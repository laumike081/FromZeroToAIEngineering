## Problem Statement

There are `n` cities. Some of them are connected, while some are not. If city `a` is connected directly with city `b`, and city `b` is connected directly with city `c`, then city `a` is connected indirectly with city `c`.

A **province** is a group of directly or indirectly connected cities and no other cities outside of the group.

You are given an `n x n` matrix `isConnected` where `isConnected[i][j] = 1` if the `i`th city and the `j`th city are directly connected, and `isConnected[i][j] = 0` otherwise.

Return the total number of **provinces**.

### Examples

**Example 1:**
```
Input: isConnected = [[1,1,0],[1,1,0],[0,0,1]]
Output: 2
Explanation:
  City 0 -- City 1    City 2
  (Province 1)        (Province 2)
```

**Example 2:**
```
Input: isConnected = [[1,0,0],[0,1,0],[0,0,1]]
Output: 3
Explanation:
  City 0    City 1    City 2
  (Prov 1)  (Prov 2)  (Prov 3)
  All cities are isolated.
```

### Constraints

- `1 <= n <= 200`
- `n == isConnected.length`
- `n == isConnected[i].length`
- `isConnected[i][j]` is `1` or `0`
- `isConnected[i][i] == 1`
- `isConnected[i][j] == isConnected[j][i]`

2. **Province**: A connected component in the graph

## Approach 1: Depth-First Search (DFS)

### Key Idea

Use DFS to explore each connected component.

### Algorithm

1. Initialize `visited` set and `provinces` counter
2. For each city:
   - If not visited, start DFS from that city
   - DFS marks all cities in the same province as visited
   - Increment province counter
3. Return province count

### DFS Helper

- Mark current city as visited
- For each neighbor (where `isConnected[city][neighbor] == 1`):
  - If not visited, recursively DFS

### Complexity

- **Time**: O(n²) - check all cells in adjacency matrix
- **Space**: O(n) - visited set + recursion stack

### Why This Works

DFS explores all reachable cities from a starting point, marking an entire province.

````python
def findCircleNum_dfs(isConnected):
    """
    DFS approach to find connected components.
    Time: O(n²)
    Space: O(n)
    """
    n = len(isConnected)
    visited = set()
    provinces = 0
    
    def dfs(city):
        """Explore all cities in the same province."""
        visited.add(city)
        
        # Visit all connected neighbors
        for neighbor in range(n):
            if isConnected[city][neighbor] == 1 and neighbor not in visited:
                dfs(neighbor)
    
    # Try starting DFS from each city
    for city in range(n):
        if city not in visited:
            dfs(city)  # Explore entire province
            provinces += 1  # Found a new province
    
    return provinces

# Test
print("Approach 1: DFS\n")
isConnected1 = [[1,1,0],[1,1,0],[0,0,1]]
print(f"Input: {isConnected1}")
print(f"Output: {findCircleNum_dfs(isConnected1)}\n")

isConnected2 = [[1,0,0],[0,1,0],[0,0,1]]
print(f"Input: {isConnected2}")
print(f"Output: {findCircleNum_dfs(isConnected2)}")
````

## Detailed DFS Trace

````python
def findCircleNum_dfs_verbose(isConnected):
    """
    DFS with detailed trace.
    """
    n = len(isConnected)
    visited = set()
    provinces = 0
    
    print(f"Adjacency Matrix ({n}x{n}):")
    for i, row in enumerate(isConnected):
        print(f"  City {i}: {row}")
    print()
    print("="*80)
    
    def dfs(city, depth=0):
        """DFS with trace."""
        indent = "  " * depth
        print(f"{indent}DFS(city={city})")
        visited.add(city)
        print(f"{indent}  Marked city {city} as visited")
        print(f"{indent}  Visited set: {sorted(visited)}")
        
        # Find neighbors
        neighbors = [i for i in range(n) if isConnected[city][i] == 1 and i not in visited]
        print(f"{indent}  Unvisited neighbors: {neighbors}")
        
        for neighbor in neighbors:
            print(f"{indent}  → Exploring neighbor {neighbor}")
            dfs(neighbor, depth + 1)
        
        print(f"{indent}  ← Backtracking from city {city}")
    
    # Main loop
    for city in range(n):
        if city not in visited:
            print(f"\nStarting new province from city {city}")
            print("-" * 80)
            dfs(city)
            provinces += 1
            print(f"\nProvince {provinces} found: {sorted([c for c in visited if c not in 
                  (visited - {city} if provinces == 1 else set())])}")
            print("="*80)
    
    print(f"\nTotal provinces: {provinces}")
    return provinces

# Trace
print("DFS Detailed Trace\n")
findCircleNum_dfs_verbose([[1,1,0],[1,1,0],[0,0,1]])
````

## Approach 2: Breadth-First Search (BFS)

### Key Idea

Use BFS to explore each connected component level by level.

### Algorithm

1. Initialize `visited` set and `provinces` counter
2. For each city:
   - If not visited, start BFS from that city
   - BFS uses a queue to explore all cities in the same province
   - Increment province counter
3. Return province count

### BFS Process

- Add starting city to queue
- While queue not empty:
  - Dequeue a city
  - For each unvisited neighbor, mark visited and enqueue

### Complexity

- **Time**: O(n²) - check all cells in adjacency matrix
- **Space**: O(n) - visited set + queue

### Advantage

No recursion stack, explores level by level.

````python
from collections import deque

def findCircleNum_bfs(isConnected):
    """
    BFS approach to find connected components.
    Time: O(n²)
    Space: O(n)
    """
    n = len(isConnected)
    visited = set()
    provinces = 0
    
    def bfs(start_city):
        """Explore all cities in the same province using BFS."""
        queue = deque([start_city])
        visited.add(start_city)
        
        while queue:
            city = queue.popleft()
            
            # Check all neighbors
            for neighbor in range(n):
                if isConnected[city][neighbor] == 1 and neighbor not in visited:
                    visited.add(neighbor)
                    queue.append(neighbor)
    
    # Try starting BFS from each city
    for city in range(n):
        if city not in visited:
            bfs(city)  # Explore entire province
            provinces += 1  # Found a new province
    
    return provinces

# Test
print("Approach 2: BFS\n")
isConnected1 = [[1,1,0],[1,1,0],[0,0,1]]
print(f"Input: {isConnected1}")
print(f"Output: {findCircleNum_bfs(isConnected1)}\n")

isConnected2 = [[1,0,0],[0,1,0],[0,0,1]]
print(f"Input: {isConnected2}")
print(f"Output: {findCircleNum_bfs(isConnected2)}")
````

## Detailed BFS Trace

````python
def findCircleNum_bfs_verbose(isConnected):
    """
    BFS with detailed trace.
    """
    n = len(isConnected)
    visited = set()
    provinces = 0
    
    print(f"Adjacency Matrix ({n}x{n}):")
    for i, row in enumerate(isConnected):
        print(f"  City {i}: {row}")
    print()
    print("="*80)
    
    def bfs(start_city):
        """BFS with trace."""
        queue = deque([start_city])
        visited.add(start_city)
        
        print(f"  Initial queue: {list(queue)}")
        print(f"  Visited: {sorted(visited)}\n")
        
        step = 0
        while queue:
            step += 1
            city = queue.popleft()
            
            print(f"  Step {step}: Dequeue city {city}")
            print(f"    Queue: {list(queue)}")
            
            # Find unvisited neighbors
            neighbors = []
            for neighbor in range(n):
                if isConnected[city][neighbor] == 1 and neighbor not in visited:
                    neighbors.append(neighbor)
                    visited.add(neighbor)
                    queue.append(neighbor)
            
            if neighbors:
                print(f"    Found neighbors: {neighbors}")
                print(f"    Added to queue: {neighbors}")
            else:
                print(f"    No unvisited neighbors")
            
            print(f"    Queue now: {list(queue)}")
            print(f"    Visited: {sorted(visited)}\n")
    
    # Main loop
    for city in range(n):
        if city not in visited:
            print(f"\nStarting new province from city {city}")
            print("-" * 80)
            bfs(city)
            provinces += 1
            print(f"Province {provinces} completed")
            print("="*80)
    
    print(f"\nTotal provinces: {provinces}")
    return provinces

# Trace
print("BFS Detailed Trace\n")
findCircleNum_bfs_verbose([[1,1,0],[1,1,0],[0,0,1]])
````

## Approach 3: Union-Find (Disjoint Set Union)

### Key Idea

Use Union-Find data structure to group connected cities.

### Algorithm

1. Initialize each city as its own parent
2. For each connection `isConnected[i][j] == 1`:
   - Union cities i and j
3. Count unique roots (provinces)

### Union-Find Operations

- **Find**: Find root of a city (with path compression)
- **Union**: Merge two sets (with union by rank)

### Complexity

- **Time**: O(n² × α(n)) ≈ O(n²) - α is inverse Ackermann (nearly constant)
- **Space**: O(n) - parent and rank arrays

### Advantage

Efficient for dynamic connectivity problems.

````python
def findCircleNum_unionfind(isConnected):
    """
    Union-Find approach.
    Time: O(n² × α(n)) ≈ O(n²)
    Space: O(n)
    """
    n = len(isConnected)
    parent = list(range(n))  # Each city is its own parent initially
    rank = [0] * n  # Rank for union by rank optimization
    
    def find(x):
        """Find root with path compression."""
        if parent[x] != x:
            parent[x] = find(parent[x])  # Path compression
        return parent[x]
    
    def union(x, y):
        """Union two sets with union by rank."""
        root_x = find(x)
        root_y = find(y)
        
        if root_x != root_y:
            # Union by rank
            if rank[root_x] < rank[root_y]:
                parent[root_x] = root_y
            elif rank[root_x] > rank[root_y]:
                parent[root_y] = root_x
            else:
                parent[root_y] = root_x
                rank[root_x] += 1
    
    # Union all connected cities
    for i in range(n):
        for j in range(i + 1, n):  # Only check upper triangle (symmetric matrix)
            if isConnected[i][j] == 1:
                union(i, j)
    
    # Count unique roots
    provinces = len(set(find(i) for i in range(n)))
    return provinces

# Test
print("Approach 3: Union-Find\n")
isConnected1 = [[1,1,0],[1,1,0],[0,0,1]]
print(f"Input: {isConnected1}")
print(f"Output: {findCircleNum_unionfind(isConnected1)}\n")

isConnected2 = [[1,0,0],[0,1,0],[0,0,1]]
print(f"Input: {isConnected2}")
print(f"Output: {findCircleNum_unionfind(isConnected2)}")
````

## Detailed Union-Find Trace

````python
def findCircleNum_unionfind_verbose(isConnected):
    """
    Union-Find with detailed trace.
    """
    n = len(isConnected)
    parent = list(range(n))
    rank = [0] * n
    
    print(f"Initial state:")
    print(f"  Parent: {parent} (each city is its own parent)")
    print(f"  Rank:   {rank}\n")
    print("="*80)
    
    def find(x):
        """Find with path compression and trace."""
        if parent[x] != x:
            original_parent = parent[x]
            parent[x] = find(parent[x])
            if original_parent != parent[x]:
                print(f"    Path compression: parent[{x}] = {parent[x]}")
        return parent[x]
    
    def union(x, y):
        """Union with trace."""
        print(f"\nUnion({x}, {y}):")
        root_x = find(x)
        root_y = find(y)
        
        print(f"  Root of {x}: {root_x}")
        print(f"  Root of {y}: {root_y}")
        
        if root_x != root_y:
            if rank[root_x] < rank[root_y]:
                parent[root_x] = root_y
                print(f"  Rank[{root_x}]={rank[root_x]} < Rank[{root_y}]={rank[root_y]}")
                print(f"  → Set parent[{root_x}] = {root_y}")
            elif rank[root_x] > rank[root_y]:
                parent[root_y] = root_x
                print(f"  Rank[{root_x}]={rank[root_x]} > Rank[{root_y}]={rank[root_y]}")
                print(f"  → Set parent[{root_y}] = {root_x}")
            else:
                parent[root_y] = root_x
                rank[root_x] += 1
                print(f"  Rank[{root_x}]={rank[root_x]-1} == Rank[{root_y}]={rank[root_y]}")
                print(f"  → Set parent[{root_y}] = {root_x}")
                print(f"  → Increment rank[{root_x}] to {rank[root_x]}")
            
            print(f"  Parent: {parent}")
            print(f"  Rank:   {rank}")
        else:
            print(f"  Already in same set (both have root {root_x})")
    
    # Process connections
    print("\nProcessing connections:\n")
    for i in range(n):
        for j in range(i + 1, n):
            if isConnected[i][j] == 1:
                union(i, j)
    
    print("\n" + "="*80)
    print("\nFinal state:")
    print(f"  Parent: {parent}")
    print(f"  Rank:   {rank}\n")
    
    # Find provinces
    print("Finding provinces:")
    roots = {}
    for i in range(n):
        root = find(i)
        if root not in roots:
            roots[root] = []
        roots[root].append(i)
        print(f"  City {i} → Root {root}")
    
    print(f"\nProvinces:")
    for idx, (root, cities) in enumerate(roots.items(), 1):
        print(f"  Province {idx}: {cities} (root: {root})")
    
    provinces = len(roots)
    print(f"\nTotal provinces: {provinces}")
    return provinces

# Trace
print("Union-Find Detailed Trace\n")
findCircleNum_unionfind_verbose([[1,1,0],[1,1,0],[0,0,1]])
````

## Performance Comparison

````python
import time
import random

def generate_graph(n, density=0.3):
    """
    Generate random symmetric adjacency matrix.
    density: probability of connection between cities
    """
    matrix = [[0] * n for _ in range(n)]
    
    # Set diagonal to 1
    for i in range(n):
        matrix[i][i] = 1
    
    # Add random connections (symmetric)
    for i in range(n):
        for j in range(i + 1, n):
            if random.random() < density:
                matrix[i][j] = 1
                matrix[j][i] = 1
    
    return matrix

def benchmark(isConnected):
    """Benchmark all approaches."""
    results = {}
    
    # DFS
    start = time.time()
    result = findCircleNum_dfs(isConnected)
    elapsed = time.time() - start
    results['DFS'] = (result, elapsed)
    
    # BFS
    start = time.time()
    result = findCircleNum_bfs(isConnected)
    elapsed = time.time() - start
    results['BFS'] = (result, elapsed)
    
    # Union-Find
    start = time.time()
    result = findCircleNum_unionfind(isConnected)
    elapsed = time.time() - start
    results['Union-Find'] = (result, elapsed)
    
    return results

# Benchmark
print("Performance Comparison\n")
print(f"{'Cities':<10} {'Approach':<15} {'Provinces':<12} {'Time (seconds)'}")
print("="*70)

for n in [10, 20, 50, 100, 150]:
    isConnected = generate_graph(n, density=0.3)
    results = benchmark(isConnected)
    
    for approach, (provinces, elapsed) in results.items():
        print(f"{n:<10} {approach:<15} {provinces:<12} {elapsed:.6f}")
    print()

print("Observation:")
print("- All approaches have O(n²) time complexity")
print("- DFS and BFS have similar performance")
print("- Union-Find is competitive with path compression")
print("- Choice depends on problem constraints and preferences")
````

## Comparison of Approaches

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| **DFS** | O(n²) | O(n) | Simple, intuitive | Recursion stack |
| **BFS** | O(n²) | O(n) | No recursion, level-by-level | Requires queue |
| **Union-Find** | O(n² × α(n)) | O(n) | Efficient for dynamic connectivity | More complex |

Where α(n) is the inverse Ackermann function (nearly constant).

### When to Use Each

- **DFS**: Simple implementation, recursive thinking
- **BFS**: Avoid recursion, explore level by level
- **Union-Find**: Dynamic connectivity, multiple queries

## Key Takeaways

### Problem Type

**Connected Components**: Count separate groups in an undirected graph.

### Graph Representation

- **Adjacency Matrix**: `isConnected[i][j] = 1` means edge between i and j
- **Symmetric**: `isConnected[i][j] == isConnected[j][i]`
- **Diagonal**: `isConnected[i][i] = 1` (city connected to itself)

### Core Algorithms

**1. DFS (Depth-First Search)**
```python
def dfs(city):
    visited.add(city)
    for neighbor in range(n):
        if isConnected[city][neighbor] == 1 and neighbor not in visited:
            dfs(neighbor)

provinces = 0
for city in range(n):
    if city not in visited:
        dfs(city)
        provinces += 1
```

**2. BFS (Breadth-First Search)**
```python
def bfs(start):
    queue = deque([start])
    visited.add(start)
    while queue:
        city = queue.popleft()
        for neighbor in range(n):
            if isConnected[city][neighbor] == 1 and neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
```

**3. Union-Find**
```python
def find(x):
    if parent[x] != x:
        parent[x] = find(parent[x])  # Path compression
    return parent[x]

def union(x, y):
    root_x, root_y = find(x), find(y)
    if root_x != root_y:
        parent[root_x] = root_y  # Union
```

### Key Insights

1. **Province = Connected Component**: All cities reachable from each other
2. **Visit each city once**: Mark visited to avoid cycles
3. **Count new explorations**: Each unvisited city starts a new province
4. **Symmetric matrix**: Only check upper/lower triangle

### Common Mistakes

❌ **Not marking cities as visited**
```python
# WRONG - infinite loop in cycles
def dfs(city):
    for neighbor in range(n):
        if isConnected[city][neighbor] == 1:
            dfs(neighbor)  # No visited check!
```

✅ **Always check visited**
```python
# CORRECT
def dfs(city):
    visited.add(city)
    for neighbor in range(n):
        if isConnected[city][neighbor] == 1 and neighbor not in visited:
            dfs(neighbor)
```

❌ **Counting provinces incorrectly**
```python
# WRONG - counts every city
for city in range(n):
    dfs(city)
    provinces += 1  # Wrong!
```

✅ **Only count new provinces**
```python
# CORRECT
for city in range(n):
    if city not in visited:  # New province!
        dfs(city)
        provinces += 1
```

### Edge Cases to Remember

1. **Single city**: n=1 → 1 province
2. **All isolated**: No connections → n provinces
3. **All connected**: Complete graph → 1 province
4. **Linear chain**: 0-1-2-3 → 1 province
5. **Multiple groups**: Separate components

### Complexity Analysis

**All Approaches:**
- **Time**: O(n²)
  - Must check all n² cells in adjacency matrix
  - DFS/BFS visit each city once, check all neighbors
  - Union-Find processes all edges
- **Space**: O(n)
  - DFS: visited set + recursion stack
  - BFS: visited set + queue
