## Problem Statement

Given the `root` of a binary tree, return its **maximum depth**.

A binary tree's **maximum depth** is the number of nodes along the longest path from the root node down to the farthest leaf node.

### Visual Example

```
Example 1:
        3
       / \
      9  20
        /  \
       15   7

Paths from root to leaves:
- 3 → 9           (depth: 2)
- 3 → 20 → 15     (depth: 3)
- 3 → 20 → 7      (depth: 3)

Maximum depth: 3
```

### Examples

**Example 1:**
```
Input: root = [3,9,20,null,null,15,7]
Output: 3
```

**Example 2:**
```
Input: root = [1,null,2]
Output: 2
```

**Example 3:**
```
Input: root = []
Output: 0
```

**Example 4:**
```
Input: root = [0]
Output: 1
```

### Constraints

- The number of nodes is in the range `[0, 10^4]`
- `-100 <= Node.val <= 100`

## Understanding Depth-First Search (DFS)

### What is DFS?

**Depth-First Search** is a tree/graph traversal algorithm that explores as far as possible along each branch before backtracking.

### DFS Traversal Orders

For binary trees, there are three main DFS orders:

```
Tree:     1
         / \
        2   3
       / \
      4   5

1. Preorder (Root → Left → Right):  1, 2, 4, 5, 3
2. Inorder (Left → Root → Right):   4, 2, 5, 1, 3
3. Postorder (Left → Right → Root): 4, 5, 2, 3, 1
```

### DFS for Maximum Depth

For finding maximum depth, we use **postorder** thinking:
- First, find depth of left subtree
- Then, find depth of right subtree
- Finally, return 1 + max(left_depth, right_depth)

### Why DFS Works for Depth?

```
        3          depth = 1 + max(left_depth, right_depth)
       / \                = 1 + max(1, 2)
      9  20               = 3
        /  \
       15   7

Left subtree (9):  depth = 1 (leaf node)
Right subtree (20): depth = 1 + max(1, 1) = 2
  - Left (15):  depth = 1
  - Right (7):  depth = 1
```

## Tree Node Definition

````python
class TreeNode:
    """Definition for a binary tree node."""
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(values):
    """Helper function to build tree from list representation."""
    if not values:
        return None
    
    root = TreeNode(values[0])
    queue = [root]
    i = 1
    
    while queue and i < len(values):
        node = queue.pop(0)
        
        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            queue.append(node.left)
        i += 1
        
        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            queue.append(node.right)
        i += 1
    
    return root

def print_tree(root, level=0, prefix="Root: "):
    """Helper function to visualize tree structure."""
    if root is not None:
        print(" " * (level * 4) + prefix + str(root.val))
        if root.left or root.right:
            if root.left:
                print_tree(root.left, level + 1, "L--- ")
            else:
                print(" " * ((level + 1) * 4) + "L--- None")
            if root.right:
                print_tree(root.right, level + 1, "R--- ")
            else:
                print(" " * ((level + 1) * 4) + "R--- None")
````

## Approach 1: Recursive DFS (Top-Down)

### Intuition

The most natural way to think about tree depth:
- If tree is empty, depth is 0
- Otherwise, depth is 1 (current node) + maximum depth of left and right subtrees

### Algorithm

```
maxDepth(root):
    if root is null:
        return 0
    
    left_depth = maxDepth(root.left)
    right_depth = maxDepth(root.right)
    
    return 1 + max(left_depth, right_depth)
```

### Complexity

- **Time**: O(n) - visit each node once
- **Space**: O(h) - recursion stack, where h is height
  - Best case (balanced): O(log n)
  - Worst case (skewed): O(n)

````python
def maxDepth(root):
    """
    Recursive DFS approach to find maximum depth.
    Time: O(n), Space: O(h)
    """
    # Base case: empty tree has depth 0
    if not root:
        return 0
    
    # Recursive case: 1 + max depth of subtrees
    left_depth = maxDepth(root.left)
    right_depth = maxDepth(root.right)
    
    return 1 + max(left_depth, right_depth)

# Test
test_cases = [
    [3, 9, 20, None, None, 15, 7],
    [1, None, 2],
    [],
    [0]
]

print("Recursive DFS Approach:\n")
for values in test_cases:
    root = build_tree(values)
    depth = maxDepth(root)
    print(f"Input:  {values}")
    print(f"Output: {depth}\n")
````

### One-Line Version

````python
def maxDepth_oneline(root):
    """Concise one-line recursive solution."""
    return 0 if not root else 1 + max(maxDepth_oneline(root.left), maxDepth_oneline(root.right))

# Test
root = build_tree([3, 9, 20, None, None, 15, 7])
print(f"One-line version: {maxDepth_oneline(root)}")
````

### Detailed Step-by-Step Trace

````python
def maxDepth_verbose(root, level=0, position="Root"):
    """Verbose version showing recursion tree."""
    indent = "  " * level
    
    if not root:
        print(f"{indent}{position}: None → depth = 0")
        return 0
    
    print(f"{indent}{position}: Node({root.val})")
    
    # Recurse on left
    print(f"{indent}  ↓ Going left from {root.val}")
    left_depth = maxDepth_verbose(root.left, level + 1, f"Left of {root.val}")
    
    # Recurse on right
    print(f"{indent}  ↓ Going right from {root.val}")
    right_depth = maxDepth_verbose(root.right, level + 1, f"Right of {root.val}")
    
    # Calculate depth
    depth = 1 + max(left_depth, right_depth)
    print(f"{indent}  ↑ Node({root.val}): 1 + max({left_depth}, {right_depth}) = {depth}")
    
    return depth

# Test
print("Detailed Recursion Trace:\n")
print("Tree structure:")
root = build_tree([3, 9, 20, None, None, 15, 7])
print_tree(root)
print("\nRecursion trace:")
result = maxDepth_verbose(root)
print(f"\nFinal result: {result}")
````

## Approach 2: Iterative DFS (Using Stack)

### Intuition

We can simulate recursion using an explicit stack:
- Push (node, current_depth) onto stack
- Track maximum depth seen
- Process nodes in DFS order

### Algorithm

```
maxDepth(root):
    if root is null:
        return 0
    
    stack = [(root, 1)]
    max_depth = 0
    
    while stack is not empty:
        node, depth = stack.pop()
        max_depth = max(max_depth, depth)
        
        if node.left:
            stack.push((node.left, depth + 1))
        if node.right:
            stack.push((node.right, depth + 1))
    
    return max_depth
```

### Complexity

- **Time**: O(n) - visit each node once
- **Space**: O(h) - stack size, where h is height

````python
def maxDepth_iterative(root):
    """
    Iterative DFS using explicit stack.
    Time: O(n), Space: O(h)
    """
    if not root:
        return 0
    
    stack = [(root, 1)]  # (node, depth)
    max_depth = 0
    
    while stack:
        node, depth = stack.pop()
        
        # Update max depth
        max_depth = max(max_depth, depth)
        
        # Push children with incremented depth
        if node.right:  # Push right first (so left is processed first)
            stack.append((node.right, depth + 1))
        if node.left:
            stack.append((node.left, depth + 1))
    
    return max_depth

# Test
print("Iterative DFS Approach:\n")
for values in test_cases:
    root = build_tree(values)
    depth = maxDepth_iterative(root)
    print(f"Input:  {values}")
    print(f"Output: {depth}\n")
````

### Detailed Step-by-Step Trace

````python
def maxDepth_iterative_verbose(root):
    """Verbose iterative DFS showing stack operations."""
    if not root:
        print("Empty tree, depth = 0")
        return 0
    
    stack = [(root, 1)]
    max_depth = 0
    step = 0
    
    print(f"Initial stack: [(Node({root.val}), depth=1)]\n")
    print("="*70)
    
    while stack:
        step += 1
        node, depth = stack.pop()
        
        print(f"Step {step}:")
        print(f"  Pop: Node({node.val}), depth={depth}")
        
        old_max = max_depth
        max_depth = max(max_depth, depth)
        if max_depth != old_max:
            print(f"  → Update max_depth: {old_max} → {max_depth}")
        else:
            print(f"  → max_depth stays: {max_depth}")
        
        # Push children
        if node.right:
            print(f"  → Push right child: Node({node.right.val}), depth={depth + 1}")
            stack.append((node.right, depth + 1))
        if node.left:
            print(f"  → Push left child: Node({node.left.val}), depth={depth + 1}")
            stack.append((node.left, depth + 1))
        
        # Show current stack
        if stack:
            stack_str = ", ".join([f"(Node({n.val}), d={d})" for n, d in stack])
            print(f"  Stack: [{stack_str}]")
        else:
            print(f"  Stack: []")
        print()
    
    print("="*70)
    print(f"Final max_depth: {max_depth}")
    return max_depth

# Test
print("Iterative DFS Trace:\n")
root = build_tree([3, 9, 20, None, None, 15, 7])
maxDepth_iterative_verbose(root)
````

## Approach 3: BFS (Level-Order Traversal)

### Intuition

Instead of DFS, we can use BFS to traverse level by level:
- Process all nodes at each level
- Count the number of levels
- Maximum depth = number of levels

### Algorithm

```
maxDepth(root):
    if root is null:
        return 0
    
    queue = [root]
    depth = 0
    
    while queue is not empty:
        depth += 1
        level_size = len(queue)
        
        for i in range(level_size):
            node = queue.dequeue()
            if node.left:
                queue.enqueue(node.left)
            if node.right:
                queue.enqueue(node.right)
    
    return depth
```

### Complexity

- **Time**: O(n) - visit each node once
- **Space**: O(w) - queue size, where w is maximum width
  - Worst case: O(n) for complete binary tree

````python
from collections import deque

def maxDepth_bfs(root):
    """
    BFS (level-order) approach to find maximum depth.
    Time: O(n), Space: O(w) where w is max width
    """
    if not root:
        return 0
    
    queue = deque([root])
    depth = 0
    
    while queue:
        depth += 1
        level_size = len(queue)
        
        # Process all nodes at current level
        for _ in range(level_size):
            node = queue.popleft()
            
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
    
    return depth

# Test
print("BFS Approach:\n")
for values in test_cases:
    root = build_tree(values)
    depth = maxDepth_bfs(root)
    print(f"Input:  {values}")
    print(f"Output: {depth}\n")
````

### Detailed Step-by-Step Trace

````python
def maxDepth_bfs_verbose(root):
    """Verbose BFS showing level-by-level processing."""
    if not root:
        print("Empty tree, depth = 0")
        return 0
    
    queue = deque([root])
    depth = 0
    
    print(f"Initial queue: [Node({root.val})]\n")
    print("="*70)
    
    while queue:
        depth += 1
        level_size = len(queue)
        
        print(f"Level {depth}:")
        print(f"  Nodes at this level: {level_size}")
        
        level_values = [node.val for node in queue]
        print(f"  Processing: {level_values}")
        
        next_level = []
        
        for i in range(level_size):
            node = queue.popleft()
            print(f"    - Node({node.val})", end="")
            
            children = []
            if node.left:
                queue.append(node.left)
                children.append(f"left={node.left.val}")
                next_level.append(node.left.val)
            if node.right:
                queue.append(node.right)
                children.append(f"right={node.right.val}")
                next_level.append(node.right.val)
            
            if children:
                print(f" → children: {', '.join(children)}")
            else:
                print(" → leaf node")
        
        if next_level:
            print(f"  Next level queue: {next_level}")
        else:
            print(f"  Next level queue: [] (done)")
        print()
    
    print("="*70)
    print(f"Total levels (depth): {depth}")
    return depth

# Test
print("BFS Level-Order Trace:\n")
root = build_tree([3, 9, 20, None, None, 15, 7])
maxDepth_bfs_verbose(root)
````

## Comparison of Approaches

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| **Recursive DFS** | O(n) | O(h) | Clean, intuitive | Stack overflow risk for deep trees |
| **Iterative DFS** | O(n) | O(h) | No recursion, explicit control | More code, less intuitive |
| **BFS** | O(n) | O(w) | Level-by-level, good for wide trees | More space for wide trees |

**Where:**
- n = number of nodes
- h = height of tree
- w = maximum width of tree

### When to Use Each

**Recursive DFS**:
- ✅ Most problems (simplest code)
- ✅ Interviews (shows understanding)
- ❌ Very deep trees (stack overflow)

**Iterative DFS**:
- ✅ Deep trees (no stack overflow)
- ✅ Need explicit control over traversal
- ❌ More verbose

**BFS**:
- ✅ Level-related problems
- ✅ Finding shortest path
- ❌ Wide trees (more space)

## DFS Pattern Recognition

### When to Use DFS for Trees

DFS is ideal for:
1. **Depth/Height problems** (this problem)
2. **Path problems** (root to leaf paths)
3. **Subtree problems** (validate BST, symmetric tree)
4. **Bottom-up aggregation** (diameter, balanced tree)

### DFS Template for Tree Problems

```python
def dfs(root):
    # Base case: null node
    if not root:
        return base_value
    
    # Recursive case: process subtrees
    left_result = dfs(root.left)
    right_result = dfs(root.right)
    
    # Combine results
    return combine(left_result, right_result, root.val)
```

## Key Takeaways

### Understanding DFS

1. **DFS explores depth-first**: Goes as deep as possible before backtracking

2. **Three DFS orders**:
   - Preorder: Root → Left → Right
   - Inorder: Left → Root → Right
   - Postorder: Left → Right → Root

3. **For depth problems**: Use postorder thinking (process children first)

### Maximum Depth Problem

1. **Recursive formula**: `depth = 1 + max(left_depth, right_depth)`

2. **Base case**: Empty tree has depth 0

3. **Three approaches**:
   - Recursive DFS (most intuitive)
   - Iterative DFS (explicit stack)
   - BFS (level-by-level)

### Complexity Analysis

- **Time**: O(n) for all approaches (visit each node once)
- **Space**: 
  - DFS: O(h) - recursion/stack depth
  - BFS: O(w) - queue width

### DFS vs BFS

| Aspect | DFS | BFS |
|--------|-----|-----|
| **Order** | Depth-first | Level-by-level |
| **Space** | O(h) | O(w) |
| **Use for** | Depth, paths, subtrees | Levels, shortest path |
| **Implementation** | Recursion or stack | Queue |

### Pattern Recognition

**Use DFS when**:
- Problem involves depth/height
- Need to explore all paths
- Bottom-up aggregation needed
- Subtree validation required

**Use BFS when**:
- Problem involves levels
- Need shortest path
- Level-order traversal required

### Remember

🎯 **DFS = Depth First** - go deep before wide  
🎯 **Recursive = Natural** - mirrors tree structure  
🎯 **Base case = Empty** - null node returns 0  
🎯 **Combine = 1 + max** - current + max of children  

Master this pattern and you'll solve dozens of tree problems!
