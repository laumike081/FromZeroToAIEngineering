# 1047. Remove All Adjacent Duplicates in String

**Difficulty**: Easy  
**Topics**: String, Stack  
**Link**: [LeetCode 1047](https://leetcode.com/problems/remove-all-adjacent-duplicates-in-string/)

## Problem Statement

You are given a string `s` consisting of lowercase English letters. A duplicate removal operation consists of choosing two adjacent and equal letters and removing them.

You repeatedly perform duplicate removal operations on `s` until no more operations can be performed.

Return the final string after all such duplicate removal operations have been performed.

### Examples

**Example 1:**
```
Input: s = "abbaca"
Output: "ca"
Explanation: 
Remove "bb" -> "aaca"
Remove "aa" -> "ca"
```

**Example 2:**
```
Input: s = "azxxzy"
Output: "ay"
Explanation: 
Remove "xx" -> "azzy"
Remove "zz" -> "ay"
```

### Constraints

- `1 <= s.length <= 10^5`
- `s` consists only of lowercase English letters.

---

## Approach 1: Stack

### Intuition

Use a stack to keep track of characters. When we encounter a character that matches the top of the stack, we pop it (removing the duplicate). Otherwise, we push it.

### Why it works

The stack naturally handles the removal of adjacent duplicates. When we remove a pair, it might create new adjacent pairs that need to be checked, and the stack handles this automatically.

### Algorithm Visualization

```
Input: "abbaca"

Stack: []

a: push -> [a]
b: push -> [a, b]
b: top is b, pop -> [a]
a: top is a, pop -> []
c: push -> [c]
a: push -> [c, a]

Result: "ca"
```

### Complexity
- **Time**: O(n) - one pass through the string
- **Space**: O(n) - worst case stack size

````python
def remove_duplicates_stack(s: str) -> str:
    """Stack approach: O(n) time, O(n) space"""
    stack = []
    
    for char in s:
        if stack and stack[-1] == char:
            stack.pop()  # Remove duplicate
        else:
            stack.append(char)  # Add new character
    
    return ''.join(stack)

# Test cases
test_cases = [
    ("abbaca", "ca"),
    ("azxxzy", "ay"),
    ("a", "a"),
    ("aa", ""),
    ("abccba", ""),
    ("abbaa", "a"),
    ("", ""),
    ("abcddcba", "abca")
]

for s, expected in test_cases:
    result = remove_duplicates_stack(s)
    print(f"{s!r:10} -> {result!r:5} (expected {expected!r})")
````

## Approach 2: Two Pointers (In-place Stack)

### Intuition

Use the input string as a stack by treating it as an array and using two pointers.

### Why it works

We maintain a write pointer that acts like the stack top. When we find a duplicate, we decrement the write pointer.

### Complexity
- **Time**: O(n)
- **Space**: O(1) - using the input string as stack

````python
def remove_duplicates_two_pointers(s: str) -> str:
    """Two pointers approach: O(n) time, O(1) space"""
    # Convert to list for in-place modification
    chars = list(s)
    write = 0  # Acts as stack top
    
    for read in range(len(chars)):
        if write > 0 and chars[write - 1] == chars[read]:
            write -= 1  # Remove duplicate
        else:
            chars[write] = chars[read]
            write += 1
    
    return ''.join(chars[:write])

# Test
for s, expected in test_cases:
    result = remove_duplicates_two_pointers(s)
    print(f"{s!r:10} -> {result!r:5} (expected {expected!r})")
````

## Approach 3: String Building

### Intuition

Build the result string character by character, removing duplicates as we go.

### Why it works

Similar to stack but using string operations instead of explicit stack.

### Complexity
- **Time**: O(n²) - string concatenation is O(n) in worst case
- **Space**: O(n)

````python
def remove_duplicates_string(s: str) -> str:
    """String building approach: O(n²) time, O(n) space"""
    result = []
    
    for char in s:
        if result and result[-1] == char:
            result.pop()
        else:
            result.append(char)
    
    return ''.join(result)

# Test
for s, expected in test_cases:
    result = remove_duplicates_string(s)
    print(f"{s!r:10} -> {result!r:5} (expected {expected!r})")
````

## Walkthrough (Stack Approach)

Input: `"abccba"`

| Step | Char | Stack | Action | Result |
|------|------|-------|--------|--------|
| 0 | - | [] | Start | ✓ |
| 1 | `a` | [`a`] | Push | ✓ |
| 2 | `b` | [`a`, `b`] | Push | ✓ |
| 3 | `c` | [`a`, `b`, `c`] | Push | ✓ |
| 4 | `c` | [`a`, `b`] | Pop duplicate | ✓ |
| 5 | `b` | [`a`] | Pop duplicate | ✓ |
| 6 | `a` | [] | Pop duplicate | ✓ |

**Final Result**: `""` (empty string)

````python
def remove_duplicates_verbose(s: str) -> str:
    """Verbose stack approach showing steps"""
    stack = []
    
    print(f"Processing: {s!r}")
    
    for i, char in enumerate(s):
        print(f"\nStep {i+1}: Char = {char!r}")
        print(f"  Stack before: {stack}")
        
        if stack and stack[-1] == char:
            removed = stack.pop()
            print(f"  Action: Remove duplicate {removed!r}")
        else:
            stack.append(char)
            print(f"  Action: Push {char!r}")
        
        print(f"  Stack after: {stack}")
    
    result = ''.join(stack)
    print(f"\nFinal result: {result!r}")
    return result

# Test with verbose output
remove_duplicates_verbose("abccba")
print("\n" + "="*40)
remove_duplicates_verbose("abbaca")
````

## Comparison

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| **Stack** | O(n) | O(n) | Intuitive, clear logic | Extra space |
| **Two Pointers** | O(n) | O(1) | Optimal space | More complex logic |
| **String Building** | O(n²) | O(n) | Simple concept | Inefficient for long strings |

**Best choice**: **Two Pointers** for optimal space, **Stack** for clarity.

## Edge Cases

| Case | Input | Output | Reason |
|------|-------|--------|--------|
| **Empty string** | "" | "" | Nothing to remove |
| **Single character** | "a" | "a" | No duplicate |
| **All same** | "aaaa" | "" | All pairs removed |
| **Odd same** | "aaa" | "a" | One remains |
| **No duplicates** | "abc" | "abc" | Nothing removed |
| **Cascade effect** | "abccba" | "" | Removal creates new pairs |
| **Long string** | "abccbaabccba" | "" | Multiple cascades |
| **Mixed** | "abbaca" | "ca" | Partial removal |

## Common Mistakes

1. **Missing cascade effect**
   ```python
   # WRONG: Only removes original pairs
   result = s.replace('aa', '').replace('bb', '')
   # CORRECT: Use stack to handle cascades
   ```

2. **Using string concatenation**
   ```python
   # WRONG: O(n²) time
   result = result + char
   # CORRECT: Use list and join
   stack.append(char)
   ```

3. **Not checking stack empty**
   ```python
   # WRONG: IndexError on empty stack
   if stack[-1] == char:
   # CORRECT: Check if stack exists
   if stack and stack[-1] == char:
   ```
