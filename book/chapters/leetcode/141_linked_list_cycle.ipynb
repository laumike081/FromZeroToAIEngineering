# 141. Linked List Cycle

## Problem Statement

Given `head`, the head of a linked list, determine if the linked list has a cycle in it.

There is a cycle in a linked list if there is some node in the list that can be reached again by continuously following the `next` pointer. Internally, `pos` is used to denote the index of the node that tail's `next` pointer is connected to. Note that `pos` is not passed as a parameter.

Return `true` if there is a cycle in the linked list. Otherwise, return `false`.

### Examples

**Example 1:**
```
Input: head = [3,2,0,-4], pos = 1
Output: true
Explanation: There is a cycle, where the tail connects to the 1st node (0-indexed).
```

**Example 2:**
```
Input: head = [1,2], pos = 0
Output: true
Explanation: There is a cycle, where the tail connects to the 0th node.
```

**Example 3:**
```
Input: head = [1], pos = -1
Output: false
Explanation: There is no cycle in the linked list.
```

### Constraints

- The number of nodes is in the range `[0, 10^4]`
- `-10^5 <= Node.val <= 10^5`
- `pos` is `-1` or a valid index in the linked-list

---

## Approach 1: Slow and Fast Pointers (Floyd's Algorithm)

### Intuition

Use two pointers moving at different speeds:
- **Slow pointer**: moves 1 step at a time
- **Fast pointer**: moves 2 steps at a time

If there's a cycle, fast will eventually "lap" slow.

### Why it works (Runner Analogy)

Imagine two runners on a circular track:
- Fast runner runs at 2x speed
- Slow runner runs at 1x speed
- Fast runner will eventually catch up to slow runner

Mathematically: The distance between fast and slow decreases by 1 each step.

### Complexity
- **Time**: O(n) - at most n steps before meeting
- **Space**: O(1) - constant extra space

````python
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None

def has_cycle(head: ListNode) -> bool:
    """Floyd's Tortoise and Hare: O(n) time, O(1) space"""
    slow = fast = head
    
    while fast and fast.next:
        slow = slow.next          # Move 1 step
        fast = fast.next.next     # Move 2 steps
        
        if slow == fast:
            return True  # Cycle detected
    
    return False  # No cycle

# --- Test ---
# Create cycle: 3 -> 2 -> 0 -> -4 -> 2 (cycle at index 1)
h = ListNode(3)
h.next = ListNode(2)
h.next.next = ListNode(0)
h.next.next.next = ListNode(-4)
h.next.next.next.next = h.next  # Create cycle
print(f"Has cycle: {has_cycle(h)}")  # True

# Create no cycle: 1 -> 2
h2 = ListNode(1)
h2.next = ListNode(2)
print(f"Has cycle: {has_cycle(h2)}")  # False
````

---

## Walkthrough (Floyd's Algorithm)

Input: `[3,2,0,-4]` with cycle at index 1

| Step | slow | fast | Action | Meeting? |
|------|------|------|--------|----------|
| 0 | 3 | 3 | Initialize | No |
| 1 | 2 | 0 | slow=slow.next, fast=fast.next.next | No |
| 2 | 0 | 2 | slow=slow.next, fast=fast.next.next | No |
| 3 | -4 | 0 | slow=slow.next, fast=fast.next.next | No |
| 4 | 2 | 2 | slow=slow.next, fast=fast.next.next | **Yes!** |

**Key observation**: Fast laps slow in 4 steps. Both meet at node 2.

## Approach 2: Hash Set

### Intuition

Track visited nodes using a hash set. If we encounter a node we've seen before, there's a cycle.

### Complexity
- **Time**: O(n) - visit each node once
- **Space**: O(n) - store all visited nodes

````python
def has_cycle_set(head: ListNode) -> bool:
    """Hash Set: O(n) time, O(n) space"""
    visited = set()
    curr = head
    
    while curr:
        if curr in visited:
            return True  # Cycle detected
        visited.add(curr)
        curr = curr.next
    
    return False  # No cycle

# --- Test ---
# Create cycle: 1 -> 2 -> 1
h = ListNode(1)
h.next = ListNode(2)
h.next.next = h
print(f"Has cycle (set): {has_cycle_set(h)}")  # True

# Create no cycle: 1 -> 2 -> 3
h2 = ListNode(1)
h2.next = ListNode(2)
h2.next.next = ListNode(3)
print(f"Has cycle (set): {has_cycle_set(h2)}")  # False
````

## Approach 3: Node Modification (Marking)

### Intuition

Temporarily modify node values to mark visited nodes. Use a special value that won't appear in the list.

### Complexity
- **Time**: O(n) - single traversal
- **Space**: O(1) - no extra storage

### Caveat
Modifies the list - not suitable if list must remain unchanged.

````python
def has_cycle_marking(head: ListNode) -> bool:
    """Node marking: O(n) time, O(1) space (modifies list)"""
    curr = head
    MARKER = float('inf')  # Value unlikely to appear in list
    
    while curr:
        if curr.val == MARKER:
            return True  # Cycle detected
        curr.val = MARKER  # Mark as visited
        curr = curr.next
    
    return False  # No cycle

# --- Test ---
# Create cycle: 1 -> 2 -> 1
h = ListNode(1)
h.next = ListNode(2)
h.next.next = h
print(f"Has cycle (marking): {has_cycle_marking(h)}")  # True

# Create no cycle: 1 -> 2 -> 3
h2 = ListNode(1)
h2.next = ListNode(2)
h2.next.next = ListNode(3)
print(f"Has cycle (marking): {has_cycle_marking(h2)}")  # False
````

## Comparison

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| **Floyd's Algorithm** | O(n) | O(1) | No extra space, elegant | Less intuitive initially |
| **Hash Set** | O(n) | O(n) | Simple to understand | Uses extra memory |
| **Node Marking** | O(n) | O(1) | No extra space, simple | Modifies the list |

**Best choice**: **Floyd's Algorithm** - optimal space complexity without modifying the list.

---

## Mathematical Proof (Why Floyd's Works)

Let:
- `μ` = distance from head to cycle start
- `λ` = length of the cycle
- `k` = steps until fast catches slow

After `k` steps:
- Slow has moved `k` steps
- Fast has moved `2k` steps

When they meet:
- Slow position: `μ + (k - μ) mod λ`
- Fast position: `μ + (2k - μ) mod λ`

Since they meet: `μ + (k - μ) ≡ μ + (2k - μ) (mod λ)`
Simplifies to: `k ≡ 2k (mod λ)`
Therefore: `k` is a multiple of `λ`

**Conclusion**: They always meet within the cycle!
