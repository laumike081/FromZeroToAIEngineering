## Problem Statement

You are given the `root` node of a binary search tree (BST) and a `value` to insert into the tree. Return the root node of the BST after the insertion. It is **guaranteed** that the new value does not exist in the original BST.

**Notice** that there may exist multiple valid ways for the insertion, as long as the tree remains a BST after insertion. You can return **any of them**.

### Visual Example

```
Original BST:        Insert 5:
      4                  4
     / \                / \
    2   7              2   7
   / \                / \  /
  1   3              1   3 5

BST Property: For each node:
- All values in left subtree < node.val
- All values in right subtree > node.val
```

### Examples

**Example 1:**
```
Input: root = [4,2,7,1,3], val = 5
Output: [4,2,7,1,3,5]
Explanation: Another accepted tree is:
      5
     / \
    2   7
   / \  
  1   3
       \
        4
```

**Example 2:**
```
Input: root = [40,20,60,10,30,50,70], val = 25
Output: [40,20,60,10,30,50,70,null,null,25]
```

**Example 3:**
```
Input: root = [4,2,7,1,3,null,null,null,null,null,null], val = 5
Output: [4,2,7,1,3,5]
```

### Constraints

- The number of nodes in the tree will be in the range `[0, 10^4]`
- `-10^8 <= Node.val <= 10^8`
- All the values `Node.val` are **unique**
- `-10^8 <= val <= 10^8`
- It's **guaranteed** that `val` does not exist in the original BST

## Understanding Binary Search Trees (BST)

### What is a BST?

A **Binary Search Tree** is a binary tree with a special property:
- **Left subtree**: All values < node.val
- **Right subtree**: All values > node.val
- This property holds for **every node** in the tree

### BST Property Visualization

```
        50
       /  \
     30    70
    / \   / \
  20  40 60  80

For node 50:
  Left subtree: {20, 30, 40} all < 50 ✓
  Right subtree: {60, 70, 80} all > 50 ✓

For node 30:
  Left subtree: {20} < 30 ✓
  Right subtree: {40} > 30 ✓
```

### Why BST is Useful

- **Search**: O(h) - compare and go left/right
- **Insert**: O(h) - find position and add
- **Delete**: O(h) - find and remove
- **Inorder traversal**: Gives sorted order

Where h = height (best: log n, worst: n)

### Insertion Strategy

To insert a value into a BST:
1. Start at root
2. Compare value with current node
3. If value < node.val → go left
4. If value > node.val → go right
5. When you reach null → insert new node there

```
Insert 25 into:     Step 1:        Step 2:        Step 3:
      40              40             40             40
     /  \            /  \           /  \           /  \
   20    60        20    60       20    60       20    60
  / \   / \       / \   / \      / \   / \      / \   / \
10  30 50  70   10  30 50  70  10  30 50  70  10  30 50  70
                                   ↓                  /
                              25 < 40?           25 < 30?
                              Yes, go left       Yes, go left
                                                 
Final:                                          Insert here!
      40
     /  \

  25
```

## Tree Node Definition

````python
class TreeNode:
    """Definition for a binary tree node."""
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(values):
    """Helper function to build tree from list representation."""
    if not values:
        return None
    
    root = TreeNode(values[0])
    queue = [root]
    i = 1
    
    while queue and i < len(values):
        node = queue.pop(0)
        
        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            queue.append(node.left)
        i += 1
        
        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            queue.append(node.right)
        i += 1
    
    return root

def print_tree(root, level=0, prefix="Root: "):
    """Helper function to visualize tree structure."""
    if root is not None:
        print(" " * (level * 4) + prefix + str(root.val))
        if root.left or root.right:
            if root.left:
                print_tree(root.left, level + 1, "L--- ")
            else:
                print(" " * ((level + 1) * 4) + "L--- None")
            if root.right:
                print_tree(root.right, level + 1, "R--- ")
            else:
                print(" " * ((level + 1) * 4) + "R--- None")

def tree_to_list(root):
    """Convert tree to list representation (level-order)."""
    if not root:
        return []
    
    result = []
    queue = [root]
    
    while queue:
        node = queue.pop(0)
        if node:
            result.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            result.append(None)
    
    # Remove trailing None values
    while result and result[-1] is None:
        result.pop()
    
    return result
````

## Approach 1: Recursive Insertion

### Intuition

The most natural way to insert into a BST is recursively:
1. **Base case**: If current node is null, create and return new node
2. **Recursive case**: Compare value with current node
   - If val < node.val → insert into left subtree
   - If val > node.val → insert into right subtree
3. Return the current node (unchanged)

### Algorithm

```
insertIntoBST(root, val):
    if root is null:
        return new TreeNode(val)
    
    if val < root.val:
        root.left = insertIntoBST(root.left, val)
    else:
        root.right = insertIntoBST(root.right, val)
    
    return root
```

### Complexity

- **Time**: O(h) where h is height
  - Best case (balanced): O(log n)
  - Worst case (skewed): O(n)
- **Space**: O(h) for recursion stack

````python
def insertIntoBST(root, val):
    """
    Recursive insertion into BST.
    Time: O(h), Space: O(h)
    """
    # Base case: found insertion point
    if not root:
        return TreeNode(val)
    
    # Recursive case: navigate to insertion point
    if val < root.val:
        root.left = insertIntoBST(root.left, val)
    else:  # val > root.val (guaranteed unique)
        root.right = insertIntoBST(root.right, val)
    
    return root

# Test
test_cases = [
    ([4, 2, 7, 1, 3], 5),
    ([40, 20, 60, 10, 30, 50, 70], 25),
    ([4, 2, 7, 1, 3], 5),
    ([], 5)
]

print("Recursive Insertion:\n")
for tree_vals, insert_val in test_cases:
    root = build_tree(tree_vals)
    print(f"Original tree: {tree_vals}")
    print(f"Insert value: {insert_val}")
    
    result_root = insertIntoBST(root, insert_val)
    result = tree_to_list(result_root)
    
    print(f"Result: {result}")
    print("Tree structure:")
    print_tree(result_root)
    print()
````

### Detailed Step-by-Step Trace

````python
def insertIntoBST_verbose(root, val, level=0, position="Root"):
    """Verbose recursive insertion showing decision path."""
    indent = "  " * level
    
    if not root:
        print(f"{indent}{position}: NULL → Insert {val} here!")
        return TreeNode(val)
    
    print(f"{indent}{position}: Node({root.val})")
    print(f"{indent}  Compare: {val} vs {root.val}")
    
    if val < root.val:
        print(f"{indent}  → {val} < {root.val}, go LEFT")
        root.left = insertIntoBST_verbose(root.left, val, level + 1, f"Left of {root.val}")
    else:
        print(f"{indent}  → {val} > {root.val}, go RIGHT")
        root.right = insertIntoBST_verbose(root.right, val, level + 1, f"Right of {root.val}")
    
    print(f"{indent}  ↑ Return Node({root.val})")
    return root

# Test
print("Detailed Recursive Insertion Trace:\n")
print("Insert 5 into [4, 2, 7, 1, 3]:\n")
print("Original tree:")
root = build_tree([4, 2, 7, 1, 3])
print_tree(root)
print("\nInsertion trace:")
print("="*70)
result = insertIntoBST_verbose(root, 5)
print("="*70)
print("\nFinal tree:")
print_tree(result)
````

## Approach 2: Iterative Insertion

### Intuition

We can avoid recursion by using iteration:
1. Handle empty tree case
2. Traverse to find parent of insertion point
3. Create new node and attach to parent

### Algorithm

```
insertIntoBST(root, val):
    if root is null:
        return new TreeNode(val)
    
    current = root
    while true:
        if val < current.val:
            if current.left is null:
                current.left = new TreeNode(val)
                break
            current = current.left
        else:
            if current.right is null:
                current.right = new TreeNode(val)
                break
            current = current.right
    
    return root
```

### Complexity

- **Time**: O(h) where h is height
- **Space**: O(1) - no recursion stack

````python
def insertIntoBST_iterative(root, val):
    """
    Iterative insertion into BST.
    Time: O(h), Space: O(1)
    """
    # Handle empty tree
    if not root:
        return TreeNode(val)
    
    # Find insertion point
    current = root
    while True:
        if val < current.val:
            if not current.left:
                current.left = TreeNode(val)
                break
            current = current.left
        else:  # val > current.val
            if not current.right:
                current.right = TreeNode(val)
                break
            current = current.right
    
    return root

# Test
print("Iterative Insertion:\n")
for tree_vals, insert_val in test_cases:
    root = build_tree(tree_vals)
    print(f"Original tree: {tree_vals}")
    print(f"Insert value: {insert_val}")
    
    result_root = insertIntoBST_iterative(root, insert_val)
    result = tree_to_list(result_root)
    
    print(f"Result: {result}\n")
````

### Detailed Step-by-Step Trace

````python
def insertIntoBST_iterative_verbose(root, val):
    """Verbose iterative insertion showing traversal path."""
    if not root:
        print("Empty tree, create new root")
        return TreeNode(val)
    
    print(f"Inserting {val} into BST\n")
    print("="*70)
    
    current = root
    step = 0
    path = []
    
    while True:
        step += 1
        path.append(current.val)
        
        print(f"Step {step}: At Node({current.val})")
        print(f"  Path so far: {' → '.join(map(str, path))}")
        print(f"  Compare: {val} vs {current.val}")
        
        if val < current.val:
            print(f"  → {val} < {current.val}, go LEFT")
            if not current.left:
                print(f"  → Left child is NULL, insert {val} here!")
                current.left = TreeNode(val)
                path.append(val)
                break
            else:
                print(f"  → Left child exists (Node {current.left.val}), continue")
                current = current.left
        else:
            print(f"  → {val} > {current.val}, go RIGHT")
            if not current.right:
                print(f"  → Right child is NULL, insert {val} here!")
                current.right = TreeNode(val)
                path.append(val)
                break
            else:
                print(f"  → Right child exists (Node {current.right.val}), continue")
                current = current.right
        print()
    
    print("="*70)
    print(f"Final path: {' → '.join(map(str, path))}")
    print(f"Inserted {val} successfully!")
    
    return root

# Test
print("Detailed Iterative Insertion Trace:\n")
root = build_tree([4, 2, 7, 1, 3])
print("Original tree:")
print_tree(root)
print()
result = insertIntoBST_iterative_verbose(root, 5)
print("\nFinal tree:")
print_tree(result)
````

## Approach 3: Parent Tracking (Alternative Iterative)

### Intuition

Another iterative approach that explicitly tracks the parent:
1. Find the parent node where insertion should happen
2. Determine if new node should be left or right child
3. Attach new node to parent

### Complexity

- **Time**: O(h)
- **Space**: O(1)

````python
def insertIntoBST_parent_tracking(root, val):
    """
    Iterative insertion with explicit parent tracking.
    Time: O(h), Space: O(1)
    """
    new_node = TreeNode(val)
    
    if not root:
        return new_node
    
    # Find parent of insertion point
    current = root
    parent = None
    
    while current:
        parent = current
        if val < current.val:
            current = current.left
        else:
            current = current.right
    
    # Attach new node to parent
    if val < parent.val:
        parent.left = new_node
    else:
        parent.right = new_node
    
    return root

# Test
print("Parent Tracking Insertion:\n")
root = build_tree([4, 2, 7, 1, 3])
print("Original tree:")
print_tree(root)
print(f"\nInsert: 5")
result = insertIntoBST_parent_tracking(root, 5)
print("\nResult tree:")
print_tree(result)
````

## Comparison of Approaches

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| **Recursive** | O(h) | O(h) | Clean, intuitive | Stack overflow for deep trees |
| **Iterative** | O(h) | O(1) | No recursion, space-efficient | More verbose |
| **Parent Tracking** | O(h) | O(1) | Explicit parent, clear logic | Slightly more code |

**Where h = height:**
- Balanced BST: h = O(log n)
- Skewed BST: h = O(n)

### When to Use Each

**Recursive**:
- ✅ Most problems (cleanest code)
- ✅ Interviews (shows understanding)
- ❌ Very deep trees

**Iterative**:
- ✅ Deep trees (no stack overflow)
- ✅ Space-critical applications
- ✅ Production code

**Parent Tracking**:
- ✅ When you need parent reference
- ✅ Educational purposes

## Edge Cases and Validation

````python
def is_valid_bst(root, min_val=float('-inf'), max_val=float('inf')):
    """Validate if tree is a valid BST."""
    if not root:
        return True
    
    if root.val <= min_val or root.val >= max_val:
        return False
    
    return (is_valid_bst(root.left, min_val, root.val) and
            is_valid_bst(root.right, root.val, max_val))

def inorder_traversal(root):
    """Get inorder traversal (should be sorted for BST)."""
    if not root:
        return []
    return inorder_traversal(root.left) + [root.val] + inorder_traversal(root.right)

print("Edge Cases and Validation:\n")

# Edge case 1: Empty tree
print("1. Insert into empty tree:")
root1 = insertIntoBST(None, 5)
print(f"   Result: {tree_to_list(root1)}")
print(f"   Valid BST: {is_valid_bst(root1)}")
print(f"   Inorder: {inorder_traversal(root1)}\n")

# Edge case 2: Single node tree
print("2. Insert into single node tree:")
root2 = build_tree([5])
root2 = insertIntoBST(root2, 3)
root2 = insertIntoBST(root2, 7)
print(f"   Result: {tree_to_list(root2)}")
print(f"   Valid BST: {is_valid_bst(root2)}")
print(f"   Inorder: {inorder_traversal(root2)}\n")

# Edge case 3: Left-skewed tree
print("3. Insert into left-skewed tree:")
root3 = build_tree([5, 3, None, 1])
print("   Before:")
print_tree(root3)
root3 = insertIntoBST(root3, 2)
print("   After inserting 2:")
print_tree(root3)
print(f"   Valid BST: {is_valid_bst(root3)}")
print(f"   Inorder: {inorder_traversal(root3)}\n")

# Edge case 4: Right-skewed tree
print("4. Insert into right-skewed tree:")
root4 = build_tree([1, None, 3, None, 5])
print("   Before:")
print_tree(root4)
root4 = insertIntoBST(root4, 4)
print("   After inserting 4:")
print_tree(root4)
print(f"   Valid BST: {is_valid_bst(root4)}")
print(f"   Inorder: {inorder_traversal(root4)}\n")

# Edge case 5: Multiple insertions
print("5. Multiple insertions:")
root5 = build_tree([50])
values_to_insert = [30, 70, 20, 40, 60, 80]
print(f"   Starting with: {tree_to_list(root5)}")
print(f"   Inserting: {values_to_insert}")
for val in values_to_insert:
    root5 = insertIntoBST(root5, val)
print("   Final tree:")
print_tree(root5)
print(f"   Valid BST: {is_valid_bst(root5)}")
print(f"   Inorder (should be sorted): {inorder_traversal(root5)}")
````

## BST Operations Pattern

### Common BST Operations

All BST operations follow similar patterns:

#### 1. Search
```python
def search(root, val):
    if not root or root.val == val:
        return root
    if val < root.val:
        return search(root.left, val)
    return search(root.right, val)
```

#### 2. Insert (This Problem)
```python
def insert(root, val):
    if not root:
        return TreeNode(val)
    if val < root.val:
        root.left = insert(root.left, val)
    else:
        root.right = insert(root.right, val)
    return root
```

#### 3. Delete
```python
def delete(root, val):
    if not root:
        return None
    if val < root.val:
        root.left = delete(root.left, val)
    elif val > root.val:
        root.right = delete(root.right, val)
    else:
        # Found node to delete
        if not root.left:
            return root.right
        if not root.right:
            return root.left
        # Node has two children: find successor
        # ... (more complex)
    return root
```

### The BST Pattern

**Template:**
```python
def bst_operation(root, val):
    # Base case
    if not root:
        return base_value
    
    # Compare and recurse
    if val < root.val:
        root.left = bst_operation(root.left, val)
    elif val > root.val:
        root.right = bst_operation(root.right, val)
    else:
        # Found the node
        handle_found_case()
    
    return root
```

## Key Takeaways

### Understanding BST

1. **BST Property**: Left < Node < Right (for every node)

2. **Why BST is useful**:
   - Search: O(h)
   - Insert: O(h)
   - Delete: O(h)
   - Inorder gives sorted sequence

3. **Height matters**:
   - Balanced: h = O(log n)
   - Skewed: h = O(n)

### Insertion Strategy

1. **Compare and navigate**:
   - val < node.val → go left
   - val > node.val → go right

2. **Insert at null position**:
   - When you reach null, that's where new node goes

3. **Multiple valid solutions**:
   - Any insertion that maintains BST property is valid

### Three Approaches

1. **Recursive** (Most common):
   - Clean, intuitive
   - O(h) space for recursion
   - Base case: null → create node
   - Recursive: navigate and return

2. **Iterative** (Space-efficient):
   - O(1) space
   - No stack overflow
   - Navigate until null, then insert

3. **Parent Tracking**:
   - Explicit parent reference
   - Find parent, then attach

### Complexity Analysis

- **Time**: O(h) for all approaches
  - Best: O(log n) - balanced tree
  - Worst: O(n) - skewed tree

- **Space**:
  - Recursive: O(h) - recursion stack
  - Iterative: O(1) - no extra space

### BST Pattern Recognition

**The BST template:**
```python
if not root:
    return base_case
if val < root.val:
    root.left = recurse(root.left)
else:
    root.right = recurse(root.right)
return root
```

**Use this pattern for**:
- Search operations
- Insert operations
- Delete operations
- Finding min/max
- Finding successor/predecessor

### Validation

**Always verify BST property**:
- Inorder traversal should be sorted
- Each node: min_val < node.val < max_val
