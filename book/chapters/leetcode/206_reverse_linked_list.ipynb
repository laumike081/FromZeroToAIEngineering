# 206. Reverse Linked List

**Difficulty**: Easy  
**Topics**: Linked List, Recursion  
**Link**: [LeetCode 206](https://leetcode.com/problems/reverse-linked-list/)

---

## Problem Statement

Given the `head` of a singly linked list, reverse the list, and return *the reversed list*.

### Examples

**Example 1:**
```
Input: head = [1,2,3,4,5]
Output: [5,4,3,2,1]
```

**Example 2:**
```
Input: head = [1,2]
Output: [2,1]
```

**Example 3:**
```
Input: head = []
Output: []
```

### Constraints

- The number of nodes in the list is in the range `[0, 5000]`
- `-5000 <= Node.val <= 5000`

````python
# Definition for singly-linked list
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    
    def __repr__(self):
        result = []
        node = self
        while node:
            result.append(str(node.val))
            node = node.next
        return " -> ".join(result)

def create_list(values):
    """Helper to create linked list from array"""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head
````

## Approach 1: Iterative (Three Pointers)

### Intuition

Use three pointers to reverse links one by one:
- **prev**: previously processed node (starts as None)
- **curr**: current node being processed
- **next**: next node to process (saved before reversing)

### Why it works

The key insight is to **save the next node** before we overwrite `curr.next`. Without this, we'd lose access to the rest of the list.

### Algorithm Visualization

```
Initial: 1 -> 2 -> 3 -> 4 -> 5 -> None
         prev=None, curr=1

Step 1: Save next=2, reverse 1->None, move pointers
         None <- 1    2 -> 3 -> 4 -> 5
         prev=1, curr=2

Step 2: Save next=3, reverse 2->1, move pointers
         None <- 1 <- 2    3 -> 4 -> 5
         prev=2, curr=3

Step 3: Save next=4, reverse 3->2, move pointers
         None <- 1 <- 2 <- 3    4 -> 5
         prev=3, curr=4

Step 4: Save next=5, reverse 4->3, move pointers
         None <- 1 <- 2 <- 3 <- 4    5
         prev=4, curr=5

Step 5: Save next=None, reverse 5->4, move pointers
         None <- 1 <- 2 <- 3 <- 4 <- 5
         prev=5, curr=None
```

### Complexity
- **Time**: O(n) - visit each node exactly once
- **Space**: O(1) - only three pointers used

````python
def reverse_list_iterative(head: ListNode) -> ListNode:
    """Iterative: O(n) time, O(1) space"""
    prev = None
    curr = head
    
    while curr:
        next_temp = curr.next  # Save next node
        curr.next = prev       # Reverse the link
        prev = curr            # Move prev forward
        curr = next_temp       # Move curr forward
    
    return prev  # prev is now the new head

# Test
head = create_list([1, 2, 3, 4, 5])
print(f"Original: {head}")
reversed_head = reverse_list_iterative(head)
print(f"Reversed: {reversed_head}")
````

## Approach 2: Recursive

### Intuition

1. **Base case**: Empty list or single node - already reversed
2. **Recursive case**: Reverse the rest of the list, then attach current node

### Why it works

The recursion reverses everything after the current node. When we return:
- `new_head` points to the last node (new head of reversed list)
- `head.next` is the node that should now point back to `head`
- We set `head.next = None` to break the original forward link

### Complexity
- **Time**: O(n) - each node visited once
- **Space**: O(n) - recursion stack depth equals list length

````python
def reverse_list_recursive(head: ListNode) -> ListNode:
    """Recursive: O(n) time, O(n) space"""
    # Base case: empty list or single node
    if not head or not head.next:
        return head
    
    # Recursively reverse the rest
    new_head = reverse_list_recursive(head.next)
    
    # Make next node point back to current
    head.next.next = head
    head.next = None
    
    return new_head

# Test
head = create_list([1, 2, 3, 4, 5])
print(f"Original: {head}")
reversed_head = reverse_list_recursive(head)
print(f"Reversed: {reversed_head}")
````

## Approach 3: Stack

### Intuition

Push all nodes onto a stack, then pop them off to build the reversed list.

### Why it works

Stacks are LIFO (Last In, First Out). When we push nodes in order 1→2→3, we get them back as 3→2→1.

### Complexity
- **Time**: O(n) - two passes (push and pop)
- **Space**: O(n) - stack stores all nodes

````python
def reverse_list_stack(head: ListNode) -> ListNode:
    """Stack: O(n) time, O(n) space"""
    if not head:
        return None
    
    # Push all nodes onto stack
    stack = []
    curr = head
    while curr:
        stack.append(curr)
        curr = curr.next
    
    # Pop nodes to build reversed list
    new_head = stack.pop()
    curr = new_head
    
    while stack:
        curr.next = stack.pop()
        curr = curr.next
    
    curr.next = None  # Terminate the list
    return new_head

# Test
head = create_list([1, 2, 3, 4, 5])
print(f"Original: {head}")
reversed_head = reverse_list_stack(head)
print(f"Reversed: {reversed_head}")
````

## Approach 4: Two Pointers (Simplified)

### Intuition

Same as iterative but with just two pointers, using a temporary variable.

### Complexity
- **Time**: O(n)
- **Space**: O(1)

````python
def reverse_list_two_pointers(head: ListNode) -> ListNode:
    """Two pointers: O(n) time, O(1) space"""
    prev = None
    while head:
        head.next, prev, head = prev, head, head.next
    return prev

# Test
head = create_list([1, 2, 3, 4, 5])
print(f"Original: {head}")
reversed_head = reverse_list_two_pointers(head)
print(f"Reversed: {reversed_head}")
````

## Detailed Walkthrough (Iterative)

Let's trace the iterative approach with `[1,2,3]`:

| Step | prev | curr | curr.next | Action | List State |
|------|------|-------|-----------|--------|------------|
| 0 | None | 1 | 2 | Initialize | 1→2→3→None |
| 1 | None | 1 | 2 | next=2, 1.next=None, prev=1, curr=2 | None←1 2→3 |
| 2 | 1 | 2 | 3 | next=3, 2.next=1, prev=2, curr=3 | None←1←2 3 |
| 3 | 2 | 3 | None | next=None, 3.next=2, prev=3, curr=None | None←1←2←3 |

**Result**: Return `prev` which is node 3, the new head.

````python
def reverse_list_verbose(head: ListNode) -> ListNode:
    """Verbose version showing each step"""
    prev = None
    curr = head
    step = 0
    
    print(f"Initial: prev={prev}, curr={curr.val if curr else None}")
    
    while curr:
        print(f"\nStep {step}:")
        print(f"  Before: prev={prev.val if prev else None}, curr={curr.val}, curr.next={curr.next.val if curr.next else None}")
        
        next_temp = curr.next
        curr.next = prev
        prev = curr
        curr = next_temp
        
        print(f"  After:  prev={prev.val if prev else None}, curr={curr.val if curr else None}")
        step += 1
    
    print(f"\nDone: new head = {prev.val if prev else None}")
    return prev

head = create_list([1, 2, 3])
reverse_list_verbose(head)
````

## Comparison

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| **Iterative (3 pointers)** | O(n) | O(1) | Optimal space, clear logic | 3 variables to track |
| **Recursive** | O(n) | O(n) | Elegant, concise | Stack overflow risk |
| **Stack** | O(n) | O(n) | Intuitive concept | Extra space overhead |
| **Two pointers** | O(n) | O(1) | Most concise | Python tuple assignment trick |

**Best choice**: **Iterative with 3 pointers** for clarity and optimal space.

## Edge Cases

| Case | Input | Output | Reason |
|------|-------|--------|--------|
| **Empty list** | [] | [] | Loop never runs, returns None |
| **Single node** | [1] | [1] | Loop runs once, returns same node |
| **Two nodes** | [1,2] | [2,1] | One reversal operation |
| **Large list** | [1,...,5000] | [5000,...,1] | Handles max constraints |
| **Negative values** | [-1,-2,-3] | [-3,-2,-1] | Values don't affect logic |

## Common Mistakes

1. **Losing reference to next node**
   ```python
   # WRONG: Overwrites curr.next before saving
   curr.next = prev  # Now we can't access curr.next!
   curr = curr.next  # This is None now!
   ```

2. **Returning wrong node**
   ```python
   # WRONG: Returns None (curr is None at end)
   return curr
   # CORRECT: Return prev (new head)
   return prev
   ```

3. **Not handling empty list**
   ```python
   # WRONG: Will try to access None.next
   while head.next:  # Fails for empty list
   # CORRECT: Check head itself
   while head:
   ```

## Key Takeaways

| Pattern | When to Use |
|---------|-------------|
| **Save before overwrite** | Always save `next` before changing `curr.next` |
| **Three pointers** | Standard in-place reversal technique |
| **Iterative vs Recursive** | Iterative for O(1) space, recursive for elegance |
| **Edge cases** | Handle empty and single-node lists |
| **Return prev** | At the end, `prev` is the new head, not `curr` |

**Remember**: The magic is in saving the next node before you overwrite the pointer!
