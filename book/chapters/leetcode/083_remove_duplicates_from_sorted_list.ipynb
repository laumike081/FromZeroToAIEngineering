# 83. Remove Duplicates from Sorted List

**Difficulty**: Easy  
**Topics**: Linked List  
**Link**: [LeetCode 83](https://leetcode.com/problems/remove-duplicates-from-sorted-list/)

## Problem Statement

Given the `head` of a sorted linked list, delete all duplicates such that each element appears only once. Return the linked list sorted as well.

### Examples

**Example 1:**
```
Input: head = [1,1,2]
Output: [1,2]
```

**Example 2:**
```
Input: head = [1,1,2,3,3]
Output: [1,2,3]
```

### Constraints

- The number of nodes in the list is in the range `[0, 300]`
- `-100 <= Node.val <= 100`
- The list is guaranteed to be **sorted** in non-decreasing order

---

## Approach 1: Iterative In-place

### Intuition

Since the list is sorted, duplicates will be adjacent. We can:
1. Traverse the list
2. When we find `curr.val == curr.next.val`, skip the duplicate
3. Otherwise, move to next node

### Why it works

**Sorted property guarantees**: All duplicates are consecutive
- If `curr.val == curr.next.val`, we know `curr.next` is a duplicate
- We can safely skip it: `curr.next = curr.next.next`
- No need to check further - all duplicates of this value are right here

### Complexity
- **Time**: O(n) - single traversal
- **Space**: O(1) - in-place modification

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def delete_duplicates(head: ListNode) -> ListNode:
    """Iterative in-place: O(n) time, O(1) space"""
    curr = head
    
    while curr and curr.next:
        if curr.val == curr.next.val:
            # Skip duplicate
            curr.next = curr.next.next
        else:
            # Move to next unique value
            curr = curr.next
    
    return head

# --- Test ---
# Create list: 1 -> 1 -> 2
h = ListNode(1, ListNode(1, ListNode(2)))
result = delete_duplicates(h)
output = []
while result:
    output.append(result.val)
    result = result.next
print(f"Test 1: {output}")  # [1, 2]

# Create list: 1 -> 1 -> 2 -> 3 -> 3
h2 = ListNode(1, ListNode(1, ListNode(2, ListNode(3, ListNode(3)))))
result2 = delete_duplicates(h2)
output2 = []
while result2:
    output2.append(result2.val)
    result2 = result2.next
print(f"Test 2: {output2}")  # [1, 2, 3]
````

---

## Walkthrough

Input: `[1,1,2,3,3]`

| Step | curr.val | curr.next.val | Action | List after step |
|------|----------|---------------|--------|----------------|
| 0 | 1 | 1 | Duplicate found, skip | 1→2→3→3 |
| 1 | 1 | 2 | Different, move curr | 1→2→3→3 |
| 2 | 2 | 3 | Different, move curr | 1→2→3→3 |
| 3 | 3 | 3 | Duplicate found, skip | 1→2→3 |
| 4 | 3 | None | End of list | 1→2→3 |

**Result**: `[1,2,3]` - all duplicates removed.

## Approach 2: Recursive

### Intuition

Recursively process the list:
1. Base case: empty list or single node
2. If current equals next, skip next and recurse
3. Otherwise, recurse on rest

### Complexity
- **Time**: O(n) - each node visited once
- **Space**: O(n) - recursion stack

````python
def delete_duplicates_recursive(head: ListNode) -> ListNode:
    """Recursive: O(n) time, O(n) space"""
    if not head or not head.next:
        return head
    
    if head.val == head.next.val:
        # Skip duplicate and continue
        return delete_duplicates_recursive(head.next)
    else:
        # Keep current and process rest
        head.next = delete_duplicates_recursive(head.next)
        return head

# --- Test ---
h = ListNode(1, ListNode(1, ListNode(2, ListNode(2, ListNode(3)))))
result = delete_duplicates_recursive(h)
output = []
while result:
    output.append(result.val)
    result = result.next
print(f"Recursive: {output}")  # [1, 2, 3]
````

## Approach 3: Two Pointers

### Intuition

Use two pointers:
- **write**: points to last unique node
- **read**: scans ahead for next unique

### Complexity
- **Time**: O(n) - single traversal
- **Space**: O(1) - constant extra space

````python
def delete_duplicates_two_pointers(head: ListNode) -> ListNode:
    """Two pointers: O(n) time, O(1) space"""
    if not head:
        return None
    
    write = head  # Last unique node
    read = head.next  # Scanner
    
    while read:
        if write.val != read.val:
            # Found new unique value
            write.next = read
            write = read
        read = read.next
    
    # Terminate list after last unique
    write.next = None
    return head

# --- Test ---
h = ListNode(1, ListNode(1, ListNode(2, ListNode(3, ListNode(3)))))
result = delete_duplicates_two_pointers(h)
output = []
while result:
    output.append(result.val)
    result = result.next
print(f"Two pointers: {output}")  # [1, 2, 3]
````

## Approach 4: Hash Set (Unsorted Generalization)

### Intuition

Use a hash set to track seen values. Works for unsorted lists too.

### Complexity
- **Time**: O(n) - single traversal
- **Space**: O(n) - store unique values

````python
def delete_duplicates_hashset(head: ListNode) -> ListNode:
    """Hash set: O(n) time, O(n) space (works for unsorted)"""
    seen = set()
    dummy = ListNode(0, head)
    prev = dummy
    curr = head
    
    while curr:
        if curr.val in seen:
            # Skip duplicate
            prev.next = curr.next
        else:
            seen.add(curr.val)
            prev = curr
        curr = curr.next
    
    return dummy.next

# --- Test ---
h = ListNode(1, ListNode(1, ListNode(2, ListNode(1, ListNode(2)))))
result = delete_duplicates_hashset(h)
output = []
while result:
    output.append(result.val)
    result = result.next
print(f"Hash set: {output}")  # [1, 2]
````

## Comparison

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| **Iterative** | O(n) | O(1) | Simple, optimal space | Only works for sorted |
| **Recursive** | O(n) | O(n) | Elegant, concise | Stack overflow risk |
| **Two Pointers** | O(n) | O(1) | Clear separation of concerns | Slightly more complex |
| **Hash Set** | O(n) | O(n) | Works for unsorted | Uses extra space |

**Best choice**: **Iterative** for sorted lists (optimal), **Hash Set** for unsorted.

## Key Takeaways

| Pattern | When to Use |
|---------|-------------|
| **Skip duplicates** | Sorted lists with adjacent duplicates |
| **Hash set** | Unsorted lists or when order doesn't matter |
| **Two pointers** | When you need to "write" to a specific position |
| **Recursive** | When elegance is preferred over space |

**Remember**: For sorted lists, duplicates are always adjacent - this is the key insight!
