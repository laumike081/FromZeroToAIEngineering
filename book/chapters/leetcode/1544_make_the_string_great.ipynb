# 1544. Make The String Great

**Difficulty**: Easy  
**Topics**: String, Stack  
**Link**: [LeetCode 1544](https://leetcode.com/problems/make-the-string-great/)

---

## Problem Statement

Given a string `s` consisting only of uppercase and lowercase letters, you can make the string great by repeatedly removing pairs of adjacent letters where one letter is the lowercase version of the other.

Specifically, you can repeatedly find two adjacent letters `s[i]` and `s[i+1]` such that:
- `s[i]` is lowercase and `s[i+1]` is the same letter in uppercase, OR
- `s[i]` is uppercase and `s[i+1]` is the same letter in lowercase

And remove both letters from the string. Continue this process until no more such pairs exist.

Return the final string after making it great.

### Examples

**Example 1:**
```
Input: s = "leEeetcode"
Output: "leetcode"
Explanation: Remove 'eE' -> "leetcode", then remove 'eE' again -> "leetcode"
```

**Example 2:**
```
Input: s = "abBAcC"
Output: ""
Explanation: Remove 'bB' -> "aAcC", then remove 'aA' -> "cC", then remove 'cC' -> ""
```

**Example 3:**
```
Input: s = "a"
Output: "a"
Explanation: No pairs to remove.
```

### Constraints

- `1 <= s.length <= 100`
- `s` consists only of lowercase and uppercase English letters.

## Approach 1: Stack

### Intuition

Use a stack to build the result. When we encounter a character, check if it forms a great pair with the top of the stack.

### Why it works

The stack naturally handles the removal process. When we remove a pair, it might create new adjacent pairs that need to be checked, and the stack handles this automatically.

### Algorithm Visualization

```
s = "leEeetcode"

Stack: []

l: push -> [l]
e: push -> [l, e]
E: top is e, remove pair -> [l]
e: push -> [l, e]
E: top is e, remove pair -> [l]
e: push -> [l, e]
e: top is e, not opposite case, push -> [l, e, e]
t: push -> [l, e, e, t]
c: push -> [l, e, e, t, c]
o: push -> [l, e, e, t, c, o]
d: push -> [l, e, e, t, c, o, d]
e: push -> [l, e, e, t, c, o, d, e]

Result: "leetcodE" -> "leetcode"
```

### Complexity
- **Time**: O(n) - one pass through the string
- **Space**: O(n) - worst case stack size

````python
def is_opposite_case(char1: str, char2: str) -> bool:
    """Check if two characters are same letter in opposite cases using ASCII difference"""
    return abs(ord(char1) - ord(char2)) == 32

def make_great_stack(s: str) -> str:
    """Stack approach: O(n) time, O(n) space"""
    stack = []
    
    for char in s:
        if stack and is_opposite_case(stack[-1], char):
            stack.pop()  # Remove the pair
        else:
            stack.append(char)
    
    return ''.join(stack)

# Test cases
test_cases = [
    ("leEeetcode", "leetcode"),
    ("abBAcC", ""),
    ("a", "a"),
    ("", ""),
    ("abAB", "abAB"),  # No opposite case pairs
    ("aabAAB", "aabAAB"),  # No opposite case pairs
    ("BbAa", ""),  # Remove 'bB' then 'aA'
    ("cC", ""),  # Single pair
    ("Ccc", "C"),  # Remove 'cC'
    ("RLrl", ""),  # Remove 'rL' then 'Rl'
]

for s, expected in test_cases:
    result = make_great_stack(s)
    print(f"{s!r:12} -> {result!r:12} (expected {expected!r})")
````

## Approach 2: Two Pointers (In-place)

### Intuition

Use the input string as a stack by maintaining a write pointer.

### Why it works

We can treat the string as a character array and overwrite it with the result, avoiding extra space.

### Complexity
- **Time**: O(n)
- **Space**: O(1) - using the input string as storage

````python
def make_great_two_pointers(s: str) -> str:
    """Two pointers approach: O(n) time, O(1) space"""
    chars = list(s)
    write = 0  # Acts as stack top
    
    for read in range(len(chars)):
        if write > 0 and is_opposite_case(chars[write - 1], chars[read]):
            write -= 1  # Remove the pair
        else:
            chars[write] = chars[read]
            write += 1
    
    return ''.join(chars[:write])

# Test
for s, expected in test_cases:
    result = make_great_two_pointers(s)
    print(f"{s!r:12} -> {result!r:12} (expected {expected!r})")
````

## Approach 3: ASCII Optimization

### Intuition

Use ASCII values to check opposite case more efficiently.

### Why it works

In ASCII, the difference between uppercase and lowercase letters is always 32:
- 'a' (97) vs 'A' (65) = 32
- 'b' (98) vs 'B' (66) = 32
- etc.

### Complexity
- **Time**: O(n)
- **Space**: O(n) - stack storage

````python
def make_great_ascii(s: str) -> str:
    """ASCII optimization approach: O(n) time, O(n) space"""
    stack = []
    
    for char in s:
        if stack and abs(ord(stack[-1]) - ord(char)) == 32:
            stack.pop()  # Remove the pair
        else:
            stack.append(char)
    
    return ''.join(stack)

# Test
for s, expected in test_cases:
    result = make_great_ascii(s)
    print(f"{s!r:12} -> {result!r:12} (expected {expected!r})")
````

## Walkthrough (Stack Approach)

Input: `"abBAcC"`

| Step | Char | Stack | Action | Result |
|------|------|-------|--------|--------|
| 0 | - | [] | Start | ✓ |
| 1 | `a` | [`a`] | Push | ✓ |
| 2 | `b` | [`a`, `b`] | Push | ✓ |
| 3 | `B` | [`a`] | Pop pair `bB` | ✓ |
| 4 | `A` | [] | Pop pair `aA` | ✓ |
| 5 | `c` | [`c`] | Push | ✓ |
| 6 | `C` | [] | Pop pair `cC` | ✓ |

**Final Result**: `""` (empty string)

````python
def make_great_verbose(s: str) -> str:
    """Verbose stack approach showing steps"""
    stack = []
    
    print(f"Processing: {s!r}")
    
    for i, char in enumerate(s):
        print(f"\nStep {i+1}: Char = {char!r}")
        print(f"  Stack before: {stack}")
        
        if stack and is_opposite_case(stack[-1], char):
            removed = stack.pop()
            print(f"  Action: Remove pair {removed!r}{char!r}")
        else:
            stack.append(char)
            print(f"  Action: Push {char!r}")
        
        print(f"  Stack after: {stack}")
    
    result = ''.join(stack)
    print(f"\nFinal result: {result!r}")
    return result

# Test with verbose output
make_great_verbose("abBAcC")
print("\n" + "="*40)
make_great_verbose("leEeetcode")
````

## Comparison

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| **Stack** | O(n) | O(n) | Intuitive, clear logic | Extra space |
| **Two Pointers** | O(n) | O(1) | Optimal space | More complex logic |
| **ASCII Optimization** | O(n) | O(n) | Fast case checking | Still needs stack |

**Best choice**: **Two Pointers** for optimal space, **ASCII Optimization** for speed.

## Common Mistakes

1. **Wrong case comparison**
   ```python
   # WRONG: Only checks if same
   if char1.lower() == char2.lower():
   # CORRECT: Check if same but different case
   if abs(ord(char1) - ord(char2)) == 32:
   ```

2. **Not checking stack empty**
   ```python
   # WRONG: IndexError on empty stack
   if is_opposite_case(stack[-1], char):
   # CORRECT: Check if stack exists
   if stack and is_opposite_case(stack[-1], char):
   ```

3. **Using string concatenation**
   ```python
   # WRONG: O(n²) time
   result = result + char
   # CORRECT: Use list and join
   stack.append(char)
   ```

## Key Takeaways

| Pattern | When to Use |
|---------|-------------|
| **Stack for adjacent removal** | When removal creates new adjacents |
| **ASCII difference of 32** | Fast uppercase/lowercase checking |
| **Cascade effect** | When operations affect neighbors |
| **LIFO property** | When recent operations affect current |
| **In-place optimization** | When space optimization needed |

**Remember**: Stack naturally handles the cascade effect of adjacent character removal!
