## Problem Statement

You are climbing a staircase. It takes `n` steps to reach the top.

Each time you can either climb **1 or 2 steps**. In how many distinct ways can you climb to the top?

### Visual Example

For `n = 3`:

```
Way 1: 1 + 1 + 1 = 3 steps
┌─┐
│3│
├─┤
│2│
├─┤
│1│
└─┘

Way 2: 1 + 2 = 3 steps
┌─┐
│3│
├─┤
│2│  ← Jump 2 steps
│ │
└─┘

Way 3: 2 + 1 = 3 steps
┌─┐
│3│
├─┤
│2│
│ │  ← Jump 2 steps
└─┘

Total: 3 distinct ways
```

### Examples

**Example 1:**
```
Input: n = 2
Output: 2
Explanation: There are two ways to climb to the top.
1. 1 step + 1 step
2. 2 steps
```

**Example 2:**
```
Input: n = 3
Output: 3
Explanation: There are three ways to climb to the top.
1. 1 step + 1 step + 1 step
2. 1 step + 2 steps
3. 2 steps + 1 step
```

### Constraints

- `1 <= n <= 45`

## Key Insight: It's Fibonacci!

### The Pattern

To reach step `n`, you can come from:
- Step `n-1` (take 1 step)
- Step `n-2` (take 2 steps)

Therefore:
```
ways(n) = ways(n-1) + ways(n-2)
```

This is the **Fibonacci sequence**!

### Building Intuition

```
n = 1: 1 way  (1)
n = 2: 2 ways (1+1, 2)
n = 3: 3 ways (ways(2) + ways(1) = 2 + 1 = 3)
n = 4: 5 ways (ways(3) + ways(2) = 3 + 2 = 5)
n = 5: 8 ways (ways(4) + ways(3) = 5 + 3 = 8)
...
```

Pattern: `1, 2, 3, 5, 8, 13, 21, ...` (Fibonacci starting from 1, 2)

## Approach 1: Recursion (Brute Force)

### Intuition

Direct translation of the recurrence relation:
- Base cases: `n=1` returns 1, `n=2` returns 2
- Recursive case: `climbStairs(n) = climbStairs(n-1) + climbStairs(n-2)`

### Algorithm

```
function climbStairs(n):
    if n == 1: return 1
    if n == 2: return 2
    return climbStairs(n-1) + climbStairs(n-2)
```

### Complexity

- **Time**: O(2^n) - exponential due to repeated calculations
- **Space**: O(n) - recursion stack depth

**Problem**: This is extremely slow for large n due to redundant calculations.

````python
def climbStairs_recursive(n):
    """
    Recursive solution (brute force).
    Time: O(2^n), Space: O(n)
    """
    if n == 1:
        return 1
    if n == 2:
        return 2
    return climbStairs_recursive(n - 1) + climbStairs_recursive(n - 2)

# Test (only small values due to exponential time)
test_cases = [1, 2, 3, 4, 5, 6]

print("Recursive Approach:\n")
for n in test_cases:
    result = climbStairs_recursive(n)
    print(f"n = {n}: {result} ways")
````

### Recursion Tree for n=5

```
                    climbStairs(5)
                   /              \
          climbStairs(4)        climbStairs(3)
           /         \            /         \
    climb(3)      climb(2)   climb(2)    climb(1)
     /    \          |          |            |
climb(2) climb(1)   [2]        [2]          [1]
   |        |
  [2]      [1]
```

Notice: `climbStairs(3)` and `climbStairs(2)` are calculated multiple times!

## Approach 2: Recursion with Memoization (Top-Down DP)

### Intuition

Cache the results of subproblems to avoid redundant calculations.

### Algorithm

```
memo = {}
function climbStairs(n):
    if n in memo: return memo[n]
    if n == 1: return 1
    if n == 2: return 2
    memo[n] = climbStairs(n-1) + climbStairs(n-2)
    return memo[n]
```

### Complexity

- **Time**: O(n) - each subproblem calculated once
- **Space**: O(n) - memo dictionary + recursion stack

````python
def climbStairs_memoization(n, memo=None):
    """
    Recursive solution with memoization.
    Time: O(n), Space: O(n)
    """
    if memo is None:
        memo = {}
    
    if n in memo:
        return memo[n]
    
    if n == 1:
        return 1
    if n == 2:
        return 2
    
    memo[n] = climbStairs_memoization(n - 1, memo) + climbStairs_memoization(n - 2, memo)
    return memo[n]

# Test
test_cases = [1, 2, 3, 4, 5, 10, 20, 30]

print("Memoization Approach:\n")
for n in test_cases:
    result = climbStairs_memoization(n)
    print(f"n = {n}: {result} ways")
````

### Verbose Memoization

Let's see how memoization avoids redundant calculations:

````python
def climbStairs_memo_verbose(n, memo=None, depth=0):
    """Verbose version showing memoization in action."""
    if memo is None:
        memo = {}
    
    indent = "  " * depth
    
    if n in memo:
        print(f"{indent}climbStairs({n}) → {memo[n]} (cached)")
        return memo[n]
    
    print(f"{indent}climbStairs({n}) - computing...")
    
    if n == 1:
        print(f"{indent}  → 1 (base case)")
        return 1
    if n == 2:
        print(f"{indent}  → 2 (base case)")
        return 2
    
    result = climbStairs_memo_verbose(n - 1, memo, depth + 1) + \
             climbStairs_memo_verbose(n - 2, memo, depth + 1)
    
    memo[n] = result
    print(f"{indent}  → {result} (computed and cached)")
    return result

# Test
print("Memoization trace for n=6:\n")
result = climbStairs_memo_verbose(6)
print(f"\nFinal result: {result}")
````

## Approach 3: Dynamic Programming (Bottom-Up)

### Intuition

Build the solution iteratively from the bottom up, starting from the base cases.

### Algorithm

```
1. Create dp array of size n+1
2. dp[1] = 1, dp[2] = 2 (base cases)
3. For i from 3 to n:
     dp[i] = dp[i-1] + dp[i-2]
4. Return dp[n]
```

### Complexity

- **Time**: O(n) - single loop
- **Space**: O(n) - dp array

````python
def climbStairs_dp(n):
    """
    Dynamic programming (bottom-up) solution.
    Time: O(n), Space: O(n)
    """
    if n == 1:
        return 1
    if n == 2:
        return 2
    
    dp = [0] * (n + 1)
    dp[1] = 1
    dp[2] = 2
    
    for i in range(3, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    
    return dp[n]

# Test
print("Dynamic Programming Approach:\n")
for n in test_cases:
    result = climbStairs_dp(n)
    print(f"n = {n}: {result} ways")
````

### Step-by-Step DP Table

For `n = 6`:

```
i   | 1 | 2 | 3 | 4 | 5 | 6 |
----+---+---+---+---+---+---+
dp  | 1 | 2 | 3 | 5 | 8 | 13|
```

- `dp[3] = dp[2] + dp[1] = 2 + 1 = 3`
- `dp[4] = dp[3] + dp[2] = 3 + 2 = 5`
- `dp[5] = dp[4] + dp[3] = 5 + 3 = 8`
- `dp[6] = dp[5] + dp[4] = 8 + 5 = 13`

````python
def climbStairs_dp_verbose(n):
    """Verbose version showing DP table construction."""
    if n == 1:
        return 1
    if n == 2:
        return 2
    
    dp = [0] * (n + 1)
    dp[1] = 1
    dp[2] = 2
    
    print(f"Building DP table for n={n}:\n")
    print(f"Initial: dp[1]={dp[1]}, dp[2]={dp[2]}\n")
    
    for i in range(3, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
        print(f"Step {i-2}: dp[{i}] = dp[{i-1}] + dp[{i-2}] = {dp[i-1]} + {dp[i-2]} = {dp[i]}")
    
    print(f"\nDP table: {dp[1:]}")
    print(f"Final answer: {dp[n]}")
    return dp[n]

# Test
climbStairs_dp_verbose(8)
````

## Approach 4: Space-Optimized DP (Optimal)

### Intuition

We only need the last two values to compute the next one, so we don't need to store the entire array.

### Algorithm

```
1. Keep two variables: prev1, prev2
2. For each step, compute: current = prev1 + prev2
3. Update: prev2 = prev1, prev1 = current
4. Return prev1
```

### Complexity

- **Time**: O(n) - single loop
- **Space**: O(1) - only two variables

````python
def climbStairs(n):
    """
    Space-optimized DP solution (optimal).
    Time: O(n), Space: O(1)
    """
    if n == 1:
        return 1
    if n == 2:
        return 2
    
    prev2 = 1  # ways(1)
    prev1 = 2  # ways(2)
    
    for i in range(3, n + 1):
        current = prev1 + prev2
        prev2 = prev1
        prev1 = current
    
    return prev1

# Test
print("Space-Optimized DP (Optimal):\n")
for n in test_cases:
    result = climbStairs(n)
    print(f"n = {n}: {result} ways")
````

### Visualization of Space-Optimized Approach

For `n = 6`:

```
Initial: prev2=1, prev1=2

i=3: current = 2 + 1 = 3
     prev2=2, prev1=3

i=4: current = 3 + 2 = 5
     prev2=3, prev1=5

i=5: current = 5 + 3 = 8
     prev2=5, prev1=8

i=6: current = 8 + 5 = 13
     prev2=8, prev1=13

Return: 13
```

````python
def climbStairs_verbose(n):
    """Verbose version showing variable updates."""
    if n == 1:
        return 1
    if n == 2:
        return 2
    
    prev2 = 1
    prev1 = 2
    
    print(f"Computing ways to climb {n} stairs:\n")
    print(f"Initial: prev2={prev2} (ways for n=1), prev1={prev1} (ways for n=2)\n")
    
    for i in range(3, n + 1):
        current = prev1 + prev2
        print(f"Step {i}: current = prev1 + prev2 = {prev1} + {prev2} = {current}")
        prev2 = prev1
        prev1 = current
        print(f"         Update: prev2={prev2}, prev1={prev1}\n")
    
    print(f"Final answer: {prev1}")
    return prev1

# Test
climbStairs_verbose(7)
````

## Comparison of Approaches

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| Recursion | O(2^n) | O(n) | Simple, intuitive | Extremely slow |
| Memoization | O(n) | O(n) | Fast, easy to implement | Uses extra space |
| DP (Bottom-Up) | O(n) | O(n) | Fast, iterative | Uses array |
| Space-Optimized DP | O(n) | O(1) | Fast, minimal space | Best solution |

### Performance Comparison

````python
import time

n = 35

# Test memoization
start = time.time()
result1 = climbStairs_memoization(n)
time1 = time.time() - start

# Test DP
start = time.time()
result2 = climbStairs_dp(n)
time2 = time.time() - start

# Test space-optimized
start = time.time()
result3 = climbStairs(n)
time3 = time.time() - start

print(f"For n = {n}:\n")
print(f"Memoization:        {time1:.6f} seconds, result = {result1}")
print(f"DP (Bottom-Up):     {time2:.6f} seconds, result = {result2}")
print(f"Space-Optimized DP: {time3:.6f} seconds, result = {result3}")
print(f"\nAll results match: {result1 == result2 == result3}")
````

## Common Mistakes

### Mistake 1: Wrong Base Cases

**Wrong**:
```python
if n == 0: return 0  # ❌ Wrong!
if n == 1: return 1
```

**Right**:
```python
if n == 1: return 1  # ✅ One way to climb 1 stair
if n == 2: return 2  # ✅ Two ways to climb 2 stairs
```

### Mistake 2: Off-by-One in DP Array

**Wrong**:
```python
dp = [0] * n  # ❌ Array too small
dp[1] = 1
dp[2] = 2
```

**Right**:
```python
dp = [0] * (n + 1)  # ✅ Need n+1 size to access dp[n]
dp[1] = 1
dp[2] = 2
```

### Mistake 3: Not Handling n=1 and n=2

**Wrong**:
```python
# Starting loop from i=1 without handling base cases
for i in range(1, n + 1):
    dp[i] = dp[i-1] + dp[i-2]  # ❌ Fails for i=1,2
```

**Right**:
```python
if n == 1: return 1
if n == 2: return 2
# Then start loop from i=3
for i in range(3, n + 1):
    dp[i] = dp[i-1] + dp[i-2]  # ✅
```

### Mistake 4: Forgetting to Update Variables

**Wrong**:
```python
for i in range(3, n + 1):
    current = prev1 + prev2
    # ❌ Forgot to update prev1 and prev2
```

**Right**:
```python
for i in range(3, n + 1):
    current = prev1 + prev2
    prev2 = prev1  # ✅ Update
    prev1 = current  # ✅ Update
```

## Why This is Fibonacci

### The Connection

The climbing stairs problem follows the Fibonacci sequence because:

1. **Recurrence relation**: `f(n) = f(n-1) + f(n-2)`
2. **Starting values**: `f(1) = 1, f(2) = 2`

This is Fibonacci with different initial values!

### Standard Fibonacci vs Climbing Stairs

```
Standard Fibonacci: 0, 1, 1, 2, 3, 5, 8, 13, 21, ...
Climbing Stairs:       1, 2, 3, 5, 8, 13, 21, ...
                       ↑  ↑
                      n=1 n=2
```

Climbing stairs is essentially Fibonacci shifted by one position!

### Why Does This Make Sense?

To reach step `n`, you must have come from either:
- Step `n-1` (all ways to reach n-1, then take 1 step)
- Step `n-2` (all ways to reach n-2, then take 2 steps)

These are **mutually exclusive** events, so we **add** them:
```
ways(n) = ways(n-1) + ways(n-2)
```

This is the fundamental property of Fibonacci!

## Key Takeaways

1. **Recognize Fibonacci pattern**: `f(n) = f(n-1) + f(n-2)`

2. **Multiple solution approaches**:
   - Recursion (exponential, too slow)
   - Memoization (linear time, linear space)
   - DP bottom-up (linear time, linear space)
   - Space-optimized DP (linear time, constant space)

3. **Space optimization**: Only need last two values, not entire array

4. **Base cases matter**: `n=1` returns 1, `n=2` returns 2

5. **Recurrence relation**: To reach step n, sum ways from n-1 and n-2

6. **Memoization vs DP**:
   - Memoization: Top-down, recursive
   - DP: Bottom-up, iterative

7. **Time complexity**: All efficient solutions are O(n)

8. **Space complexity**: Can be optimized from O(n) to O(1)

### The Pattern

**Dynamic Programming with Fibonacci-like recurrence**:
- Identify the recurrence relation
- Start with base cases
- Build solution iteratively or with memoization
- Optimize space by keeping only necessary values

This pattern appears in many DP problems and is fundamental to understanding dynamic programming!
