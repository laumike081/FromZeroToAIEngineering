## Problem Statement

Given an array `arr`, replace every element in that array with the greatest element among the elements to its right, and replace the last element with `-1`.

After doing so, return the array.

### Visual Example

```
Input:  [17, 18, 5, 4, 6, 1]

For each element, find max to the right:
arr[0] = 17 → max(18, 5, 4, 6, 1) = 18
arr[1] = 18 → max(5, 4, 6, 1) = 6
arr[2] = 5  → max(4, 6, 1) = 6
arr[3] = 4  → max(6, 1) = 6
arr[4] = 6  → max(1) = 1
arr[5] = 1  → -1 (last element)

Output: [18, 6, 6, 6, 1, -1]
```

### Examples

**Example 1:**
```
Input: arr = [17,18,5,4,6,1]
Output: [18,6,6,6,1,-1]
Explanation: 
- index 0 --> the greatest element to the right of index 0 is index 1 (18).
- index 1 --> the greatest element to the right of index 1 is index 4 (6).
- index 2 --> the greatest element to the right of index 2 is index 4 (6).
- index 3 --> the greatest element to the right of index 3 is index 4 (6).
- index 4 --> the greatest element to the right of index 4 is index 5 (1).
- index 5 --> there are no elements to the right of index 5, so we put -1.
```

**Example 2:**
```
Input: arr = [400]
Output: [-1]
Explanation: There are no elements to the right of index 0.
```

### Constraints

- `1 <= arr.length <= 10^4`
- `1 <= arr[i] <= 10^5`

## Key Insight: Work Backwards!

### The Naive Approach Problem

If we go left-to-right, we need to scan right for each element:
```
For arr[0]: scan arr[1:] to find max → O(n)
For arr[1]: scan arr[2:] to find max → O(n-1)
For arr[2]: scan arr[3:] to find max → O(n-2)
...
Total: O(n²)
```

### The Optimal Insight

**Work backwards (right-to-left)!**
- Keep track of the maximum seen so far
- Each element becomes the max we've seen
- Update max as we go
- Time: O(n), Space: O(1)

```
Original: [17, 18, 5, 4, 6, 1]
          
Start from right, max = -1:
i=5: arr[5]=1  → result[5]=-1,  max=1
i=4: arr[4]=6  → result[4]=1,   max=6
i=3: arr[3]=4  → result[3]=6,   max=6
i=2: arr[2]=5  → result[2]=6,   max=6
i=1: arr[1]=18 → result[1]=6,   max=18
i=0: arr[0]=17 → result[0]=18,  max=18

Result: [18, 6, 6, 6, 1, -1]
```

## Approach 1: Brute Force (Naive)

### Intuition

For each element, scan all elements to its right to find the maximum.

### Algorithm

```
For each index i from 0 to n-2:
    max_right = -infinity
    For each index j from i+1 to n-1:
        max_right = max(max_right, arr[j])
    arr[i] = max_right
arr[n-1] = -1
```

### Complexity

- **Time**: O(n²) - nested loops
- **Space**: O(1) - in-place modification

**Problem**: Too slow for large arrays!

````python
def replaceElements_bruteforce(arr):
    """
    Brute force: For each element, scan right to find max.
    Time: O(n²), Space: O(1)
    """
    n = len(arr)
    
    for i in range(n - 1):
        max_right = -1
        # Find max among all elements to the right
        for j in range(i + 1, n):
            max_right = max(max_right, arr[j])
        arr[i] = max_right
    
    # Last element is always -1
    arr[n - 1] = -1
    
    return arr

# Test
test_cases = [
    [17, 18, 5, 4, 6, 1],
    [400],
    [1, 2, 3, 4, 5]
]

print("Brute Force Approach:\n")
for arr in test_cases:
    original = arr.copy()
    result = replaceElements_bruteforce(arr.copy())
    print(f"Input:  {original}")
    print(f"Output: {result}\n")
````

## Approach 2: Backwards Iteration (Optimal)

### Intuition

Instead of looking forward for each element, traverse backwards and keep track of the maximum seen so far.

**Key Observations:**
1. The last element is always `-1`
2. For any element, the "max to the right" is the max we've seen while going backwards
3. After processing an element, it might become the new max for elements to its left

### Algorithm

```
1. Start from the last element
2. Initialize max_right = -1
3. For each element from right to left:
     a. Store current element value
     b. Replace current element with max_right
     c. Update max_right = max(max_right, stored_value)
4. Return array
```

### Complexity

- **Time**: O(n) - single pass
- **Space**: O(1) - in-place modification

````python
def replaceElements(arr):
    """
    Optimal solution: Traverse backwards, track max.
    Time: O(n), Space: O(1)
    """
    n = len(arr)
    max_right = -1
    
    # Traverse from right to left
    for i in range(n - 1, -1, -1):
        current = arr[i]        # Store current value
        arr[i] = max_right      # Replace with max to the right
        max_right = max(max_right, current)  # Update max
    
    return arr

# Test
print("Optimal Backwards Iteration:\n")
for arr in test_cases:
    original = arr.copy()
    result = replaceElements(arr.copy())
    print(f"Input:  {original}")
    print(f"Output: {result}\n")
````

### Detailed Step-by-Step Trace

````python
def replaceElements_verbose(arr):
    """Verbose version showing each step."""
    n = len(arr)
    max_right = -1
    
    print(f"Original array: {arr}")
    print(f"Length: {n}\n")
    print("="*70)
    print("Processing from RIGHT to LEFT:")
    print("="*70)
    
    for i in range(n - 1, -1, -1):
        current = arr[i]
        
        print(f"\nStep {n - i}:")
        print(f"  Index: {i}")
        print(f"  Current value: arr[{i}] = {current}")
        print(f"  Max to the right so far: {max_right}")
        
        arr[i] = max_right
        print(f"  Replace arr[{i}] with {max_right}")
        
        old_max = max_right
        max_right = max(max_right, current)
        
        if max_right != old_max:
            print(f"  Update max: {old_max} → {max_right} (current {current} is larger)")
        else:
            print(f"  Max stays: {max_right} (current {current} is not larger)")
        
        print(f"  Array now: {arr}")
    
    print(f"\n{'='*70}")
    print(f"Final result: {arr}")
    return arr

# Test
replaceElements_verbose([17, 18, 5, 4, 6, 1])
````

## Why Backwards Works

### The Key Insight

**When going backwards:**
- We've already seen all elements to the right
- We know the maximum among them
- We can replace current element immediately
- Then update max for the next iteration

**When going forwards:**
- We haven't seen elements to the right yet
- We need to scan forward to find max
- This causes O(n²) time complexity

### The Pattern: Backwards Traversal

This pattern appears in many problems:
- When you need information about elements to the right
- When you can accumulate information as you go backwards
- When forward traversal would require repeated scans

**Examples:**
- Next greater element
- Stock span problems
- Running max/min from right

## Common Mistakes

### Mistake 1: Forgetting to Store Current Value

**Wrong**:
```python
for i in range(n - 1, -1, -1):
    arr[i] = max_right  # ❌ Lost current value!
    max_right = max(max_right, arr[i])  # Using new value, not old
```

**Right**:
```python
for i in range(n - 1, -1, -1):
    current = arr[i]  # ✅ Store first
    arr[i] = max_right
    max_right = max(max_right, current)  # Use stored value
```

### Mistake 2: Wrong Initial Value for max_right

**Wrong**:
```python
max_right = 0  # ❌ What if all elements are negative?
max_right = float('-inf')  # ❌ Last element should be -1, not -inf
```

**Right**:
```python
max_right = -1  # ✅ Correct per problem requirements
```

### Mistake 3: Going Forward Instead of Backward

**Wrong**:
```python
for i in range(n):  # ❌ Forward traversal
    # Need to scan right for each element → O(n²)
    max_right = max(arr[i+1:])  # Inefficient
    arr[i] = max_right
```

**Right**:
```python
for i in range(n - 1, -1, -1):  # ✅ Backward traversal
    # Already know max from previous iterations → O(n)
```

### Mistake 4: Off-by-One in Range

**Wrong**:
```python
for i in range(n - 1, 0, -1):  # ❌ Stops at 1, misses index 0
```

**Right**:
```python
for i in range(n - 1, -1, -1):  # ✅ Goes down to 0
```

## Comparison of Approaches

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| Brute Force | O(n²) | O(1) | Simple to understand | Too slow for large inputs |
| Backwards Iteration | O(n) | O(1) | Optimal, single pass | Requires backwards thinking |

### Performance Comparison

````python
import time
import random

def compare_approaches(size=1000):
    """Compare brute force vs optimal approach."""
    random.seed(42)
    arr = [random.randint(1, 1000) for _ in range(size)]
    
    # Brute force
    arr1 = arr.copy()
    start = time.time()
    replaceElements_bruteforce(arr1)
    time_brute = time.time() - start
    
    # Optimal
    arr2 = arr.copy()
    start = time.time()
    replaceElements(arr2)
    time_optimal = time.time() - start
    
    print(f"Array size: {size}")
    print(f"Brute Force:  {time_brute:.6f} seconds")
    print(f"Optimal:      {time_optimal:.6f} seconds")
    print(f"Speedup:      {time_brute/time_optimal:.2f}x faster\n")

# Test with different sizes
for size in [100, 500, 1000, 2000]:
    compare_approaches(size)
````

## Key Takeaways

1. **Backwards traversal pattern**: When you need information about elements to the right, consider going backwards

2. **Avoid nested loops**: O(n²) brute force can often be optimized to O(n) with clever traversal

3. **Track running maximum**: Keep track of max/min as you traverse to avoid repeated scans

4. **Store before modify**: When modifying in-place, store the current value before overwriting

5. **In-place optimization**: Can achieve O(1) space by modifying the input array

6. **Edge cases matter**: Single element, ascending/descending order, all same values

7. **Time-space tradeoff**: Sometimes O(1) space requires more clever algorithms

### The Backwards Traversal Pattern

**When to use:**
- Need information about elements to the right
- Can accumulate information (max, min, sum) as you go
- Forward traversal would require repeated scans

**How it works:**
1. Start from the rightmost element
2. Maintain running state (max, min, etc.)
3. Process current element using accumulated state
4. Update state for next iteration

**Benefits:**
- Single pass: O(n) instead of O(n²)
- In-place: O(1) space
- Simple logic once you see the pattern

This pattern is fundamental for many array problems and is worth mastering!
