# 24. Swap Nodes in Pairs

**Difficulty**: Medium  
**Topics**: Linked List, Recursion  
**Link**: [LeetCode 24](https://leetcode.com/problems/swap-nodes-in-pairs/)

---

## Problem Statement

Given a linked list, swap every two adjacent nodes and return its head. You must solve the problem without modifying the values in the list's nodes (i.e., only nodes themselves may be changed.)

### Examples

**Example 1:**
```
Input: head = [1,2,3,4]
Output: [2,1,4,3]
```

**Example 2:**
```
Input: head = []
Output: []
```

**Example 3:**
```
Input: head = [1]
Output: [1]
```

### Constraints

- The number of nodes in the list is in the range `[0, 100]`
- `0 <= Node.val <= 100`

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    
    def __repr__(self):
        result = []
        node = self
        while node:
            result.append(str(node.val))
            node = node.next
        return " -> ".join(result)

def create_list(values):
    """Helper to create linked list from array"""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head
````

## Approach 1: Iterative with Dummy Head

### Intuition

Use a dummy node before head to simplify edge cases. Then iterate through pairs and swap them.

### Why it works

The dummy node ensures we always have a `prev` node to work with, even when swapping the first two nodes.

### Algorithm

```
dummy -> 1 -> 2 -> 3 -> 4 -> None
  ^      ^    ^
prev    first second

Swap: prev.next = second, first.next = second.next, second.next = first

Result: dummy -> 2 -> 1 -> 3 -> 4 -> None
```

### Complexity
- **Time**: O(n) - visit each node once
- **Space**: O(1) - constant extra space

````python
def swap_pairs_iterative(head: ListNode) -> ListNode:
    """Iterative with dummy head: O(n) time, O(1) space"""
    dummy = ListNode(0, head)
    prev = dummy
    
    while prev.next and prev.next.next:
        # Identify the pair to swap
        first = prev.next
        second = prev.next.next
        
        # Perform the swap
        prev.next = second
        first.next = second.next
        second.next = first
        
        # Move to next pair
        prev = first
    
    return dummy.next

# Test
head = create_list([1, 2, 3, 4])
print(f"Original: {head}")
swapped = swap_pairs_iterative(head)
print(f"Swapped: {swapped}")  # [2, 1, 4, 3]

# Test odd length
head2 = create_list([1, 2, 3])
print(f"\nOriginal (odd): {head2}")
swapped2 = swap_pairs_iterative(head2)
print(f"Swapped (odd): {swapped2}")  # [2, 1, 3]
````

## Approach 2: Recursive

### Intuition

Recursively swap the first two nodes, then recursively process the rest of the list.

### Why it works

Each recursive call handles one pair and returns the head of the swapped sublist.

### Complexity
- **Time**: O(n) - each node visited once
- **Space**: O(n) - recursion stack

````python
def swap_pairs_recursive(head: ListNode) -> ListNode:
    """Recursive: O(n) time, O(n) space"""
    # Base case: empty list or single node
    if not head or not head.next:
        return head
    
    # Nodes to swap
    first = head
    second = head.next
    
    # Swap and recursively process rest
    first.next = swap_pairs_recursive(second.next)
    second.next = first
    
    return second

# Test
head = create_list([1, 2, 3, 4, 5, 6])
print(f"Original: {head}")
swapped = swap_pairs_recursive(head)
print(f"Swapped: {swapped}")  # [2, 1, 4, 3, 6, 5]
````

## Approach 3: Three Pointers

### Intuition

Similar to iterative but without dummy node. Handle first pair separately.

### Complexity
- **Time**: O(n)
- **Space**: O(1)

````python
def swap_pairs_three_pointers(head: ListNode) -> ListNode:
    """Three pointers: O(n) time, O(1) space"""
    if not head or not head.next:
        return head
    
    # Handle first pair separately
    prev = None
    curr = head
    new_head = head.next
    
    while curr and curr.next:
        next_pair = curr.next.next
        second = curr.next
        
        # Swap current pair
        second.next = curr
        curr.next = next_pair
        
        # Link previous pair to current swapped pair
        if prev:
            prev.next = second
        
        # Move to next pair
        prev = curr
        curr = next_pair
    
    return new_head

# Test
head = create_list([1, 2, 3, 4])
print(f"Original: {head}")
swapped = swap_pairs_three_pointers(head)
print(f"Swapped: {swapped}")
````

## Walkthrough (Iterative with Dummy)

Input: `[1,2,3,4]`

| Step | prev | first | second | Action | List State |
|------|------|-------|--------|--------|------------|
| 0 | dummy | 1 | 2 | Initialize | dummy→1→2→3→4 |
| 1 | dummy | 1 | 2 | prev.next=2, 1.next=3, 2.next=1 | dummy→2→1→3→4 |
| 2 | 1 | 3 | 4 | prev.next=4, 3.next=None, 4.next=3 | dummy→2→1→4→3 |
| 3 | 3 | None | None | Stop | dummy→2→1→4→3 |

**Result**: Return `dummy.next` which is node 2.

---

## Comparison

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| **Iterative + Dummy** | O(n) | O(1) | Simple, handles all cases | Extra dummy node |
| **Recursive** | O(n) | O(n) | Elegant, concise | Stack overflow risk |
| **Three Pointers** | O(n) | O(1) | No dummy needed | More complex logic |

**Best choice**: **Iterative with dummy** for clarity and reliability.

## Edge Cases

| Case | Input | Output | Reason |
|------|-------|--------|--------|
| **Empty list** | [] | [] | Returns None |
| **Single node** | [1] | [1] | No pair to swap |
| **Two nodes** | [1,2] | [2,1] | One swap |
| **Odd length** | [1,2,3] | [2,1,3] | Last node unchanged |
| **Even length** | [1,2,3,4] | [2,1,4,3] | All pairs swapped |

## Common Mistakes

1. **Not checking for second node**
   ```python
   # WRONG: Will error on odd length
   while curr:
       second = curr.next  # None on last node!
   # CORRECT: Check both exist
   while curr and curr.next:
   ```

2. **Losing reference to rest of list**
   ```python
   # WRONG: Overwrites before saving
   curr.next = second.next  # Now we lost second.next!
   # CORRECT: Save next_pair first
   next_pair = second.next
   ```

3. **Not linking previous pair**
   ```python
   # WRONG: Pairs become disconnected
   # Need to link prev.next to second
   ```

## Key Takeaways

| Pattern | When to Use |
|---------|-------------|
| **Dummy head** | Simplify edge cases, especially head changes |
| **Check pair existence** | Always verify `curr and curr.next` |
| **Save references** | Store next nodes before overwriting |
| **Recursive elegance** | When stack space is acceptable |
| **Pair swapping** | Pattern: `prev.next=second, first.next=second.next, second.next=first` |

**Remember**: The dummy node is your friend for list head operations!
