## Problem Statement

Given an integer array `nums` sorted in **non-decreasing order**, return an array of **the squares of each number** sorted in non-decreasing order.

### Visual Example

```
Input:  [-4, -1, 0, 3, 10]
         ↓   ↓   ↓  ↓   ↓
Square: [16,  1, 0, 9, 100]
         ↓   ↓   ↓  ↓   ↓
Sort:   [0,  1, 9, 16, 100]  ← Output
```

### Examples

**Example 1:**
```
Input: nums = [-4,-1,0,3,10]
Output: [0,1,9,16,100]
Explanation: After squaring, the array becomes [16,1,0,9,100].
After sorting, it becomes [0,1,9,16,100].
```

**Example 2:**
```
Input: nums = [-7,-3,2,3,11]
Output: [4,9,9,49,121]
```

### Constraints

- `1 <= nums.length <= 10^4`
- `-10^4 <= nums[i] <= 10^4`
- `nums` is sorted in **non-decreasing** order

### Follow-up

Squaring each element and sorting the new array is very trivial. Could you find an **O(n)** solution using a different approach?

---

## Approach 1: Square and Sort (Brute Force)

### Intuition

The most straightforward approach is to:
1. Square each element
2. Sort the resulting array

### Algorithm

```
1. Create a new array with squared values
2. Sort the array
3. Return the sorted array
```

### Complexity

- **Time**: O(n log n) - dominated by sorting
- **Space**: O(n) or O(log n) depending on sorting algorithm

````python
def sortedSquares_bruteforce(nums):
    """
    Square and sort approach.
    Time: O(n log n), Space: O(n)
    """
    # Square each element
    squared = [num * num for num in nums]
    
    # Sort the squared array
    squared.sort()
    
    return squared

# Test cases
test_cases = [
    [-4, -1, 0, 3, 10],
    [-7, -3, 2, 3, 11],
    [-5, -3, -2, -1],
    [1, 2, 3, 4, 5],
]

print("Brute Force Approach:\n")
for nums in test_cases:
    result = sortedSquares_bruteforce(nums)
    print(f"Input:  {nums}")
    print(f"Output: {result}\n")
````

## Approach 2: Two Pointers (Optimal)

### Intuition

**Key Insight**: The array is already sorted, but when we square negative numbers, they become positive and potentially large.

**Observation**:
- The **largest squared values** come from either:
  - The **leftmost** element (most negative)
  - The **rightmost** element (most positive)
- The **smallest squared values** are near the middle (close to zero)

**Strategy**: Use two pointers at the ends and build the result array **from largest to smallest** (fill from right to left).

### Visual Walkthrough

For `[-4, -1, 0, 3, 10]`:

```
Initial:
[-4, -1, 0, 3, 10]
  ↑              ↑
 left          right

Squares: 16 vs 100 → 100 is larger
Result: [_, _, _, _, 100]
Move right pointer left

[-4, -1, 0, 3, 10]
  ↑           ↑
 left       right

Squares: 16 vs 9 → 16 is larger
Result: [_, _, _, 16, 100]
Move left pointer right

[-4, -1, 0, 3, 10]
      ↑      ↑
    left   right

Squares: 1 vs 9 → 9 is larger
Result: [_, _, 9, 16, 100]
Move right pointer left

[-4, -1, 0, 3, 10]
      ↑   ↑
    left right

Squares: 1 vs 0 → 1 is larger
Result: [_, 1, 9, 16, 100]
Move left pointer right

[-4, -1, 0, 3, 10]
         ↑
      left=right

Last element: 0
Result: [0, 1, 9, 16, 100] ✅
```

### Algorithm

```
1. Initialize two pointers: left = 0, right = n-1
2. Create result array of size n
3. Fill result from right to left (position = n-1 down to 0):
     Compare nums[left]² vs nums[right]²
     Place the larger one at current position
     Move the pointer of the larger value
4. Return result
```

### Why This Works

- **Sorted property**: Original array is sorted
- **Squared values**: Largest squares are at the extremes
- **Build backwards**: Fill result from largest to smallest
- **One pass**: Each element visited exactly once

### Complexity

- **Time**: O(n) - single pass through the array
- **Space**: O(n) - for the result array (O(1) if we don't count output)

````python
def sortedSquares(nums):
    """
    Optimal two-pointer solution.
    Time: O(n), Space: O(n)
    """
    n = len(nums)
    result = [0] * n
    
    left = 0
    right = n - 1
    position = n - 1  # Fill from right to left
    
    while left <= right:
        left_square = nums[left] * nums[left]
        right_square = nums[right] * nums[right]
        
        if left_square > right_square:
            result[position] = left_square
            left += 1
        else:
            result[position] = right_square
            right -= 1
        
        position -= 1
    
    return result

# Test
print("Two Pointers Approach (Optimal):\n")
for nums in test_cases:
    result = sortedSquares(nums)
    print(f"Input:  {nums}")
    print(f"Output: {result}\n")
````

### Detailed Step-by-Step Trace

Let's trace through `[-4, -1, 0, 3, 10]` with detailed output:

````python
def sortedSquares_verbose(nums):
    """Verbose version showing each step."""
    n = len(nums)
    result = [0] * n
    
    left = 0
    right = n - 1
    position = n - 1
    
    print(f"Input array: {nums}")
    print(f"Initial: left={left}, right={right}, position={position}\n")
    
    step = 1
    while left <= right:
        left_square = nums[left] * nums[left]
        right_square = nums[right] * nums[right]
        
        print(f"Step {step}:")
        print(f"  left={left}, nums[left]={nums[left]}, square={left_square}")
        print(f"  right={right}, nums[right]={nums[right]}, square={right_square}")
        
        if left_square > right_square:
            result[position] = left_square
            print(f"  {left_square} > {right_square}, place {left_square} at position {position}")
            print(f"  Move left pointer: {left} → {left + 1}")
            left += 1
        else:
            result[position] = right_square
            print(f"  {right_square} >= {left_square}, place {right_square} at position {position}")
            print(f"  Move right pointer: {right} → {right - 1}")
            right -= 1
        
        position -= 1
        print(f"  Result so far: {result}")
        print(f"  Next position: {position}\n")
        step += 1
    
    print(f"Final result: {result}")
    return result

# Test
sortedSquares_verbose([-4, -1, 0, 3, 10])
````

|----------|----------------|------------------|------|------|

## Common Mistakes

### Mistake 1: Filling Result from Left to Right

**Wrong**:
```python
position = 0  # Start from left
while left <= right:
    # Compare and place smaller value
    if left_square < right_square:
        result[position] = left_square
        left += 1
    else:
        result[position] = right_square
        right -= 1
    position += 1
```

**Problem**: This tries to place the **smaller** value first, but we can't determine which end has the smaller value without additional logic.

**Right**:
```python
position = n - 1  # Start from right
while left <= right:
    # Compare and place larger value
    if left_square > right_square:
        result[position] = left_square
        left += 1
    else:
        result[position] = right_square
        right -= 1
    position -= 1
```

### Mistake 2: Wrong Pointer Movement

**Wrong**:
```python
if left_square > right_square:
    result[position] = left_square
    right -= 1  # ❌ Wrong pointer!
```

**Right**:
```python
if left_square > right_square:
    result[position] = left_square
    left += 1  # ✅ Move the pointer we used
```

### Mistake 3: Not Handling Equal Values

**Wrong**:
```python
if left_square > right_square:
    # ...
elif left_square < right_square:
    # ...
# Missing: what if they're equal?
```

**Right**:
```python
if left_square > right_square:
    result[position] = left_square
    left += 1
else:  # Handles both < and ==
    result[position] = right_square
    right -= 1
```

### Mistake 4: Forgetting to Decrement Position

**Wrong**:
```python
while left <= right:
    if left_square > right_square:
        result[position] = left_square
        left += 1
    # Forgot: position -= 1
```

**Right**:
```python
while left <= right:
    if left_square > right_square:
        result[position] = left_square
        left += 1
    else:
        result[position] = right_square
        right -= 1
    position -= 1  # ✅ Always decrement
```

## Why Two Pointers Works: The Key Insight

### The Problem with Squaring

When we square a sorted array with negative numbers:

```
Original: [-5, -3, -1, 0, 2, 4]
          ↓   ↓   ↓   ↓  ↓  ↓
Squared:  [25, 9,  1, 0, 4, 16]
```

The sorted property is **lost** because:
- Negative numbers become positive when squared
- More negative → larger when squared

### The Insight

**Largest squared values are always at the ends**:
- Either the **most negative** number (leftmost)
- Or the **most positive** number (rightmost)

**Smallest squared values are near zero** (middle of array):
- Numbers closest to zero have smallest squares

### Building from Largest to Smallest

Since we know the largest values are at the ends:
1. Compare the two ends
2. Place the larger one at the end of result
3. Move that pointer inward
4. Repeat until all elements processed

This gives us **O(n)** time with a single pass!

## Key Takeaways

1. **Sorted array property** can be leveraged even after transformation

2. **Two pointers** is powerful for sorted arrays:
   - Start from both ends
   - Move pointers based on comparison
   - Build result strategically

3. **Build backwards** when largest values are at extremes

4. **O(n) vs O(n log n)**: Two pointers avoids sorting overhead

5. **Squaring negative numbers**:
   - More negative → larger square
   - Breaks sorted order
   - But extremes still have largest values

6. **Fill direction matters**: Right-to-left for largest-first

7. **Pointer movement**: Move the pointer whose value you just used

8. **Handle equality**: Use `else` instead of `elif` to cover `==` case

### The Pattern

**Two-pointer technique on sorted arrays**:
- When you need to merge, compare, or build from sorted data
- Start from both ends when extremes are important
- Move pointers inward based on comparison
- Often achieves O(n) time complexity

This pattern appears in many array problems and is essential for optimal solutions!
