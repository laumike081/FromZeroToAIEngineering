## Problem Statement

You are given an integer array `target` and an integer `n`.

You have an empty stack with the two following operations:
- **"Push"**: pushes an integer to the top of the stack.
- **"Pop"**: removes the integer on the top of the stack.

You also have a stream of the integers in the range `[1, n]`.

Use the two stack operations to make the numbers in the stack (from the bottom to the top) equal to `target`. You should follow the following rules:
- If the stream of the integers is not empty, pick the next integer from the stream and push it to the top of the stack.
- If the stack is not empty, pop the integer at the top of the stack.
- If, at any moment, the elements in the stack (from the bottom to the top) are equal to `target`, do not read new integers from the stream and do not do more operations on the stack.

Return the stack operations needed to build `target` following the mentioned rules. If there are multiple valid answers, return **any of them**.

### Examples

**Example 1:**
```
Input: target = [1,3], n = 3
Output: ["Push","Push","Pop","Push"]
Explanation: 
Initially the stack s is empty. The last element is the top of the stack.
Read 1 from stream and push it. s = [1].
Read 2 from stream and push it. s = [1,2].
Pop the integer on the top of the stack. s = [1].
Read 3 from stream and push it. s = [1,3].
```

**Example 2:**
```
Input: target = [1,2,3], n = 3
Output: ["Push","Push","Push"]
Explanation: Initially the stack s is empty. The last element is the top of the stack.
Read 1 from stream and push it. s = [1].
Read 2 from stream and push it. s = [1,2].
Read 3 from stream and push it. s = [1,2,3].
```

**Example 3:**
```
Input: target = [1,2], n = 4
Output: ["Push","Push"]
Explanation: We only need to read the first 2 numbers and stop.
```

### Constraints

- `1 <= target.length <= 100`
- `1 <= n <= 100`
- `1 <= target[i] <= n`
- `target` is strictly increasing.

## Understanding the Problem

### Key Insights

1. **Stream**: Numbers come in order: 1, 2, 3, ..., n
2. **Target**: Strictly increasing array (subset of stream)
3. **Goal**: Build target array using Push/Pop operations

### Strategy

For each number in the stream (1 to n):
- If it's in target → **Push** it
- If it's not in target → **Push** then **Pop** it
- Stop when we've built the entire target

### Visual Example

```
target = [1, 3, 5], n = 5

Stream: 1, 2, 3, 4, 5

Number 1: In target? YES → Push
  Stack: [1]
  Operations: ["Push"]

Number 2: In target? NO → Push, Pop
  Stack: [1, 2] → [1]
  Operations: ["Push", "Push", "Pop"]

Number 3: In target? YES → Push
  Stack: [1, 3]
  Operations: ["Push", "Push", "Pop", "Push"]

Number 4: In target? NO → Push, Pop
  Stack: [1, 3, 4] → [1, 3]
  Operations: ["Push", "Push", "Pop", "Push", "Push", "Pop"]

Number 5: In target? YES → Push
  Stack: [1, 3, 5] ← Target reached!
  Operations: ["Push", "Push", "Pop", "Push", "Push", "Pop", "Push"]
```

### Key Observation

Since target is **strictly increasing**, we can iterate through stream numbers and check if each matches the next target value.

## Approach 1: Simulation with Set

### Algorithm

1. Convert target to a set for O(1) lookup
2. Iterate through stream (1 to n)
3. For each number:
   - If in target: add "Push"
   - If not in target: add "Push", "Pop"
4. Stop when we've processed all target elements

### Complexity

- **Time**: O(n) - iterate through stream up to max(target)
- **Space**: O(m) - where m = len(target) for the set

### Why Use Set?

O(1) lookup to check if number is in target.

````python
def buildArray_set(target, n):
    """
    Simulation with set for O(1) lookup.
    Time: O(n)
    Space: O(m) where m = len(target)
    """
    target_set = set(target)
    operations = []
    
    # Process stream from 1 to max(target)
    for num in range(1, target[-1] + 1):
        if num in target_set:
            operations.append("Push")
        else:
            operations.append("Push")
            operations.append("Pop")
    
    return operations

# Test
print("Approach 1: Simulation with Set\n")
target1, n1 = [1, 3], 3
print(f"target = {target1}, n = {n1}")
print(f"Result: {buildArray_set(target1, n1)}\n")

target2, n2 = [1, 2, 3], 3
print(f"target = {target2}, n = {n2}")
print(f"Result: {buildArray_set(target2, n2)}\n")

target3, n3 = [1, 2], 4
print(f"target = {target3}, n = {n3}")
print(f"Result: {buildArray_set(target3, n3)}")
````

## Approach 2: Two Pointers (Optimal)

### Key Idea

Since target is **strictly increasing**, use a pointer to track current target element.

### Algorithm

1. Initialize pointer `i = 0` for target
2. Iterate stream number from 1 to max(target)
3. For each stream number:
   - If equals `target[i]`: Push, increment i
   - Else: Push, Pop
4. Return operations

### Complexity

- **Time**: O(max(target)) - iterate up to largest target value
- **Space**: O(1) - only use pointer and result list

### Advantage

No extra space for set, cleaner logic.

````python
def buildArray(target, n):
    """
    Two pointers approach (optimal).
    Time: O(max(target))
    Space: O(1)
    """
    operations = []
    target_idx = 0
    
    # Process stream from 1 to max(target)
    for stream_num in range(1, target[-1] + 1):
        if stream_num == target[target_idx]:
            # This number is in target
            operations.append("Push")
            target_idx += 1
        else:
            # This number is not in target
            operations.append("Push")
            operations.append("Pop")
    
    return operations

# Test
print("Approach 2: Two Pointers (Optimal)\n")
target1, n1 = [1, 3], 3
print(f"target = {target1}, n = {n1}")
print(f"Result: {buildArray(target1, n1)}\n")

target2, n2 = [1, 2, 3], 3
print(f"target = {target2}, n = {n2}")
print(f"Result: {buildArray(target2, n2)}\n")

target3, n3 = [1, 2], 4
print(f"target = {target3}, n = {n3}")
print(f"Result: {buildArray(target3, n3)}")
````

## Detailed Trace with Simulation

````python
def buildArray_verbose(target, n):
    """
    Detailed trace showing stack state.
    """
    operations = []
    stack = []
    target_idx = 0
    
    print(f"Target: {target}")
    print(f"Stream: 1 to {n}")
    print(f"We only need up to {target[-1]}\n")
    print("="*80)
    
    for stream_num in range(1, target[-1] + 1):
        print(f"\nStream number: {stream_num}")
        
        if stream_num == target[target_idx]:
            # In target
            print(f"  ✓ {stream_num} is in target (target[{target_idx}])")
            print(f"  Action: Push {stream_num}")
            operations.append("Push")
            stack.append(stream_num)
            target_idx += 1
        else:
            # Not in target
            print(f"  ✗ {stream_num} is NOT in target")
            print(f"  Action: Push {stream_num}, then Pop")
            operations.append("Push")
            operations.append("Pop")
            stack.append(stream_num)
            print(f"  Stack after Push: {stack}")
            stack.pop()
        
        print(f"  Stack: {stack}")
        print(f"  Operations so far: {operations}")
        
        # Check if done
        if stack == target:
            print(f"\n  🎯 Stack matches target! Done.")
            break
    
    print("\n" + "="*80)
    print(f"\nFinal Stack: {stack}")
    print(f"Final Operations: {operations}")
    print(f"Total operations: {len(operations)}")
    return operations

# Trace example 1
print("Example 1: Detailed Trace\n")
buildArray_verbose([1, 3], 3)

print("\n" + "="*80 + "\n")

# Trace example 2
print("Example 2: Detailed Trace\n")
buildArray_verbose([1, 2, 3], 3)
````

## Approach 3: Gap-Based Solution

### Alternative Perspective

Think in terms of **gaps** between consecutive target elements.

### Algorithm

1. For each target element:
   - Calculate gap from previous element
   - Add (gap - 1) × ["Push", "Pop"]
   - Add one "Push" for the target element

### Example

```
target = [1, 3, 5]

Element 1: gap = 1 - 0 = 1
  → 0 Push/Pop pairs, 1 Push
  → ["Push"]

Element 3: gap = 3 - 1 = 2
  → 1 Push/Pop pair (for number 2), 1 Push
  → ["Push", "Pop", "Push"]

Element 5: gap = 5 - 3 = 2
  → 1 Push/Pop pair (for number 4), 1 Push
  → ["Push", "Pop", "Push"]

Total: ["Push", "Push", "Pop", "Push", "Push", "Pop", "Push"]
```

````python
def buildArray_gaps(target, n):
    """
    Gap-based solution.
    Time: O(max(target))
    Space: O(1)
    """
    operations = []
    prev = 0
    
    for num in target:
        gap = num - prev
        
        # Add Push/Pop for numbers we skip
        for _ in range(gap - 1):
            operations.append("Push")
            operations.append("Pop")
        
        # Add Push for target number
        operations.append("Push")
        
        prev = num
    
    return operations

# Test
print("Approach 3: Gap-Based Solution\n")
target1, n1 = [1, 3], 3
print(f"target = {target1}, n = {n1}")
print(f"Result: {buildArray_gaps(target1, n1)}\n")

target2, n2 = [1, 2, 3], 3
print(f"target = {target2}, n = {n2}")
print(f"Result: {buildArray_gaps(target2, n2)}\n")

target3, n3 = [2, 4, 6], 6
print(f"target = {target3}, n = {n3}")
print(f"Result: {buildArray_gaps(target3, n3)}")
````

## Gap-Based Trace

````python
def buildArray_gaps_verbose(target, n):
    """
    Gap-based solution with detailed trace.
    """
    operations = []
    prev = 0
    
    print(f"Target: {target}\n")
    print("="*80)
    
    for i, num in enumerate(target):
        gap = num - prev
        
        print(f"\nTarget element {i}: {num}")
        print(f"  Previous: {prev}")
        print(f"  Gap: {num} - {prev} = {gap}")
        print(f"  Numbers to skip: {gap - 1}")
        
        # Skip numbers
        if gap > 1:
            skipped = list(range(prev + 1, num))
            print(f"  Skipped numbers: {skipped}")
            for skip_num in skipped:
                print(f"    Push {skip_num}, Pop {skip_num}")
                operations.append("Push")
                operations.append("Pop")
        
        # Push target number
        print(f"  Push {num} (target element)")
        operations.append("Push")
        
        print(f"  Operations so far: {operations}")
        
        prev = num
    
    print("\n" + "="*80)
    print(f"\nFinal Operations: {operations}")
    print(f"Total: {len(operations)}")
    return operations

# Trace
print("Gap-Based Approach Trace\n")
buildArray_gaps_verbose([1, 3, 5], 5)
````

## Performance Analysis

````python
import time

def benchmark(target, n):
    """
    Benchmark all approaches.
    """
    results = {}
    
    # Approach 1: Set
    start = time.time()
    result = buildArray_set(target, n)
    elapsed = time.time() - start
    results['Set'] = (len(result), elapsed)
    
    # Approach 2: Two Pointers
    start = time.time()
    result = buildArray(target, n)
    elapsed = time.time() - start
    results['Two Pointers'] = (len(result), elapsed)
    
    # Approach 3: Gaps
    start = time.time()
    result = buildArray_gaps(target, n)
    elapsed = time.time() - start
    results['Gaps'] = (len(result), elapsed)
    
    return results

# Benchmark
print("Performance Comparison\n")
print(f"{'Target Size':<15} {'Approach':<20} {'Operations':<15} {'Time (seconds)'}")
print("="*70)

test_cases = [
    (list(range(1, 11)), 10),           # [1..10]
    (list(range(1, 51, 2)), 50),        # [1,3,5..49]
    (list(range(1, 101)), 100),         # [1..100]
]

for target, n in test_cases:
    results = benchmark(target, n)
    
    for approach, (ops, elapsed) in results.items():
        print(f"{len(target):<15} {approach:<20} {ops:<15} {elapsed:.6f}")
    print()

print("Observation:")
print("- All approaches have similar O(max(target)) time complexity")
print("- Two Pointers is most space-efficient (no set)")
print("- Gaps approach is most intuitive conceptually")
````

## Key Takeaways

### Problem Understanding

1. **Stream is ordered**: 1, 2, 3, ..., n
2. **Target is strictly increasing**: Subset of stream
3. **Two operations**: Push (add to stack), Pop (remove from stack)
4. **Goal**: Build target array using these operations

### Core Strategy

For each number in stream (1 to max(target)):
- **If in target**: Push it
- **If not in target**: Push then Pop (to skip it)

### Key Insights

1. **Only process up to max(target)**: No need to go beyond
2. **Target is sorted**: Can use pointer instead of set
3. **Gaps matter**: Numbers between target elements need Push/Pop

### Algorithm Pattern

```python
operations = []
target_idx = 0

for stream_num in range(1, target[-1] + 1):
    if stream_num == target[target_idx]:
        operations.append("Push")
        target_idx += 1
    else:
        operations.append("Push")
        operations.append("Pop")

return operations
```

### Common Mistakes

❌ **Processing entire stream up to n**
```python
# WRONG - unnecessary iterations
for num in range(1, n + 1):
```

✅ **Only process up to max(target)**
```python
# CORRECT - stop early
for num in range(1, target[-1] + 1):
```

❌ **Using set when not needed**
```python
# WRONG - extra space
target_set = set(target)
if num in target_set:
```

✅ **Using pointer for sorted array**
```python
# CORRECT - O(1) space
if num == target[target_idx]:
```

### Edge Cases to Remember

1. **Single element**: Just one Push
2. **Consecutive numbers**: All Push, no Pop
3. **Large gaps**: Many Push/Pop pairs
4. **Target starts > 1**: Skip initial numbers
5. **n > max(target)**: Stop at max(target)

### Complexity Analysis

**Two Pointers (Optimal):**
- **Time**: O(max(target))
  - Iterate from 1 to max(target)
  - Each iteration: O(1)
- **Space**: O(1)
  - Only use: target_idx, operations list
  - Operations list is output, doesn't count

### Gap-Based Thinking

Alternative way to think:
- **Gap = current - previous**
- **Skip (gap - 1) numbers**: Each needs Push/Pop
- **Push current number**: Add to stack

### Remember

🎯 **Stream is ordered** 1 to n  
🎯 **Target is strictly increasing**  
🎯 **Stop at max(target)**, not n  
🎯 **Use pointer** for sorted target  
🎯 **Push/Pop for skipped** numbers  
🎯 **O(max(target)) time**, O(1) space
