# 844. Backspace String Compare

**Difficulty**: Easy  
**Topics**: String, Stack, Two Pointers  
**Link**: [LeetCode 844](https://leetcode.com/problems/backspace-string-compare/)

---

## Problem Statement

Given two strings `s` and `t`, return `true` if they are equal when both are typed into empty text editors.

The character `'#'` means a backspace character.

### Examples

**Example 1:**
```
Input: s = "ab#c", t = "ad#c"
Output: true
Explanation: Both s and t become "ac".
```

**Example 2:**
```
Input: s = "ab##", t = "c#d#"
Output: true
Explanation: Both s and t become "".
```

**Example 3:**
```
Input: s = "a#c", t = "b"
Output: false
Explanation: s becomes "c" while t becomes "b".
```

### Constraints

- `1 <= s.length, t.length <= 200`
- `s` and `t` only contain lowercase letters and `'#'` characters.

---

## Approach 1: Stack Simulation

### Intuition

Use a stack to simulate the typing process. When we encounter a character, push it to the stack. When we encounter `'#', pop from the stack if not empty.

### Why it works

The stack naturally models the behavior of a text editor - adding characters and removing the most recent one when backspacing.

### Algorithm Visualization

```
s = "ab#c"

Stack: []

a: push -> [a]
b: push -> [a, b]
#: pop -> [a]
c: push -> [a, c]

Result: "ac"
```

### Complexity
- **Time**: O(n + m) - where n and m are lengths of s and t
- **Space**: O(n + m) - worst case stack sizes

````python
def backspace_compare_stack(s: str, t: str) -> bool:
    """Stack simulation: O(n+m) time, O(n+m) space"""
    def process_string(string):
        stack = []
        for char in string:
            if char == '#':
                if stack:
                    stack.pop()
            else:
                stack.append(char)
        return ''.join(stack)
    
    return process_string(s) == process_string(t)

# Test cases
test_cases = [
    ("ab#c", "ad#c", True),
    ("ab##", "c#d#", True),
    ("a#c", "b", False),
    ("", "", True),
    ("#", "", True),
    ("a#b#c", "c", True),
    ("xy#z", "xz", True),
    ("xy##", "", True)
]

for s, t, expected in test_cases:
    result = backspace_compare_stack(s, t)
    print(f"{s!r:8} vs {t!r:8} -> {result} (expected {expected})")
````

## Approach 2: Two Pointers (Reverse)

### Intuition

Traverse both strings from the end, skipping characters that are deleted by backspaces.

### Why it works

When traversing from the end, we can count how many characters to skip based on the number of backspaces encountered.

### Algorithm

1. Start from the end of both strings
2. For each string:
   - Count backspaces
   - Skip that many characters
3. Compare current characters
4. Move pointers backward

### Complexity
- **Time**: O(n + m)
- **Space**: O(1) - constant extra space

````python
def backspace_compare_two_pointers(s: str, t: str) -> bool:
    """Two pointers reverse: O(n+m) time, O(1) space"""
    i, j = len(s) - 1, len(t) - 1
    
    while i >= 0 or j >= 0:
        # Find next valid character in s
        skip_s = 0
        while i >= 0:
            if s[i] == '#':
                skip_s += 1
                i -= 1
            elif skip_s > 0:
                skip_s -= 1
                i -= 1
            else:
                break
        
        # Find next valid character in t
        skip_t = 0
        while j >= 0:
            if t[j] == '#':
                skip_t += 1
                j -= 1
            elif skip_t > 0:
                skip_t -= 1
                j -= 1
            else:
                break
        
        # Compare characters
        if i >= 0 and j >= 0:
            if s[i] != t[j]:
                return False
        elif i >= 0 or j >= 0:
            # One string has characters left, other doesn't
            return False
        
        i -= 1
        j -= 1
    
    return True

# Test
for s, t, expected in test_cases:
    result = backspace_compare_two_pointers(s, t)
    print(f"{s!r:8} vs {t!r:8} -> {result} (expected {expected})")
````

## Approach 3: In-place Modification

### Intuition

Modify the strings in-place to remove backspaces, then compare.

### Why it works

We can use the string as a character array and build the result from the beginning.

### Complexity
- **Time**: O(n + m)
- **Space**: O(1) - modifying in-place

````python
def backspace_compare_inplace(s: str, t: str) -> bool:
    """In-place modification: O(n+m) time, O(1) space"""
    def process_string(string):
        chars = list(string)
        write = 0
        
        for read in range(len(chars)):
            if chars[read] != '#':
                chars[write] = chars[read]
                write += 1
            elif write > 0:
                write -= 1
        
        return ''.join(chars[:write])
    
    return process_string(s) == process_string(t)

# Test
for s, t, expected in test_cases:
    result = backspace_compare_inplace(s, t)
    print(f"{s!r:8} vs {t!r:8} -> {result} (expected {expected})")
````

## Walkthrough (Two Pointers Approach)

Input: `s = "ab#c"`, `t = "ad#c"`

**Step 1: Initialize pointers**
- i = 2 (last index of s), j = 2 (last index of t)
- s[i] = 'c', t[j] = 'c'

**Step 2: Compare 'c' vs 'c'**
- Equal, continue
- i = 1, j = 1

**Step 3: Process backspace in s**
- s[i] = '#', skip_s = 1
- i = 0, skip_s = 1 (still skipping)
- s[i] = 'a', skip_s = 0 (skip done)
- i = -1 (stop)

**Step 4: Process backspace in t**
- t[j] = '#', skip_t = 1
- j = 0, skip_t = 1 (still skipping)
- t[j] = 'a', skip_t = 0 (skip done)
- j = -1 (stop)

**Step 5: Both pointers at -1**
- Both strings exhausted, equal ✓

````python
def backspace_compare_verbose(s: str, t: str) -> bool:
    """Verbose two pointers approach"""
    i, j = len(s) - 1, len(t) - 1
    
    print(f"Comparing {s!r} vs {t!r}")
    
    while i >= 0 or j >= 0:
        print(f"\nPointers: i={i}, j={j}")
        
        # Find next valid in s
        skip_s = 0
        while i >= 0:
            if s[i] == '#':
                skip_s += 1
                print(f"  s[{i}] = '#', skip_s = {skip_s}")
                i -= 1
            elif skip_s > 0:
                print(f"  s[{i}] = '{s[i]}', skipping (skip_s={skip_s})")
                skip_s -= 1
                i -= 1
            else:
                break
        
        # Find next valid in t
        skip_t = 0
        while j >= 0:
            if t[j] == '#':
                skip_t += 1
                print(f"  t[{j}] = '#', skip_t = {skip_t}")
                j -= 1
            elif skip_t > 0:
                print(f"  t[{j}] = '{t[j]}', skipping (skip_t={skip_t})")
                skip_t -= 1
                j -= 1
            else:
                break
        
        # Compare
        if i >= 0 and j >= 0:
            print(f"  Comparing '{s[i]}' vs '{t[j]}'")
            if s[i] != t[j]:
                print(f"  Not equal!")
                return False
        elif i >= 0 or j >= 0:
            print(f"  One string exhausted: i={i}, j={j}")
            return False
        else:
            print(f"  Both strings exhausted")
        
        i -= 1
        j -= 1
    
    print("\nStrings are equal!")
    return True

# Test with verbose output
backspace_compare_verbose("ab#c", "ad#c")
````

## Comparison

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| **Stack Simulation** | O(n+m) | O(n+m) | Intuitive, easy to understand | Extra space |
| **Two Pointers** | O(n+m) | O(1) | Optimal space | More complex logic |
| **In-place** | O(n+m) | O(1) | No extra storage | Modifies input (conceptually) |

**Best choice**: **Two Pointers** for optimal space, **Stack** for clarity.

## Edge Cases

| Case | Input | Output | Reason |
|------|-------|--------|--------|
| **Empty strings** | "", "" | True | Both empty |
| **Only backspaces** | "####", "###" | True | Both become empty |
| **More backspaces** | "a###", "b##" | True | Both become empty |
| **Single char** | "a", "a" | True | Same single character |

| **Different lengths** | "a#", "" | True | Both become empty |
| **Complex case** | "xy#z", "xz" | True | 'y' is backspaced |
| **All backspaced** | "abc###", "" | True | All characters removed |

## Common Mistakes

1. **Not checking empty stack**
   ```python
   # WRONG: IndexError on empty stack
   if char == '#':
       stack.pop()
   # CORRECT: Check if stack exists
   if char == '#' and stack:
       stack.pop()
   ```

2. **Wrong skip logic**
   ```python
   # WRONG: Doesn't handle consecutive backspaces
   if s[i] == '#':
       i -= 2
   # CORRECT: Count backspaces
   skip = 0
   while i >= 0 and (s[i] == '#' or skip > 0):
   ```

3. **Early termination**
   ```python
   # WRONG: Might miss characters
   while i >= 0 and j >= 0:
   # CORRECT: Handle different lengths
   while i >= 0 or j >= 0:
   ```

## Key Takeaways

| Pattern | When to Use |
|---------|-------------|
| **Stack simulation** | When processing affects previous elements |
| **Two pointers reverse** | When backspaces affect preceding characters |
| **Skip counting** | When you need to skip multiple elements |
| **In-place processing** | When space optimization is critical |
| **Reverse traversal** | When operations affect previous elements |

**Remember**: Process from the end when dealing with backspaces - it's much easier!
