## Problem Statement

In a linked list of size `n`, where `n` is **even**, the `i`th node (0-indexed) of the linked list is known as the **twin** of the `(n-1-i)`th node.

The **twin sum** is defined as the sum of a node and its twin.

Given the `head` of a linked list with even length, return the **maximum twin sum** of the linked list.

### Examples

**Example 1:**
```
Input: head = [5,4,2,1]
Output: 6
Explanation: Twins are (5,1) and (4,2). Max sum = 5+1 = 6.
```

**Example 2:**
```
Input: head = [4,2,2,3]
Output: 7
Explanation: Twins are (4,3) and (2,2). Max sum = 4+3 = 7.
```

**Example 3:**
```
Input: head = [1,100000]
Output: 100001
Explanation: Twins are (1,100000). Max sum = 1+100000 = 100001.
```

### Constraints

- The number of nodes is in the range `[2, 10^5]`
- `1 <= Node.val <= 10^5`
- The length of the linked list is **even**

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    
    def __repr__(self):
        result = []
        node = self
        while node:
            result.append(str(node.val))
            node = node.next
        return " -> ".join(result)

def create_list(values):
    """Helper to create linked list from array"""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head
````

## Approach 1: Reverse Second Half

### Intuition

1. Find the middle of the list using slow/fast pointers
2. Reverse the second half of the list
3. Compare nodes from first half with reversed second half

### Why it works

For an even-length list, the twins are:
- First half: nodes 0, 1, 2, ..., n/2-1
- Second half: nodes n-1, n-2, ..., n/2

By reversing the second half, we align twins in the same direction for easy comparison.

### Visualization

```
Original: 5 -> 4 -> 2 -> 1
          |    |    |    |
Twin:    0    1    2    3
         ^         ^
         |         |
        twin     twin

After reversing second half:
5 -> 4    1 -> 2
|    |    |    |
0    1    3    2

Now we can compare: (5,1) and (4,2)
```

### Complexity
- **Time**: O(n) - one pass to find middle, one to reverse, one to compare
- **Space**: O(1) - in-place reversal

````python
def pair_sum_reverse_half(head: ListNode) -> int:
    """Reverse second half: O(n) time, O(1) space"""
    # Step 1: Find middle using slow/fast pointers
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    
    # Step 2: Reverse second half
    prev = None
    curr = slow
    while curr:
        next_node = curr.next
        curr.next = prev
        prev = curr
        curr = next_node
    
    # Step 3: Compare pairs and find max sum
    max_sum = 0
    first = head
    second = prev  # Head of reversed second half
    
    while second:
        max_sum = max(max_sum, first.val + second.val)
        first = first.next
        second = second.next
    
    return max_sum

# Test
head = create_list([5, 4, 2, 1])
print(f"List: {head}")
print(f"Max twin sum: {pair_sum_reverse_half(head)}")  # 6

head2 = create_list([4, 2, 2, 3])
print(f"\nList: {head2}")
print(f"Max twin sum: {pair_sum_reverse_half(head2)}")  # 7

head3 = create_list([1, 100000])
print(f"\nList: {head3}")
print(f"Max twin sum: {pair_sum_reverse_half(head3)}")  # 100001
````

## Approach 2: Stack

### Intuition

Push first half of nodes onto stack, then traverse second half and calculate sums.

### Why it works

Stack gives us LIFO access, so when we traverse the second half in forward direction, we get the corresponding twin from the first half in reverse order.

### Complexity
- **Time**: O(n)
- **Space**: O(n) - stack stores n/2 nodes

````python
def pair_sum_stack(head: ListNode) -> int:
    """Stack: O(n) time, O(n) space"""
    # Find middle
    slow = fast = head
    stack = []
    
    # Push first half onto stack
    while fast and fast.next:
        stack.append(slow.val)
        slow = slow.next
        fast = fast.next.next
    
    # Calculate twin sums
    max_sum = 0
    while slow:
        twin_sum = stack.pop() + slow.val
        max_sum = max(max_sum, twin_sum)
        slow = slow.next
    
    return max_sum

# Test
head = create_list([5, 4, 2, 1])
print(f"List: {head}")
print(f"Max twin sum (stack): {pair_sum_stack(head)}")  # 6
````

## Approach 3: Array Conversion

### Intuition

Convert linked list to array, then compare symmetric elements.

### Complexity
- **Time**: O(n)
- **Space**: O(n) - array stores all nodes

````python
def pair_sum_array(head: ListNode) -> int:
    """Array: O(n) time, O(n) space"""
    # Convert to array
    values = []
    curr = head
    while curr:
        values.append(curr.val)
        curr = curr.next
    
    # Calculate twin sums
    n = len(values)
    max_sum = 0
    
    for i in range(n // 2):
        twin_sum = values[i] + values[n - 1 - i]
        max_sum = max(max_sum, twin_sum)
    
    return max_sum

# Test
head = create_list([5, 4, 2, 1])
print(f"List: {head}")
print(f"Max twin sum (array): {pair_sum_array(head)}")  # 6
````

## Walkthrough (Reverse Second Half)

Input: `[5,4,2,1]`

**Step 1: Find middle**
- slow moves: 5 → 4 → 2
- fast moves: 5 → 2 → None
- slow stops at 2 (start of second half)

**Step 2: Reverse second half**
- Original second half: 2 → 1
- After reversal: 1 → 2

**Step 3: Compare pairs**
- First half: 5 → 4
- Reversed second: 1 → 2
- Pairs: (5,1) = 6, (4,2) = 6
- Max: 6

````python
def pair_sum_verbose(head: ListNode) -> int:
    """Verbose version showing steps"""
    print(f"Original list: {head}")
    
    # Find middle
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    print(f"Middle node: {slow.val}")
    
    # Reverse second half
    prev = None
    curr = slow
    while curr:
        next_node = curr.next
        curr.next = prev
        prev = curr
        curr = next_node
    print(f"Reversed second half starts with: {prev.val}")
    
    # Compare pairs
    max_sum = 0
    first = head
    second = prev
    i = 0
    
    while second:
        twin_sum = first.val + second.val
        print(f"Pair {i}: ({first.val}, {second.val}) = {twin_sum}")
        max_sum = max(max_sum, twin_sum)
        first = first.next
        second = second.next
        i += 1
    
    print(f"Maximum twin sum: {max_sum}")
    return max_sum

head = create_list([5, 4, 2, 1])
pair_sum_verbose(head)
````

## Comparison

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| **Reverse second half** | O(n) | O(1) | Optimal space, no extra storage | Modifies list structure |
| **Stack** | O(n) | O(n) | Simple logic, preserves list | Extra space overhead |
| **Array** | O(n) | O(n) | Easiest to understand | Most space usage |

**Best choice**: **Reverse second half** for optimal space complexity.

## Common Mistakes

1. **Wrong middle finding**
   ```python
   # WRONG: For even length, slow ends at n/2+1
   while fast:
       slow = slow.next
       fast = fast.next.next
   # CORRECT: Use while fast and fast.next
   ```

2. **Not handling even length correctly**
   - The problem guarantees even length, but some solutions might assume odd

3. **Integer overflow**
   - In Python this isn't an issue, but in other languages, use long long

4. **Modifying original list**
   - If list must be preserved, use stack or array approach

## Key Takeaways

| Pattern | When to Use |
|---------|-------------|
| **Reverse second half** | Compare symmetric pairs in linked list |
| **Slow/Fast pointers** | Find middle of linked list efficiently |
| **Stack for LIFO** | When you need reverse access without modification |
| **Array conversion** | When random access is more important than space |
| **Twin relationship** | For index i, twin is n-1-i in 0-indexed array/list |

**Remember**: For even-length lists, slow/fast with `while fast and fast.next` finds the exact middle!
