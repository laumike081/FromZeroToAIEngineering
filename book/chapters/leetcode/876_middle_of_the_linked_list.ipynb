# 876. Middle of the Linked List

**Difficulty**: Easy  
**Topics**: Linked List, Two Pointers  
**Link**: [LeetCode 876](https://leetcode.com/problems/middle-of-the-linked-list/)

---

## Problem Statement

Given the `head` of a singly linked list, return the middle node of the linked list.

If there are two middle nodes, return the **second middle node**.

### Examples

**Example 1:**
```
Input: head = [1,2,3,4,5]
Output: [3,4,5]
Explanation: The middle node of the list is node 3.
```

**Example 2:**
```
Input: head = [1,2,3,4,5,6]
Output: [4,5,6]
Explanation: Since the list has two middle nodes with values 3 and 4, we return the second one.
```

### Constraints

- The number of nodes in the list is in the range `[1, 100]`
- `1 <= Node.val <= 100`

---

## Approach 1: Slow and Fast Pointers

### Intuition

Use two pointers moving at different speeds:
- **Slow pointer**: moves 1 step at a time
- **Fast pointer**: moves 2 steps at a time

When fast reaches the end, slow will be at the middle.

### Why it works

For a list of length `n`:
- Fast moves `2k` steps while slow moves `k` steps
- When fast can't move 2 steps anymore, slow is at `⌊n/2⌋`
- For even `n`, this gives the **second** middle node

### Complexity
- **Time**: O(n) - single traversal
- **Space**: O(1) - constant extra space

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def middle_node(head: ListNode) -> ListNode:
    """Slow/Fast pointers: O(n) time, O(1) space"""
    slow = fast = head
    
    while fast and fast.next:
        slow = slow.next      # Move 1 step
        fast = fast.next.next # Move 2 steps
    
    return slow  # slow is at middle

# --- Test ---
# Create list: 1 -> 2 -> 3 -> 4 -> 5
h = ListNode(1, ListNode(2, ListNode(3, ListNode(4, ListNode(5)))))
middle = middle_node(h)
result = []
while middle:
    result.append(middle.val)
    middle = middle.next
print(f"Odd length [1,2,3,4,5]: {result}")  # [3, 4, 5]

# Create list: 1 -> 2 -> 3 -> 4 -> 5 -> 6
h2 = ListNode(1, ListNode(2, ListNode(3, ListNode(4, ListNode(5, ListNode(6))))))
middle2 = middle_node(h2)
result2 = []
while middle2:
    result2.append(middle2.val)
    middle2 = middle2.next
print(f"Even length [1,2,3,4,5,6]: {result2}")  # [4, 5, 6]
````

---

## Walkthrough

### Odd Length: [1,2,3,4,5]

| Step | slow | fast | Action | Continue? |
|------|------|------|--------|------------|
| 0 | 1 | 1 | Initialize | Yes |
| 1 | 2 | 3 | slow=slow.next, fast=fast.next.next | Yes |
| 2 | 3 | 5 | slow=slow.next, fast=fast.next.next | Yes |
| 3 | 4 | None | slow=slow.next, fast=fast.next.next | No |

**Result**: slow at node 3 (middle)

### Even Length: [1,2,3,4,5,6]

| Step | slow | fast | Action | Continue? |
|------|------|------|--------|------------|
| 0 | 1 | 1 | Initialize | Yes |
| 1 | 2 | 3 | slow=slow.next, fast=fast.next.next | Yes |
| 2 | 3 | 5 | slow=slow.next, fast=fast.next.next | Yes |
| 3 | 4 | None | slow=slow.next, fast=fast.next.next | No |

**Result**: slow at node 4 (second middle)

## Approach 2: Count and Navigate

### Intuition

Two-pass approach:
1. First pass: count total nodes
2. Second pass: navigate to middle position

### Complexity
- **Time**: O(n) - two traversals
- **Space**: O(1) - constant extra space

````python
def middle_node_count(head: ListNode) -> ListNode:
    """Count then navigate: O(n) time, O(1) space"""
    # Count nodes
    count = 0
    curr = head
    while curr:
        count += 1
        curr = curr.next
    
    # Navigate to middle (second middle for even)
    middle_idx = count // 2
    curr = head
    for _ in range(middle_idx):
        curr = curr.next
    
    return curr

# --- Test ---
h = ListNode(1, ListNode(2, ListNode(3, ListNode(4, ListNode(5)))))
middle = middle_node_count(h)
result = []
while middle:
    result.append(middle.val)
    middle = middle.next
print(f"Count method: {result}")  # [3, 4, 5]
````

## Approach 3: Array Conversion

### Intuition

Convert linked list to array, then return middle element.

### Complexity
- **Time**: O(n) - one traversal + conversion
- **Space**: O(n) - store all nodes in array

````python
def middle_node_array(head: ListNode) -> ListNode:
    """Array conversion: O(n) time, O(n) space"""
    nodes = []
    curr = head
    while curr:
        nodes.append(curr)
        curr = curr.next
    
    # Return middle (second middle for even)
    middle_idx = len(nodes) // 2
    return nodes[middle_idx]

# --- Test ---
h = ListNode(1, ListNode(2, ListNode(3, ListNode(4, ListNode(5, ListNode(6))))))
middle = middle_node_array(h)
result = []
while middle:
    result.append(middle.val)
    middle = middle.next
print(f"Array method: {result}")  # [4, 5, 6]
````

## Approach 4: Single Pass with Counter

### Intuition

Keep track of position while traversing. Return when reaching middle.

### Complexity
- **Time**: O(n) - single traversal
- **Space**: O(1) - constant extra space

````python
def middle_node_single_pass(head: ListNode) -> ListNode:
    """Single pass with counter: O(n) time, O(1) space"""
    # First pass to count (we need total length)
    length = 0
    curr = head
    while curr:
        length += 1
        curr = curr.next
    
    # Second pass to find middle
    middle_idx = length // 2
    curr = head
    for _ in range(middle_idx):
        curr = curr.next
    
    return curr

# --- Test ---
h = ListNode(1, ListNode(2, ListNode(3)))
middle = middle_node_single_pass(h)
result = []
while middle:
    result.append(middle.val)
    middle = middle.next
print(f"Single pass: {result}")  # [2, 3]
````

---

## Comparison

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| **Slow/Fast pointers** | O(n) | O(1) | Single pass, elegant | Less intuitive |
| **Count then navigate** | O(n) | O(1) | Simple logic | Two passes |
| **Array conversion** | O(n) | O(n) | Easy indexing | Uses extra space |
| **Single pass + counter** | O(n) | O(1) | Clear position tracking | Still two passes |

**Best choice**: **Slow/Fast pointers** - optimal single pass with O(1) space.
