# 20. Valid Parentheses

**Difficulty**: Easy  
**Topics**: String, Stack  
**Link**: [LeetCode 20](https://leetcode.com/problems/valid-parentheses/)

---

## Problem Statement

Given a string `s` containing just the characters `'('`, `')'`, `'{'`, `'}'`, `'['` and `']'`, determine if the input string is valid.

An input string is valid if:
1. Open brackets must be closed by the same type of brackets.
2. Open brackets must be closed in the correct order.
3. Every close bracket has a corresponding open bracket of the same type.

### Examples

**Example 1:**
```
Input: s = "()"
Output: true
```

**Example 2:**
```
Input: s = "()[]{}"
Output: true
```

**Example 3:**
```
Input: s = "(]"
Output: false
```

### Constraints

- `1 <= s.length <= 10^4`
- `s` consists only of `'('`, `')'`, `'{'`, `'}'`, `'['` and `']'`.

---

## Approach 1: Stack

### Intuition

Use a stack to keep track of opening brackets. When we encounter a closing bracket, check if it matches the top of the stack.

### Why it works

The stack naturally handles the LIFO (Last In, First Out) nature of nested brackets. The most recent opening bracket must be closed first.

### Algorithm

1. Create a mapping of closing to opening brackets
2. Iterate through the string:
   - If opening bracket, push to stack
   - If closing bracket, check if stack is empty or top doesn't match
3. At the end, stack should be empty

### Complexity
- **Time**: O(n) - one pass through the string
- **Space**: O(n) - worst case stack size (all opening brackets)

````python
def is_valid_stack(s: str) -> bool:
    """Stack approach: O(n) time, O(n) space"""
    # Mapping of closing to opening brackets
    bracket_map = {')': '(', '}': '{', ']': '['}
    
    stack = []
    
    for char in s:
        if char in bracket_map.values():  # Opening bracket
            stack.append(char)
        else:  # Closing bracket
            if not stack or stack[-1] != bracket_map[char]:
                return False
            stack.pop()
    
    return len(stack) == 0

# Test cases
test_cases = [
    ("()", True),
    ("()[]{}", True),
    ("(]", False),
    ("([)]", False),
    ("{[]}", True),
    ("", True),  # Empty string is valid
    ("((", False),
    (")", False)
]

for s, expected in test_cases:
    result = is_valid_stack(s)
    print(f"{s!r:8} -> {result} (expected {expected})")
````

---

## Approach 2: Replace Pairs

### Intuition

Repeatedly remove valid pairs until no more can be removed.

### Why it works

If the string is valid, we can keep removing adjacent valid pairs until empty.

### Complexity
- **Time**: O(n²) - worst case O(n) passes over O(n) string
- **Space**: O(1) - no extra storage

````python
def is_valid_replace(s: str) -> bool:
    """Replace pairs approach: O(n²) time, O(1) space"""
    while True:
        found = False
        
        # Remove all valid pairs
        if '()' in s:
            s = s.replace('()', '')
            found = True
        if '[]' in s:
            s = s.replace('[]', '')
            found = True
        if '{}' in s:
            s = s.replace('{}', '')
            found = True
        
        if not found:
            break
    
    return len(s) == 0

# Test
for s, expected in test_cases:
    result = is_valid_replace(s)
    print(f"{s!r:8} -> {result} (expected {expected})")
````

## Approach 3: Counting (Limited)

### Intuition

Use counters for each bracket type to track balance.

### Why it works (with limitations)

This works for simple cases but fails for nested different types like `([)]`.

### Complexity
- **Time**: O(n)
- **Space**: O(1)

````python
def is_valid_counting(s: str) -> bool:
    """Counting approach: O(n) time, O(1) space (limited)"""
    # This approach only works for single bracket type
    # Fails for mixed types like ([)]
    
    count = 0
    for char in s:
        if char == '(':
            count += 1
        elif char == ')':
            count -= 1
            if count < 0:
                return False
    
    return count == 0

# Test with single bracket type
single_tests = [
    ("()", True),
    ("((()))", True),
    ("(()", False),
    (")()", False)
]

print("Counting approach (single bracket type only):")
for s, expected in single_tests:
    result = is_valid_counting(s)
    print(f"{s!r:8} -> {result} (expected {expected})")

print("\nFails for mixed types:")
print(f"([)] -> {is_valid_counting('([)]')} (should be False)")
````

## Walkthrough (Stack Approach)

Input: `"{[()]})"`

| Step | Char | Stack | Action | Valid? |
|------|------|-------|--------|--------|
| 0 | - | [] | Start | ✓ |
| 1 | `{` | [`{`] | Push opening | ✓ |
| 2 | `[` | [`{`, `[`] | Push opening | ✓ |
| 3 | `(` | [`{`, `[`, `(`] | Push opening | ✓ |
| 4 | `)` | [`{`, `[`] | Pop matching `(` | ✓ |
| 5 | `]` | [`{`] | Pop matching `[` | ✓ |
| 6 | `}` | [] | Pop matching `{` | ✓ |
| 7 | `)` | [] | Stack empty, can't pop | ✗ |

**Result**: Invalid (extra closing bracket)

````python
def is_valid_verbose(s: str) -> bool:
    """Verbose stack approach showing steps"""
    bracket_map = {')': '(', '}': '{', ']': '['}
    stack = []
    
    print(f"Validating: {s!r}")
    
    for i, char in enumerate(s):
        print(f"\nStep {i+1}: Char = {char!r}")
        print(f"  Stack before: {stack}")
        
        if char in bracket_map.values():  # Opening
            stack.append(char)
            print(f"  Action: Push {char!r}")
        else:  # Closing
            if not stack:
                print(f"  Action: Stack empty, invalid!")
                return False
            if stack[-1] != bracket_map[char]:
                print(f"  Action: Mismatch! Expected {stack[-1]!r}, got {bracket_map[char]!r}")
                return False
            stack.pop()
            print(f"  Action: Pop matching {bracket_map[char]!r}")
        
        print(f"  Stack after: {stack}")
    
    print(f"\nFinal stack: {stack}")
    result = len(stack) == 0
    print(f"Result: {result}")
    return result

# Test with verbose output
is_valid_verbose("{[()]}")
print("\n" + "="*40)
is_valid_verbose("{[()]})")
````

## Comparison

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| **Stack** | O(n) | O(n) | Correct for all cases, standard solution | Extra space |
| **Replace Pairs** | O(n²) | O(1) | No extra space, intuitive | Inefficient for long strings |
| **Counting** | O(n) | O(1) | Fast, no space | Only works for single bracket type |

**Best choice**: **Stack** - optimal and handles all cases correctly.

## Edge Cases

| Case | Input | Output | Reason |
|------|-------|--------|--------|
| **Empty string** | "" | True | No brackets to validate |
| **Single opening** | "(" | False | Unclosed bracket |
| **Single closing** | ")" | False | No matching opening |
| **All opening** | "((({" | False | Unclosed brackets |
| **All closing** | ")]" | False | No matching openings |
| **Nested valid** | "({[]})" | True | Properly nested |
| **Interleaved invalid** | "([)]" | False | Wrong nesting order |
| **Long valid** | "()()()()" | True | Sequential pairs |
| **Complex valid** | "{[()({[]})]}" | True | Deep nesting |

## Common Mistakes

1. **Wrong bracket mapping**
   ```python
   # WRONG: Maps opening to closing
   bracket_map = {'(': ')', '[': ']', '{': '}'}
   # CORRECT: Maps closing to opening
   bracket_map = {')': '(', ']': '[', '}': '{'}
   ```

2. **Not checking empty stack**
   ```python
   # WRONG: Stack might be empty
   if stack[-1] != bracket_map[char]:
   # CORRECT: Check if stack is empty first
   if not stack or stack[-1] != bracket_map[char]:
   ```

3. **Only checking final stack**
   ```python
   # WRONG: Might have invalid nesting
   return len(stack) == 0
   # CORRECT: Check matches during iteration
   ```

## Key Takeaways

| Pattern | When to Use |
|---------|-------------|
| **Stack for matching** | Parentheses, brackets, HTML tags |
| **LIFO property** | Nested structures, undo operations |
| **Mapping pairs** | Quick lookup of matching pairs |
| **Early validation** | Check invalid cases immediately |
| **Empty check** | Always validate before accessing stack top |

**Remember**: Stack is the natural choice for nested structure validation!
