## Problem Statement

Given an array `nums` of integers, return how many of them contain an **even number of digits**.

### Examples

**Example 1:**
```
Input: nums = [12,345,2,6,7896]
Output: 2
Explanation:
- 12 contains 2 digits (even number of digits)
- 345 contains 3 digits (odd number of digits)
- 2 contains 1 digit (odd number of digits)
- 6 contains 1 digit (odd number of digits)
- 7896 contains 4 digits (even number of digits)
Therefore only 12 and 7896 contain an even number of digits.
```

**Example 2:**
```
Input: nums = [555,901,482,1771]
Output: 1
Explanation:
Only 1771 contains an even number of digits (4 digits).
```

### Constraints

- `1 <= nums.length <= 500`
- `1 <= nums[i] <= 10^5`

## Approach 1: Convert to String

### Intuition

The simplest approach is to convert each number to a string and count its length. If the length is even, increment the counter.

### Algorithm

```
1. Initialize count = 0
2. For each number in nums:
     Convert number to string
     If length of string is even:
         count++
3. Return count
```

### Complexity

- **Time**: O(n × d) where n = array length, d = average digits per number
  - Converting to string takes O(d) time
- **Space**: O(d) for the string representation

````python
def findNumbers_string(nums):
    """
    Count numbers with even number of digits using string conversion.
    Time: O(n × d), Space: O(d)
    """
    count = 0
    
    for num in nums:
        # Convert to string and check length
        if len(str(num)) % 2 == 0:
            count += 1
    
    return count

# Test cases
test_cases = [
    [12, 345, 2, 6, 7896],
    [555, 901, 482, 1771],
    [1, 22, 333, 4444, 55555],
]

for nums in test_cases:
    result = findNumbers_string(nums)
    print(f"Input: {nums}")
    print(f"Output: {result}\n")
````

### Detailed Walkthrough

Let's trace through `[12, 345, 2, 6, 7896]`:

````python
def findNumbers_verbose(nums):
    """Verbose version showing each step."""
    count = 0
    
    print(f"Processing array: {nums}\n")
    
    for i, num in enumerate(nums):
        str_num = str(num)
        digit_count = len(str_num)
        is_even = digit_count % 2 == 0
        
        print(f"Index {i}: num = {num}")
        print(f"  String: '{str_num}'")
        print(f"  Digit count: {digit_count}")
        print(f"  Is even? {is_even}")
        
        if is_even:
            count += 1
            print(f"  ✅ Count incremented to {count}")
        else:
            print(f"  ❌ Not counted (odd digits)")
        print()
    
    print(f"Final count: {count}")
    return count

# Test
findNumbers_verbose([12, 345, 2, 6, 7896])
````

## Approach 2: Mathematical Approach (Count Digits)

### Intuition

Instead of converting to string, we can count digits mathematically by repeatedly dividing by 10.

### Algorithm

```
1. For each number:
     digit_count = 0
     temp = number
     While temp > 0:
         digit_count++
         temp = temp // 10
     If digit_count is even:
         count++
2. Return count
```

### Why This Works

Each division by 10 removes one digit:
- 1234 → 123 → 12 → 1 → 0 (4 divisions = 4 digits)

### Complexity

- **Time**: O(n × d) where d = number of digits
- **Space**: O(1) - no extra space needed

````python
def findNumbers_math(nums):
    """
    Count numbers with even number of digits using math.
    Time: O(n × d), Space: O(1)
    """
    count = 0
    
    for num in nums:
        # Count digits by dividing by 10
        digit_count = 0
        temp = num
        
        while temp > 0:
            digit_count += 1
            temp //= 10
        
        # Check if even
        if digit_count % 2 == 0:
            count += 1
    
    return count

# Test
print("Mathematical approach:")
for nums in test_cases:
    result = findNumbers_math(nums)
    print(f"Input: {nums}")
    print(f"Output: {result}\n")
````

### Step-by-Step Example

Counting digits in 7896:

```
Initial: temp = 7896, digit_count = 0

Iteration 1:
  digit_count = 1
  temp = 7896 // 10 = 789

Iteration 2:
  digit_count = 2
  temp = 789 // 10 = 78

Iteration 3:
  digit_count = 3
  temp = 78 // 10 = 7

Iteration 4:
  digit_count = 4
  temp = 7 // 10 = 0

Loop ends (temp = 0)
Final: digit_count = 4 (even!) ✅
```

````python
def count_digits_verbose(num):
    """Show how digit counting works."""
    print(f"Counting digits in {num}:")
    digit_count = 0
    temp = num
    
    while temp > 0:
        digit_count += 1
        print(f"  Iteration {digit_count}: temp = {temp}, temp // 10 = {temp // 10}")
        temp //= 10
    
    print(f"  Final digit count: {digit_count}")
    print(f"  Is even? {digit_count % 2 == 0}\n")
    return digit_count

# Test
count_digits_verbose(7896)
count_digits_verbose(345)
count_digits_verbose(12)
````

## Approach 3: Using Logarithm

### Intuition

The number of digits in a positive integer `n` is `floor(log10(n)) + 1`.

**Why?**
- log₁₀(100) = 2, so 100 has 2 + 1 = 3 digits ✅
- log₁₀(999) ≈ 2.999, floor = 2, so 999 has 2 + 1 = 3 digits ✅
- log₁₀(1000) = 3, so 1000 has 3 + 1 = 4 digits ✅

### Algorithm

```
1. For each number:
     digit_count = floor(log10(num)) + 1
     If digit_count is even:
         count++
2. Return count
```

### Complexity

- **Time**: O(n) - log operation is O(1)
- **Space**: O(1)

````python
import math

def findNumbers_log(nums):
    """
    Count numbers with even number of digits using logarithm.
    Time: O(n), Space: O(1)
    """
    count = 0
    
    for num in nums:
        # Calculate number of digits using log
        digit_count = math.floor(math.log10(num)) + 1
        
        if digit_count % 2 == 0:
            count += 1
    
    return count

# Test
print("Logarithm approach:")
for nums in test_cases:
    result = findNumbers_log(nums)
    print(f"Input: {nums}")
    print(f"Output: {result}\n")
````

## Approach 4: One-Liner with List Comprehension

### Intuition

Python allows us to write this very concisely using list comprehension and sum.

### Implementation

````python
def findNumbers(nums):
    """
    Pythonic one-liner solution.
    Time: O(n × d), Space: O(1)
    """
    return sum(len(str(num)) % 2 == 0 for num in nums)

# Test
print("One-liner approach:")
for nums in test_cases:
    result = findNumbers(nums)
    print(f"Input: {nums}")
    print(f"Output: {result}\n")
````

### How It Works

```python
sum(len(str(num)) % 2 == 0 for num in nums)
```

Breaking it down:
1. `for num in nums` - iterate through each number
2. `str(num)` - convert to string
3. `len(str(num))` - get digit count
4. `len(str(num)) % 2 == 0` - check if even (returns True/False)
5. `sum(...)` - sum the True values (True = 1, False = 0)

**Example**: `[12, 345, 2]`
- 12: len("12") = 2, 2 % 2 == 0 → True (1)
- 345: len("345") = 3, 3 % 2 == 0 → False (0)
- 2: len("2") = 1, 1 % 2 == 0 → False (0)
- sum([1, 0, 0]) = 1

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity | Readability | Notes |
|----------|----------------|------------------|-------------|-------|
| String Conversion | O(n × d) | O(d) | ⭐⭐⭐⭐⭐ | Simplest, most readable |
| Mathematical | O(n × d) | O(1) | ⭐⭐⭐⭐ | No string allocation |
| Logarithm | O(n) | O(1) | ⭐⭐⭐ | Fastest, but uses floating point |
| One-Liner | O(n × d) | O(1) | ⭐⭐⭐⭐⭐ | Pythonic, concise |

Where:
- n = length of array
- d = average number of digits

### Which to Use?

**For interviews**: String conversion or one-liner (clear and Pythonic)

**For production**: 
- If performance critical: Logarithm approach
- Otherwise: String conversion (most maintainable)

## Common Mistakes

### Mistake 1: Off-by-One Error in Logarithm

**Wrong**:
```python
digit_count = math.floor(math.log10(num))  # Missing +1
```

**Right**:
```python
digit_count = math.floor(math.log10(num)) + 1  # ✅
```

### Mistake 2: Not Handling Single Digit Numbers in Log

For num = 1:
- log₁₀(1) = 0
- floor(0) + 1 = 1 ✅

This works correctly!

### Mistake 3: Using Wrong Modulo Check

**Wrong**:
```python
if digit_count % 2 == 1:  # Checking for odd instead of even
```

**Right**:
```python
if digit_count % 2 == 0:  # ✅ Check for even
```

### Mistake 4: Forgetting to Initialize Counter

**Wrong**:
```python
# count not initialized
for num in nums:
    if len(str(num)) % 2 == 0:
        count += 1  # Error: count not defined
```

**Right**:
```python
count = 0  # ✅ Initialize first
for num in nums:
    if len(str(num)) % 2 == 0:
        count += 1
```
