## Pattern Overview

### What is the Slow-Fast Pointer Pattern?

The **slow-fast pointer pattern** (also called **two-pointer technique**) uses two pointers moving at different speeds through an array:

- **Slow pointer**: Points to the position where we want to write/place elements
- **Fast pointer**: Scans through the array to find elements that meet our criteria

### Why Use It?

✅ **In-place modification**: O(1) space complexity  
✅ **Single pass**: O(n) time complexity  
✅ **No extra arrays**: Modifies the original array  
✅ **Versatile**: Works for many array manipulation problems  

### Common Use Cases

- Moving elements (zeros, duplicates, etc.)
- Removing elements
- Partitioning arrays
- Sorting by certain criteria

## Core Concept

### The Basic Template

```python
def two_pointer_template(arr):
    slow = 0  # Position to write
    
    for fast in range(len(arr)):  # Scan through array
        if condition(arr[fast]):  # Check if element meets criteria
            arr[slow] = arr[fast]  # Place element at slow position
            slow += 1              # Move slow pointer forward
    
    return slow  # Often returns new length or position
```

### How It Works

```
Initial: [3, 0, 1, 0, 2, 0, 4]
         ↑
       slow, fast

Step 1: fast finds 3 (meets criteria)
        arr[slow] = arr[fast] = 3
        slow++
        [3, 0, 1, 0, 2, 0, 4]
            ↑  ↑
          slow fast

Step 2: fast finds 0 (doesn't meet criteria)
        Skip, only fast moves
        [3, 0, 1, 0, 2, 0, 4]
            ↑     ↑
          slow   fast

Step 3: fast finds 1 (meets criteria)
        arr[slow] = arr[fast] = 1
        slow++
        [3, 1, 1, 0, 2, 0, 4]
               ↑     ↑
             slow   fast

... and so on
```

### Key Invariants

1. **Elements before slow**: Already processed and meet criteria
2. **Elements from slow to fast**: Don't meet criteria (or garbage)
3. **Elements after fast**: Not yet processed

```
[valid elements | invalid/garbage | unprocessed]
 0 ... slow-1   | slow ... fast-1 | fast ... n-1
```

## Problem 1: Move Zeros

**LeetCode 283**: Move all zeros to the end while maintaining the relative order of non-zero elements.

### Problem Statement

```
Input:  [0, 1, 0, 3, 12]
Output: [1, 3, 12, 0, 0]
```

### Approach: Two-Pointer

**Strategy**:
1. **Slow pointer**: Position to place next non-zero element
2. **Fast pointer**: Scans for non-zero elements
3. When fast finds non-zero, place it at slow and increment slow
4. Fill remaining positions with zeros

````python
def moveZeroes(nums):
    """
    Move all zeros to the end, maintain order of non-zeros.
    Time: O(n), Space: O(1)
    
    LeetCode 283: Move Zeros
    """
    slow = 0  # Position to place next non-zero
    
    # Move all non-zero elements to the front
    for fast in range(len(nums)):
        if nums[fast] != 0:  # Found non-zero
            nums[slow] = nums[fast]
            slow += 1
    
    # Fill remaining positions with zeros
    for i in range(slow, len(nums)):
        nums[i] = 0
    
    return nums

# Test
test_cases = [
    [0, 1, 0, 3, 12],
    [0],
    [1, 2, 3],
    [0, 0, 1]
]

print("Move Zeros:\n")
for arr in test_cases:
    original = arr.copy()
    result = moveZeroes(arr.copy())
    print(f"Input:  {original}")
    print(f"Output: {result}\n")
````

### Optimized Version: Swap Instead of Overwrite

````python
def moveZeroes_optimized(nums):
    """
    Optimized: Swap elements instead of overwrite + fill.
    Single pass, fewer writes.
    """
    slow = 0
    
    for fast in range(len(nums)):
        if nums[fast] != 0:
            # Swap non-zero with position at slow
            nums[slow], nums[fast] = nums[fast], nums[slow]
            slow += 1
    
    return nums

# Test
print("Move Zeros (Optimized - Swap):\n")
for arr in test_cases:
    original = arr.copy()
    result = moveZeroes_optimized(arr.copy())
    print(f"Input:  {original}")
    print(f"Output: {result}\n")
````

### Detailed Step-by-Step Trace

````python
def moveZeroes_verbose(nums):
    """Verbose version showing each step."""
    print(f"Original: {nums}\n")
    print("="*70)
    print("PHASE 1: Move non-zeros to front")
    print("="*70)
    
    slow = 0
    
    for fast in range(len(nums)):
        print(f"\nStep {fast + 1}:")
        print(f"  fast={fast}, nums[fast]={nums[fast]}")
        print(f"  slow={slow}")
        
        if nums[fast] != 0:
            print(f"  → Non-zero found! Place {nums[fast]} at position {slow}")
            nums[slow] = nums[fast]
            slow += 1
            print(f"  → Array: {nums}, slow now at {slow}")
        else:
            print(f"  → Zero found, skip (only fast moves)")
            print(f"  → Array: {nums}")
    
    print(f"\n{'='*70}")
    print("PHASE 2: Fill remaining with zeros")
    print("="*70)
    print(f"Fill positions {slow} to {len(nums)-1} with zeros\n")
    
    for i in range(slow, len(nums)):
        nums[i] = 0
    
    print(f"Final result: {nums}")
    return nums

# Test
moveZeroes_verbose([0, 1, 0, 3, 12])
````

## Problem 2: Remove Element

**LeetCode 27**: Remove all instances of a value in-place and return the new length.

### Problem Statement

```
Input:  nums = [3, 2, 2, 3], val = 3
Output: 2, nums = [2, 2, _, _]
```

### Approach: Two-Pointer

**Strategy**:
1. **Slow pointer**: Position to place next valid element
2. **Fast pointer**: Scans for elements != val
3. When fast finds valid element, place it at slow and increment slow
4. Return slow (new length)

````python
def removeElement(nums, val):
    """
    Remove all instances of val in-place.
    Time: O(n), Space: O(1)
    
    LeetCode 27: Remove Element
    """
    slow = 0  # Position to place next valid element
    
    for fast in range(len(nums)):
        if nums[fast] != val:  # Found valid element
            nums[slow] = nums[fast]
            slow += 1
    
    return slow  # New length

# Test
test_cases = [
    ([3, 2, 2, 3], 3),
    ([0, 1, 2, 2, 3, 0, 4, 2], 2),
    ([1], 1),
    ([1, 2, 3], 4)
]

print("Remove Element:\n")
for nums, val in test_cases:
    original = nums.copy()
    arr = nums.copy()
    new_length = removeElement(arr, val)
    print(f"Input:  nums={original}, val={val}")
    print(f"Output: length={new_length}, nums={arr[:new_length]} (first {new_length} elements)\n")
````

### Detailed Step-by-Step Trace

````python
def removeElement_verbose(nums, val):
    """Verbose version showing each step."""
    print(f"Original: {nums}, remove value: {val}\n")
    print("="*70)
    
    slow = 0
    
    for fast in range(len(nums)):
        print(f"Step {fast + 1}:")
        print(f"  fast={fast}, nums[fast]={nums[fast]}")
        print(f"  slow={slow}")
        
        if nums[fast] != val:
            print(f"  → {nums[fast]} != {val}, keep it! Place at position {slow}")
            nums[slow] = nums[fast]
            slow += 1
            print(f"  → Array: {nums}, slow now at {slow}")
        else:
            print(f"  → {nums[fast]} == {val}, skip it!")
            print(f"  → Array: {nums}")
        print()
    
    print("="*70)
    print(f"New length: {slow}")
    print(f"Valid elements: {nums[:slow]}")
    print(f"Full array: {nums}")
    return slow

# Test
removeElement_verbose([3, 2, 2, 3], 3)
````

## Problem 3: Remove Duplicates

**LeetCode 26**: Remove duplicates from sorted array in-place.

### Problem Statement

```
Input:  [1, 1, 2, 2, 3]
Output: 3, nums = [1, 2, 3, _, _]
```

### Approach: Two-Pointer

**Strategy**:
1. **Slow pointer**: Position of last unique element
2. **Fast pointer**: Scans for new unique elements
3. When fast finds element different from slow, it's unique
4. Increment slow and place new unique element

````python
def removeDuplicates(nums):
    """
    Remove duplicates from sorted array in-place.
    Time: O(n), Space: O(1)
    
    LeetCode 26: Remove Duplicates from Sorted Array
    """
    if not nums:
        return 0
    
    slow = 0  # Position of last unique element
    
    for fast in range(1, len(nums)):
        if nums[fast] != nums[slow]:  # Found new unique element
            slow += 1
            nums[slow] = nums[fast]
    
    return slow + 1  # Length (slow is index, so +1)

# Test
test_cases = [
    [1, 1, 2],
    [0, 0, 1, 1, 1, 2, 2, 3, 3, 4],
    [1, 2, 3],
    [1, 1, 1, 1]
]

print("Remove Duplicates:\n")
for nums in test_cases:
    original = nums.copy()
    arr = nums.copy()
    new_length = removeDuplicates(arr)
    print(f"Input:  {original}")
    print(f"Output: length={new_length}, nums={arr[:new_length]}\n")
````

### Detailed Step-by-Step Trace

````python
def removeDuplicates_verbose(nums):
    """Verbose version showing each step."""
    if not nums:
        return 0
    
    print(f"Original (sorted): {nums}\n")
    print("="*70)
    
    slow = 0
    print(f"Initial: slow=0, nums[slow]={nums[slow]} (first element is always unique)\n")
    
    for fast in range(1, len(nums)):
        print(f"Step {fast}:")
        print(f"  fast={fast}, nums[fast]={nums[fast]}")
        print(f"  slow={slow}, nums[slow]={nums[slow]}")
        
        if nums[fast] != nums[slow]:
            print(f"  → {nums[fast]} != {nums[slow]}, new unique element!")
            slow += 1
            nums[slow] = nums[fast]
            print(f"  → Increment slow to {slow}, place {nums[fast]} there")
            print(f"  → Array: {nums}")
        else:
            print(f"  → {nums[fast]} == {nums[slow]}, duplicate! Skip")
            print(f"  → Array: {nums}")
        print()
    
    print("="*70)
    print(f"New length: {slow + 1}")
    print(f"Unique elements: {nums[:slow + 1]}")
    return slow + 1

# Test
removeDuplicates_verbose([1, 1, 2, 2, 3])
````

## Problem 4: Sort Array by Parity

**LeetCode 905**: Move all even numbers to the front, odd numbers to the back.

### Problem Statement

```
Input:  [3, 1, 2, 4]
Output: [2, 4, 3, 1] (or any arrangement with evens first)
```

### Approach: Two-Pointer

**Strategy**:
1. **Slow pointer**: Position to place next even number
2. **Fast pointer**: Scans for even numbers
3. When fast finds even, swap with slow and increment slow
4. Result: evens before slow, odds after slow

````python
def sortArrayByParity(nums):
    """
    Move all even numbers to front, odd to back.
    Time: O(n), Space: O(1)
    
    LeetCode 905: Sort Array by Parity
    """
    slow = 0  # Position to place next even number
    
    for fast in range(len(nums)):
        if nums[fast] % 2 == 0:  # Found even number
            nums[slow], nums[fast] = nums[fast], nums[slow]
            slow += 1
    
    return nums

# Test
test_cases = [
    [3, 1, 2, 4],
    [0],
    [1, 2, 3, 4, 5, 6],
    [2, 4, 6, 8]
]

print("Sort Array by Parity:\n")
for nums in test_cases:
    original = nums.copy()
    result = sortArrayByParity(nums.copy())
    print(f"Input:  {original}")
    print(f"Output: {result}")
    
    # Verify: count evens before first odd
    first_odd_idx = next((i for i, x in enumerate(result) if x % 2 == 1), len(result))
    evens_first = all(x % 2 == 0 for x in result[:first_odd_idx])
    odds_last = all(x % 2 == 1 for x in result[first_odd_idx:])
    print(f"Valid: {evens_first and odds_last}\n")
````

### Detailed Step-by-Step Trace

````python
def sortArrayByParity_verbose(nums):
    """Verbose version showing each step."""
    print(f"Original: {nums}\n")
    print("="*70)
    print("Goal: Move all EVEN numbers to front, ODD to back")
    print("="*70 + "\n")
    
    slow = 0
    
    for fast in range(len(nums)):
        print(f"Step {fast + 1}:")
        print(f"  fast={fast}, nums[fast]={nums[fast]} ({'EVEN' if nums[fast] % 2 == 0 else 'ODD'})")
        print(f"  slow={slow}")
        
        if nums[fast] % 2 == 0:
            print(f"  → Even number found! Swap with position {slow}")
            print(f"  → Before swap: {nums}")
            nums[slow], nums[fast] = nums[fast], nums[slow]
            print(f"  → After swap:  {nums}")
            slow += 1
            print(f"  → slow now at {slow}")
        else:
            print(f"  → Odd number, skip (only fast moves)")
            print(f"  → Array: {nums}")
        print()
    
    print("="*70)
    print(f"Final result: {nums}")
    print(f"Even numbers (0 to {slow-1}): {nums[:slow]}")
    print(f"Odd numbers ({slow} to end): {nums[slow:]}")
    return nums

# Test
sortArrayByParity_verbose([3, 1, 2, 4])
````

## Pattern Variations

### Variation 1: Overwrite + Fill (Move Zeros)

```python
slow = 0
for fast in range(len(nums)):
    if condition(nums[fast]):
        nums[slow] = nums[fast]  # Overwrite
        slow += 1
# Fill remaining with default value
for i in range(slow, len(nums)):
    nums[i] = default_value
```

**Use when**: Need to fill remaining positions with specific value

### Variation 2: Swap (Sort by Parity)

```python
slow = 0
for fast in range(len(nums)):
    if condition(nums[fast]):
        nums[slow], nums[fast] = nums[fast], nums[slow]  # Swap
        slow += 1
```

**Use when**: Need to preserve all elements, just reorder them

### Variation 3: Return New Length (Remove Element)

```python
slow = 0
for fast in range(len(nums)):
    if condition(nums[fast]):
        nums[slow] = nums[fast]
        slow += 1
return slow  # New length
```

**Use when**: Need to return count of valid elements

### Variation 4: Compare with Previous (Remove Duplicates)

```python
slow = 0
for fast in range(1, len(nums)):  # Start from 1
    if nums[fast] != nums[slow]:  # Compare with slow
        slow += 1
        nums[slow] = nums[fast]
return slow + 1
```

**Use when**: Need to compare with previously placed element

## When to Use This Pattern

### Identifying the Pattern

Use slow-fast pointer when you see:

✅ **In-place modification** required  
✅ **O(1) space** constraint  
✅ **Filtering/partitioning** elements  
✅ **Maintaining relative order** (usually)  
✅ **Single pass** through array  

### Common Problem Keywords

- "Move all X to the end/front"
- "Remove all instances of X"
- "Remove duplicates"
- "Partition array by condition"
- "In-place"
- "O(1) extra space"

### Decision Tree

```
Need to modify array in-place?
  |
  ├─ YES → Can use two pointers?
  │         |
  │         ├─ Filtering elements → Slow-Fast Pointer
  │         ├─ Partitioning → Slow-Fast Pointer
  │         ├─ Two ends → Left-Right Pointer
  │         └─ Sliding window → Window Pointer
  │
  └─ NO → Can use extra space
            └─ Use new array/set/map
```

## Comparison of All Four Problems

| Problem | Condition | Action | Post-Process | Return |
|---------|-----------|--------|--------------|--------|
| **Move Zeros** | `nums[fast] != 0` | Overwrite | Fill with 0s | Array |
| **Remove Element** | `nums[fast] != val` | Overwrite | None | Length |
| **Remove Duplicates** | `nums[fast] != nums[slow]` | Overwrite | None | Length |
| **Sort by Parity** | `nums[fast] % 2 == 0` | Swap | None | Array |

### Visual Comparison

````python
def compare_all_patterns():
    """Compare all four patterns side by side."""
    
    # Test array
    arr = [0, 1, 0, 3, 12]
    
    print("Original array:", arr)
    print("\n" + "="*70)
    
    # Move Zeros
    result1 = moveZeroes(arr.copy())
    print("Move Zeros:        ", result1)
    
    # Remove Element (remove 0s)
    arr2 = arr.copy()
    length2 = removeElement(arr2, 0)
    print(f"Remove Element (0): {arr2[:length2]} (length={length2})")
    
    # Remove Duplicates (need sorted array)
    arr3 = [0, 0, 1, 1, 2, 3, 3]
    length3 = removeDuplicates(arr3)
    print(f"Remove Duplicates:  {arr3[:length3]} (from sorted {[0, 0, 1, 1, 2, 3, 3]})")
    
    # Sort by Parity
    result4 = sortArrayByParity(arr.copy())
    print("Sort by Parity:    ", result4)
    
    print("\n" + "="*70)
    print("\nAll use the same slow-fast pointer pattern!")
    print("Difference is in the CONDITION and ACTION")

compare_all_patterns()
````

## Key Takeaways

### The Slow-Fast Pointer Pattern

1. **Two pointers, different speeds**:
   - Slow: Position to write
   - Fast: Scans for valid elements

2. **Core invariant**:
   ```
   [valid elements | invalid/garbage | unprocessed]
    0 ... slow-1   | slow ... fast-1 | fast ... n-1
   ```

3. **Time complexity**: O(n) - single pass

4. **Space complexity**: O(1) - in-place

5. **When to use**:
   - In-place array modification
   - Filtering/partitioning elements
   - O(1) space constraint

### Common Variations

| Variation | When to Use | Example |
|-----------|-------------|----------|
| **Overwrite + Fill** | Need specific value in remaining positions | Move Zeros |
| **Swap** | Preserve all elements, just reorder | Sort by Parity |
| **Return Length** | Need count of valid elements | Remove Element |
| **Compare Previous** | Need to check against last valid element | Remove Duplicates |

### Problem-Solving Steps

1. **Identify**: Does problem need in-place O(1) space?
2. **Define condition**: What makes an element valid?
3. **Choose action**: Overwrite, swap, or compare?
4. **Handle post-process**: Fill remaining? Return length?
5. **Verify invariants**: Elements before slow are valid

### Related Patterns

- **Left-Right Pointer**: Two pointers from both ends (Two Sum II, Container With Most Water)
- **Sliding Window**: Two pointers defining a window (Longest Substring, Max Sum Subarray)
- **Fast-Slow Cycle Detection**: Different speeds to detect cycles (Linked List Cycle)
