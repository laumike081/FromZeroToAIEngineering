## Overview

### Sorting Algorithm Comparison

| Algorithm | Time (Best) | Time (Average) | Time (Worst) | Space | Stable | In-Place |
|-----------|-------------|----------------|--------------|-------|--------|----------|
| **Insertion Sort** | O(n) | O(n²) | O(n²) | O(1) | ✅ Yes | ✅ Yes |
| **Merge Sort** | O(n log n) | O(n log n) | O(n log n) | O(n) | ✅ Yes | ❌ No |
| **Quick Sort** | O(n log n) | O(n log n) | O(n²) | O(log n) | ❌ No | ✅ Yes |
| **Bucket Sort** | O(n + k) | O(n + k) | O(n²) | O(n + k) | ✅ Yes | ❌ No |

**Key Terms:**
- **Stable**: Maintains relative order of equal elements
- **In-Place**: Requires O(1) extra space (excluding recursion stack)
- **k**: Number of buckets (for bucket sort)

## Insertion Sort

### Concept

Insertion sort builds the final sorted array one item at a time. It's like sorting playing cards in your hands:
- Pick up one card at a time
- Insert it into its correct position among the already sorted cards
- Shift other cards as needed

### How It Works

```
Initial: [5, 2, 4, 6, 1, 3]
         ↑
       sorted

Step 1: [5, 2, 4, 6, 1, 3]  → Insert 2
        [2, 5, 4, 6, 1, 3]  (2 < 5, shift 5 right)
         ↑  ↑
       sorted

Step 2: [2, 5, 4, 6, 1, 3]  → Insert 4
        [2, 4, 5, 6, 1, 3]  (4 < 5, shift 5 right)
         ↑  ↑  ↑
       sorted

Step 3: [2, 4, 5, 6, 1, 3]  → Insert 6
        [2, 4, 5, 6, 1, 3]  (6 > 5, no shift)
         ↑  ↑  ↑  ↑
       sorted

Step 4: [2, 4, 5, 6, 1, 3]  → Insert 1
        [1, 2, 4, 5, 6, 3]  (1 < all, shift all right)
         ↑  ↑  ↑  ↑  ↑
       sorted

Step 5: [1, 2, 4, 5, 6, 3]  → Insert 3
        [1, 2, 3, 4, 5, 6]  (3 between 2 and 4)
         ↑  ↑  ↑  ↑  ↑  ↑
       sorted (done!)
```

### Algorithm

````python
def insertion_sort(arr):
    """
    Insertion Sort: Build sorted array one element at a time.
    
    Time Complexity:
        - Best: O(n) - already sorted
        - Average: O(n²)
        - Worst: O(n²) - reverse sorted
    Space Complexity: O(1) - in-place
    Stable: Yes
    """
    n = len(arr)
    
    # Start from second element (index 1)
    for i in range(1, n):
        key = arr[i]  # Element to be inserted
        j = i - 1     # Start comparing with previous elements
        
        # Shift elements greater than key to the right
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        
        # Insert key at correct position
        arr[j + 1] = key
    
    return arr

# Test
test_arr = [64, 34, 25, 12, 22, 11, 90]
print(f"Original: {test_arr}")
print(f"Sorted:   {insertion_sort(test_arr.copy())}")
````

### Detailed Step-by-Step Trace

````python
def insertion_sort_verbose(arr):
    """Insertion sort with detailed step-by-step output."""
    n = len(arr)
    print(f"Starting array: {arr}\n")
    
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        
        print(f"Step {i}: Inserting {key} (index {i})")
        print(f"  Current array: {arr}")
        print(f"  Sorted portion: {arr[:i]}")
        
        shifts = 0
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            shifts += 1
            j -= 1
        
        arr[j + 1] = key
        
        if shifts > 0:
            print(f"  Shifted {shifts} element(s) right")
            print(f"  Inserted {key} at index {j + 1}")
        else:
            print(f"  {key} already in correct position")
        
        print(f"  Result: {arr}\n")
    
    print(f"Final sorted array: {arr}")
    return arr

# Test
insertion_sort_verbose([5, 2, 4, 6, 1, 3])
````

### When to Use Insertion Sort

**Best Use Cases:**
- ✅ Small datasets (n < 50)
- ✅ Nearly sorted data (best case O(n))
- ✅ Online sorting (data arrives one at a time)
- ✅ When simplicity is important
- ✅ When stability is required

**Avoid When:**
- ❌ Large datasets
- ❌ Reverse sorted data

### Complexity Analysis

**Time Complexity:**
- **Best Case: O(n)** - Array is already sorted, only n-1 comparisons
- **Average Case: O(n²)** - Random order, ~n²/4 comparisons
- **Worst Case: O(n²)** - Reverse sorted, n(n-1)/2 comparisons

**Space Complexity: O(1)** - Only uses a constant amount of extra space

**Why O(n²)?**
- Outer loop runs n times
- Inner loop runs up to i times for each i
- Total: 1 + 2 + 3 + ... + n = n(n+1)/2 = O(n²)

## Merge Sort

### Concept

Merge sort is a **divide-and-conquer** algorithm:
1. **Divide**: Split array into two halves
2. **Conquer**: Recursively sort each half
3. **Combine**: Merge the two sorted halves

### How It Works

```
Original: [38, 27, 43, 3, 9, 82, 10]

DIVIDE:
                [38, 27, 43, 3, 9, 82, 10]
                /                        \
        [38, 27, 43, 3]              [9, 82, 10]
         /           \                /         \
    [38, 27]      [43, 3]         [9, 82]      [10]
     /    \        /    \          /    \
   [38]  [27]    [43]  [3]       [9]  [82]

MERGE:
   [38]  [27]    [43]  [3]       [9]  [82]     [10]
     \    /        \    /          \    /
    [27, 38]      [3, 43]         [9, 82]      [10]
         \           /                \         /
        [3, 27, 38, 43]              [9, 10, 82]
                \                      /
              [3, 9, 10, 27, 38, 43, 82]
```

### Algorithm

````python
def merge_sort(arr):
    """
    Merge Sort: Divide and conquer sorting algorithm.
    
    Time Complexity: O(n log n) - always
    Space Complexity: O(n) - requires extra space for merging
    Stable: Yes
    """
    # Base case: array of size 0 or 1 is already sorted
    if len(arr) <= 1:
        return arr
    
    # Divide: find middle point
    mid = len(arr) // 2
    
    # Conquer: recursively sort both halves
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    
    # Combine: merge the sorted halves
    return merge(left, right)

def merge(left, right):
    """
    Merge two sorted arrays into one sorted array.
    """
    result = []
    i = j = 0
    
    # Compare elements from left and right, add smaller to result
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:  # <= for stability
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    # Add remaining elements (one array will be exhausted)
    result.extend(left[i:])
    result.extend(right[j:])
    
    return result

# Test
test_arr = [38, 27, 43, 3, 9, 82, 10]
print(f"Original: {test_arr}")
print(f"Sorted:   {merge_sort(test_arr)}")
````

### Detailed Step-by-Step Trace

````python
def merge_sort_verbose(arr, depth=0):
    """Merge sort with detailed recursion trace."""
    indent = "  " * depth
    
    print(f"{indent}merge_sort({arr})")
    
    if len(arr) <= 1:
        print(f"{indent}  → Base case, return {arr}")
        return arr
    
    mid = len(arr) // 2
    print(f"{indent}  Divide at index {mid}")
    print(f"{indent}  Left:  {arr[:mid]}")
    print(f"{indent}  Right: {arr[mid:]}")
    
    left = merge_sort_verbose(arr[:mid], depth + 1)
    right = merge_sort_verbose(arr[mid:], depth + 1)
    
    result = merge_verbose(left, right, depth)
    print(f"{indent}  → Merged result: {result}\n")
    
    return result

def merge_verbose(left, right, depth):
    """Merge with detailed output."""
    indent = "  " * depth
    print(f"{indent}  Merging {left} and {right}")
    
    result = []
    i = j = 0
    
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            print(f"{indent}    {left[i]} <= {right[j]}, take {left[i]} from left")
            i += 1
        else:
            result.append(right[j])
            print(f"{indent}    {left[i]} > {right[j]}, take {right[j]} from right")
            j += 1
    
    if i < len(left):
        print(f"{indent}    Add remaining from left: {left[i:]}")
    if j < len(right):
        print(f"{indent}    Add remaining from right: {right[j:]}")
    
    result.extend(left[i:])
    result.extend(right[j:])
    
    return result

# Test
print("Merge Sort Trace:\n")
merge_sort_verbose([38, 27, 43, 3])
````

### When to Use Merge Sort

**Best Use Cases:**
- ✅ Guaranteed O(n log n) performance needed
- ✅ Stability is required
- ✅ Sorting linked lists (no random access needed)
- ✅ External sorting (data doesn't fit in memory)
- ✅ When worst-case performance matters

**Avoid When:**
- ❌ Memory is limited (requires O(n) extra space)
- ❌ Small datasets (overhead not worth it)

### Complexity Analysis

**Time Complexity: O(n log n)** - always
- **Divide**: O(log n) levels of recursion (halving each time)
- **Merge**: O(n) work at each level (merge all elements)
- **Total**: O(n) × O(log n) = O(n log n)

**Space Complexity: O(n)**
- Requires extra space for temporary arrays during merging
- Recursion stack: O(log n)
- Total: O(n) for merge arrays

**Why O(n log n)?**
```
Level 0:           [n elements]           → n work
Level 1:     [n/2]         [n/2]          → n work
Level 2:  [n/4] [n/4]   [n/4] [n/4]      → n work
...
Level log n: [1] [1] [1] ... [1]          → n work

Total levels: log n
Work per level: n
Total: n × log n = O(n log n)
```

## Quick Sort

### Concept

Quick sort is a **divide-and-conquer** algorithm using a **pivot**:
1. **Choose a pivot** element
2. **Partition**: Rearrange so elements < pivot are on left, elements > pivot are on right
3. **Recursively** sort left and right partitions

### How It Works

```
Original: [10, 80, 30, 90, 40, 50, 70]
          Choose pivot = 70 (last element)

PARTITION:
[10, 80, 30, 90, 40, 50, 70]
 ↑                       ↑
 i                    pivot

Compare each element with pivot:
10 < 70 → [10, 80, 30, 90, 40, 50, 70]  (10 stays)
80 > 70 → [10, 80, 30, 90, 40, 50, 70]  (80 stays)
30 < 70 → [10, 30, 80, 90, 40, 50, 70]  (swap 30 and 80)
90 > 70 → [10, 30, 80, 90, 40, 50, 70]  (90 stays)
40 < 70 → [10, 30, 40, 90, 80, 50, 70]  (swap 40 and 80)
50 < 70 → [10, 30, 40, 50, 80, 90, 70]  (swap 50 and 90)

Place pivot: [10, 30, 40, 50, 70, 90, 80]
                           ↑
                        pivot in
                     correct position

Now recursively sort:
Left:  [10, 30, 40, 50]
Right: [90, 80]
```

### Algorithm

````python
def quick_sort(arr):
    """
    Quick Sort: Divide and conquer using pivot partitioning.
    
    Time Complexity:
        - Best: O(n log n) - balanced partitions
        - Average: O(n log n)
        - Worst: O(n²) - already sorted with bad pivot choice
    Space Complexity: O(log n) - recursion stack
    Stable: No (standard implementation)
    """
    if len(arr) <= 1:
        return arr
    
    # Choose pivot (middle element for better average case)
    pivot = arr[len(arr) // 2]
    
    # Partition into three parts
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    
    # Recursively sort and combine
    return quick_sort(left) + middle + quick_sort(right)

# Test
test_arr = [10, 80, 30, 90, 40, 50, 70]
print(f"Original: {test_arr}")
print(f"Sorted:   {quick_sort(test_arr)}")
````

### In-Place Quick Sort (More Efficient)

````python
def quick_sort_inplace(arr, low=0, high=None):
    """
    In-place quick sort using Lomuto partition scheme.
    More space-efficient than the simple version.
    """
    if high is None:
        high = len(arr) - 1
    
    if low < high:
        # Partition and get pivot index
        pivot_idx = partition(arr, low, high)
        
        # Recursively sort left and right partitions
        quick_sort_inplace(arr, low, pivot_idx - 1)
        quick_sort_inplace(arr, pivot_idx + 1, high)
    
    return arr

def partition(arr, low, high):
    """
    Lomuto partition: Choose last element as pivot.
    Rearrange so elements <= pivot are on left.
    """
    pivot = arr[high]  # Choose last element as pivot
    i = low - 1        # Index of smaller element
    
    for j in range(low, high):
        # If current element is smaller than or equal to pivot
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]  # Swap
    
    # Place pivot in correct position
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    
    return i + 1  # Return pivot index

# Test
test_arr = [10, 80, 30, 90, 40, 50, 70]
print(f"Original: {test_arr}")
print(f"Sorted:   {quick_sort_inplace(test_arr.copy())}")
````

### Detailed Step-by-Step Trace

````python
def quick_sort_verbose(arr, depth=0):
    """Quick sort with detailed recursion trace."""
    indent = "  " * depth
    
    print(f"{indent}quick_sort({arr})")
    
    if len(arr) <= 1:
        print(f"{indent}  → Base case, return {arr}")
        return arr
    
    pivot = arr[len(arr) // 2]
    print(f"{indent}  Pivot: {pivot}")
    
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    
    print(f"{indent}  Left (< {pivot}):  {left}")
    print(f"{indent}  Middle (= {pivot}): {middle}")
    print(f"{indent}  Right (> {pivot}):  {right}\n")
    
    sorted_left = quick_sort_verbose(left, depth + 1) if left else []
    sorted_right = quick_sort_verbose(right, depth + 1) if right else []
    
    result = sorted_left + middle + sorted_right
    print(f"{indent}  → Combined: {result}\n")
    
    return result

# Test
print("Quick Sort Trace:\n")
quick_sort_verbose([10, 80, 30, 90, 40])
````

### When to Use Quick Sort

**Best Use Cases:**
- ✅ General-purpose sorting (fastest in practice)
- ✅ Large datasets
- ✅ When average-case performance matters
- ✅ In-place sorting needed
- ✅ Cache-friendly (good locality of reference)

**Avoid When:**
- ❌ Worst-case O(n²) is unacceptable
- ❌ Stability is required
- ❌ Already sorted data with bad pivot choice

### Complexity Analysis

**Time Complexity:**
- **Best Case: O(n log n)** - Balanced partitions (pivot is median)
- **Average Case: O(n log n)** - Random pivot choices
- **Worst Case: O(n²)** - Unbalanced partitions (already sorted, bad pivot)

**Space Complexity:**
- **In-place version: O(log n)** - Recursion stack for balanced partitions
- **Simple version: O(n)** - Creates new arrays

**Why O(n log n) on average?**
```
Best case (balanced):
Level 0:           [n elements]           → n work
Level 1:     [n/2]         [n/2]          → n work
Level 2:  [n/4] [n/4]   [n/4] [n/4]      → n work
...
Total: log n levels × n work = O(n log n)

Worst case (unbalanced):
Level 0: [n]                              → n work
Level 1: [n-1]                            → n-1 work
Level 2: [n-2]                            → n-2 work
...
Total: n + (n-1) + (n-2) + ... + 1 = O(n²)
```

### Pivot Selection Strategies

1. **First element**: Simple but bad for sorted data
2. **Last element**: Simple but bad for sorted data
3. **Middle element**: Better for partially sorted data
4. **Random element**: Good average case
5. **Median-of-three**: Choose median of first, middle, last (best in practice)

## Bucket Sort

### Concept

Bucket sort is a **distribution sort** that works by:
1. **Distribute** elements into buckets based on their value
2. **Sort** each bucket individually (using another algorithm)
3. **Concatenate** all sorted buckets

**Key Idea**: If elements are uniformly distributed, each bucket will have few elements, making sorting fast.

### How It Works

```
Input: [0.78, 0.17, 0.39, 0.26, 0.72, 0.94, 0.21, 0.12, 0.23, 0.68]
Range: [0, 1)
Buckets: 5

DISTRIBUTE (bucket = floor(value × num_buckets)):
Bucket 0 [0.0-0.2): [0.17, 0.12]
Bucket 1 [0.2-0.4): [0.39, 0.26, 0.21, 0.23]
Bucket 2 [0.4-0.6): []
Bucket 3 [0.6-0.8): [0.78, 0.72, 0.68]
Bucket 4 [0.8-1.0): [0.94]

SORT each bucket:
Bucket 0: [0.12, 0.17]
Bucket 1: [0.21, 0.23, 0.26, 0.39]
Bucket 2: []
Bucket 3: [0.68, 0.72, 0.78]
Bucket 4: [0.94]

CONCATENATE:
[0.12, 0.17, 0.21, 0.23, 0.26, 0.39, 0.68, 0.72, 0.78, 0.94]
```

### Algorithm

````python
def bucket_sort(arr, num_buckets=10):
    """
    Bucket Sort: Distribute elements into buckets, sort each, concatenate.
    
    Best for: Uniformly distributed data in a known range
    
    Time Complexity:
        - Best: O(n + k) - uniform distribution
        - Average: O(n + k)
        - Worst: O(n²) - all elements in one bucket
    Space Complexity: O(n + k) - buckets and output
    Stable: Yes (if underlying sort is stable)
    
    Args:
        arr: Array of numbers to sort
        num_buckets: Number of buckets to use
    """
    if len(arr) == 0:
        return arr
    
    # Find min and max to determine range
    min_val = min(arr)
    max_val = max(arr)
    
    # Handle edge case where all elements are the same
    if min_val == max_val:
        return arr
    
    # Create empty buckets
    buckets = [[] for _ in range(num_buckets)]
    
    # Distribute elements into buckets
    bucket_range = (max_val - min_val) / num_buckets
    
    for num in arr:
        # Calculate bucket index
        index = int((num - min_val) / bucket_range)
        # Handle edge case for max value
        if index == num_buckets:
            index -= 1
        buckets[index].append(num)
    
    # Sort each bucket (using insertion sort for small buckets)
    for i in range(num_buckets):
        buckets[i] = insertion_sort(buckets[i])
    
    # Concatenate all buckets
    result = []
    for bucket in buckets:
        result.extend(bucket)
    
    return result

# Helper function for bucket sort
def insertion_sort(arr):
    """Simple insertion sort for sorting individual buckets."""
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

# Test with floating point numbers
test_arr = [0.78, 0.17, 0.39, 0.26, 0.72, 0.94, 0.21, 0.12, 0.23, 0.68]
print(f"Original: {test_arr}")
print(f"Sorted:   {bucket_sort(test_arr, num_buckets=5)}")

# Test with integers
test_arr2 = [29, 25, 3, 49, 9, 37, 21, 43]
print(f"\nOriginal: {test_arr2}")
print(f"Sorted:   {bucket_sort(test_arr2, num_buckets=5)}")
````

### Detailed Step-by-Step Trace

````python
def bucket_sort_verbose(arr, num_buckets=5):
    """Bucket sort with detailed step-by-step output."""
    print(f"Input array: {arr}")
    print(f"Number of buckets: {num_buckets}\n")
    
    if len(arr) == 0:
        return arr
    
    min_val = min(arr)
    max_val = max(arr)
    
    print(f"Range: [{min_val}, {max_val}]")
    
    if min_val == max_val:
        print("All elements are the same, already sorted!")
        return arr
    
    buckets = [[] for _ in range(num_buckets)]
    bucket_range = (max_val - min_val) / num_buckets
    
    print(f"Bucket range: {bucket_range:.4f}\n")
    print("="*60)
    print("STEP 1: DISTRIBUTE INTO BUCKETS")
    print("="*60)
    
    for num in arr:
        index = int((num - min_val) / bucket_range)
        if index == num_buckets:
            index -= 1
        
        bucket_start = min_val + index * bucket_range
        bucket_end = min_val + (index + 1) * bucket_range
        
        print(f"  {num:.2f} → Bucket {index} [{bucket_start:.2f}, {bucket_end:.2f})")
        buckets[index].append(num)
    
    print(f"\n{'='*60}")
    print("BUCKETS AFTER DISTRIBUTION:")
    print("="*60)
    for i, bucket in enumerate(buckets):
        bucket_start = min_val + i * bucket_range
        bucket_end = min_val + (i + 1) * bucket_range
        print(f"Bucket {i} [{bucket_start:.2f}, {bucket_end:.2f}): {bucket}")
    
    print(f"\n{'='*60}")
    print("STEP 2: SORT EACH BUCKET")
    print("="*60)
    
    for i in range(num_buckets):
        if buckets[i]:
            print(f"Bucket {i} before: {buckets[i]}")
            buckets[i] = insertion_sort(buckets[i])
            print(f"Bucket {i} after:  {buckets[i]}\n")
    
    print("="*60)
    print("STEP 3: CONCATENATE BUCKETS")
    print("="*60)
    
    result = []
    for i, bucket in enumerate(buckets):
        if bucket:
            print(f"Adding Bucket {i}: {bucket}")
            result.extend(bucket)
    
    print(f"\nFinal sorted array: {result}")
    return result

# Test
bucket_sort_verbose([0.78, 0.17, 0.39, 0.26, 0.72], num_buckets=3)
````

### When to Use Bucket Sort

**Best Use Cases:**
- ✅ Uniformly distributed data
- ✅ Data in a known range
- ✅ Floating-point numbers in [0, 1)
- ✅ When O(n) sorting is needed
- ✅ External sorting with limited memory

**Avoid When:**
- ❌ Data is not uniformly distributed
- ❌ Range is unknown or very large
- ❌ Data clusters in few buckets

### Complexity Analysis

**Time Complexity:**
- **Best Case: O(n + k)** - Uniform distribution, k buckets
  - O(n) to distribute into buckets
  - O(k) to sort k buckets with ~1 element each
  - O(n) to concatenate

- **Average Case: O(n + k)** - Assuming uniform distribution
  - Each bucket has ~n/k elements
  - Sorting each bucket: O((n/k)²)
  - Total for k buckets: k × O((n/k)²) = O(n²/k)
  - If k = n, then O(n)

- **Worst Case: O(n²)** - All elements in one bucket
  - Degenerates to insertion sort

**Space Complexity: O(n + k)**
- O(k) for k buckets
- O(n) for storing elements

### Choosing Number of Buckets

**Rule of thumb**: Use k = n buckets for best performance
- Too few buckets: More elements per bucket, slower sorting
- Too many buckets: More overhead, wasted space
- Optimal: k ≈ n gives O(n) expected time

### Variants

1. **Radix Sort**: Bucket sort on individual digits
2. **Counting Sort**: Special case where buckets represent distinct values
3. **Histogram Sort**: Bucket sort with frequency counting

## Comparison and When to Use Each

### Performance Comparison

````python
import time
import random

def compare_sorting_algorithms(size=1000):
    """Compare performance of different sorting algorithms."""
    # Generate random data
    random.seed(42)
    data = [random.random() for _ in range(size)]
    
    algorithms = [
        ("Insertion Sort", insertion_sort),
        ("Merge Sort", merge_sort),
        ("Quick Sort", quick_sort),
        ("Bucket Sort", lambda arr: bucket_sort(arr, num_buckets=int(len(arr)**0.5))),
        ("Python sorted()", sorted)
    ]
    
    print(f"Comparing sorting algorithms on {size} random elements:\n")
    print(f"{'Algorithm':<20} {'Time (seconds)':<15} {'Speedup vs Insertion'}")
    print("="*60)
    
    baseline_time = None
    
    for name, func in algorithms:
        test_data = data.copy()
        
        start = time.time()
        result = func(test_data)
        elapsed = time.time() - start
        
        if baseline_time is None:
            baseline_time = elapsed
            speedup = "1.00x (baseline)"
        else:
            speedup = f"{baseline_time/elapsed:.2f}x faster"
        
        print(f"{name:<20} {elapsed:<15.6f} {speedup}")

# Test with different sizes
for size in [100, 1000, 5000]:
    compare_sorting_algorithms(size)
    print()
````

## Key Takeaways

### Insertion Sort
- ✅ **Use when**: Small datasets, nearly sorted data, simplicity needed
- 📊 **Complexity**: O(n²) average, O(n) best case
- 💡 **Key insight**: Like sorting cards in your hand
- ⚡ **Adaptive**: Runs in O(n) on sorted data

### Merge Sort
- ✅ **Use when**: Guaranteed O(n log n) needed, stability required
- 📊 **Complexity**: Always O(n log n)
- 💡 **Key insight**: Divide and conquer with merging
- 🎯 **Predictable**: No worst-case degradation

### Quick Sort
- ✅ **Use when**: General purpose, large datasets, in-place needed
- 📊 **Complexity**: O(n log n) average, O(n²) worst
- 💡 **Key insight**: Partition around pivot
- ⚡ **Fastest**: In practice for random data

### Bucket Sort
- ✅ **Use when**: Uniformly distributed data in known range
- 📊 **Complexity**: O(n + k) average, O(n²) worst
- 💡 **Key insight**: Distribute then sort
- 🚀 **Linear time**: When k = O(n) and uniform distribution

### General Principles

1. **No single best algorithm**: Choice depends on data characteristics
2. **Comparison sorts**: Lower bound is O(n log n)
3. **Non-comparison sorts**: Can beat O(n log n) with assumptions (bucket sort)
4. **Trade-offs**: Time vs space, stability vs performance
5. **In practice**: Use Python's `sorted()` (Timsort) unless you have specific needs

### Python's Timsort

Python's built-in `sorted()` and `.sort()` use **Timsort**:
- Hybrid of merge sort and insertion sort
- O(n log n) worst case
- Adaptive: O(n) on nearly sorted data
- Stable
- Highly optimized for real-world data

**When to implement your own:**
- Learning purposes
- Specific constraints (e.g., in-place, limited memory)
- Special data characteristics (e.g., uniform distribution for bucket sort)
- Embedded systems or performance-critical code
