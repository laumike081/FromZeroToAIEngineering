## Introduction

Stacks and Queues are fundamental data structures that form the backbone of many algorithms and systems. They are linear data structures that follow specific ordering principles:

- **Stack**: LIFO (Last In, First Out)
- **Queue**: FIFO (First In, First Out)

These simple concepts have powerful applications in computer science, from function calls to task scheduling.

## Stacks

### Definition & Properties

A **Stack** is a linear data structure that follows the **LIFO** (Last In, First Out) principle. Think of it like a stack of plates - you can only add or remove from the top.

**Key Properties:**
- Elements are added and removed from the same end (top)
- Only the top element is accessible at any time
- Order of elements is reversed when removed

### Visual Representation

```
Top -> | 5 |  (push 5)
      | 4 |  (push 4)
      | 3 |  (push 3)
      | 2 |  (push 2)
      | 1 |  (push 1)
      +---+
      Bottom

pop() -> removes 5 (last pushed)
```

### Implementation

````python
class Stack:
    """Stack implementation using Python list"""
    
    def __init__(self):
        self.items = []
    
    def push(self, item):
        """Add item to top of stack"""
        self.items.append(item)
    
    def pop(self):
        """Remove and return top item"""
        if not self.is_empty():
            return self.items.pop()
        raise IndexError("pop from empty stack")
    
    def peek(self):
        """Return top item without removing"""
        if not self.is_empty():
            return self.items[-1]
        raise IndexError("peek from empty stack")
    
    def is_empty(self):
        """Check if stack is empty"""
        return len(self.items) == 0
    
    def size(self):
        """Return stack size"""
        return len(self.items)
    
    def __str__(self):
        return f"Stack({self.items})"

# Test the stack
stack = Stack()
stack.push(1)
stack.push(2)
stack.push(3)

print(f"Stack: {stack}")
print(f"Top element: {stack.peek()}")
print(f"Popped: {stack.pop()}")
print(f"Stack after pop: {stack}")
print(f"Is empty: {stack.is_empty()}")
print(f"Size: {stack.size()}")
````

### Common Operations

| Operation | Description | Time Complexity |
|-----------|-------------|------------------|
| `push(item)` | Add item to top | O(1) |
| `pop()` | Remove and return top | O(1) |
| `peek()` | View top without removing | O(1) |
| `is_empty()` | Check if empty | O(1) |
| `size()` | Get number of elements | O(1) |

### Applications

#### 1. Function Call Stack

When you call a function, Python uses a stack to keep track of function calls and local variables.

````python
def function_call_stack_demo():
    """Demonstrates function call stack behavior"""
    def func_a():
        print("Entering func_a")
        func_b()
        print("Exiting func_a")
    
    def func_b():
        print("Entering func_b")
        func_c()
        print("Exiting func_b")
    
    def func_c():
        print("Entering func_c")
        print("Base case reached")
        print("Exiting func_c")
    
    print("Starting function call stack demo")
    func_a()
    print("Function call stack demo complete")

function_call_stack_demo()
````

#### 2. Valid Parentheses Checking

Stacks are perfect for checking balanced parentheses, brackets, and braces.

````python
def is_valid_parentheses(s: str) -> bool:
    """Check if string has valid parentheses using stack"""
    stack = []
    mapping = {')': '(', '}': '{', ']': '['}
    
    for char in s:
        if char in mapping.values():  # Opening bracket
            stack.append(char)
        elif char in mapping:  # Closing bracket
            if not stack or stack[-1] != mapping[char]:
                return False
            stack.pop()
    
    return len(stack) == 0

# Test cases
test_cases = [
    ("()", True),
    "()[]{}",
    "(]",
    "([{}])",
    "([)]"
]

for test in test_cases:
    if isinstance(test, tuple):
        s, expected = test
        result = is_valid_parentheses(s)
        print(f"{s!r:10} -> {result} (expected {expected})")
    else:
        result = is_valid_parentheses(test)
        print(f"{test!r:10} -> {result}")
````

#### 3. Reverse a String

Stack naturally reverses the order of elements.

````python
def reverse_string(s: str) -> str:
    """Reverse string using stack"""
    stack = []
    
    # Push all characters
    for char in s:
        stack.append(char)
    
    # Pop all characters to reverse
    reversed_str = ""
    while stack:
        reversed_str += stack.pop()
    
    return reversed_str

print(f"Original: 'hello'")
print(f"Reversed: '{reverse_string('hello')}'")
print(f"Original: 'Python'")
print(f"Reversed: '{reverse_string('Python')}'")
````

## Queues

### Definition & Properties

A **Queue** is a linear data structure that follows the **FIFO** (First In, First Out) principle. Think of it like a line at a grocery store - the first person to join the line is the first to be served.

**Key Properties:**
- Elements are added at the rear (enqueue)
- Elements are removed from the front (dequeue)
- Order of elements is preserved

### Visual Representation

```
Front -> [1][2][3][4][5] <- Rear

enqueue(6) -> [1][2][3][4][5][6]
dequeue()  -> removes 1
           [2][3][4][5][6]
```

### Implementation

````python
from collections import deque

class Queue:
    """Queue implementation using collections.deque"""
    
    def __init__(self):
        self.items = deque()
    
    def enqueue(self, item):
        """Add item to rear of queue"""
        self.items.append(item)
    
    def dequeue(self):
        """Remove and return front item"""
        if not self.is_empty():
            return self.items.popleft()
        raise IndexError("dequeue from empty queue")
    
    def front(self):
        """Return front item without removing"""
        if not self.is_empty():
            return self.items[0]
        raise IndexError("front from empty queue")
    
    def rear(self):
        """Return rear item without removing"""
        if not self.is_empty():
            return self.items[-1]
        raise IndexError("rear from empty queue")
    
    def is_empty(self):
        """Check if queue is empty"""
        return len(self.items) == 0
    
    def size(self):
        """Return queue size"""
        return len(self.items)
    
    def __str__(self):
        return f"Queue({list(self.items)})"

# Test the queue
queue = Queue()
queue.enqueue(1)
queue.enqueue(2)
queue.enqueue(3)

print(f"Queue: {queue}")
print(f"Front element: {queue.front()}")
print(f"Rear element: {queue.rear()}")
print(f"Dequeued: {queue.dequeue()}")
print(f"Queue after dequeue: {queue}")
print(f"Is empty: {queue.is_empty()}")
print(f"Size: {queue.size()}")
````

### Common Operations

| Operation | Description | Time Complexity |
|-----------|-------------|------------------|
| `enqueue(item)` | Add item to rear | O(1) |
| `dequeue()` | Remove and return front | O(1) |
| `front()` | View front without removing | O(1) |
| `rear()` | View rear without removing | O(1) |
| `is_empty()` | Check if empty | O(1) |
| `size()` | Get number of elements | O(1) |

### Applications

#### 1. Task Scheduling

Queues are perfect for managing tasks in the order they arrive.

````python
import time
import random

class TaskScheduler:
    """Simple task scheduler using queue"""
    
    def __init__(self):
        self.queue = Queue()
    
    def add_task(self, task_name, duration):
        """Add a task to the scheduler"""
        self.queue.enqueue((task_name, duration))
        print(f"Task '{task_name}' added to queue")
    
    def run_tasks(self):
        """Execute all tasks in queue order"""
        print("\nStarting task execution...")
        
        while not self.queue.is_empty():
            task_name, duration = self.queue.dequeue()
            print(f"Executing task: {task_name} ({duration}s)")
            time.sleep(duration)  # Simulate task execution
            print(f"Task '{task_name}' completed")
        
        print("All tasks completed!")

# Demo
scheduler = TaskScheduler()
scheduler.add_task("Download file", 0.5)
scheduler.add_task("Process data", 0.3)
scheduler.add_task("Send email", 0.2)
scheduler.add_task("Update database", 0.4)

scheduler.run_tasks()
````

#### 2. Level Order Tree Traversal

Queue enables breadth-first traversal of tree structures.

````python
class TreeNode:
    """Binary tree node"""
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def level_order_traversal(root):
    """Breadth-first traversal using queue"""
    if not root:
        return []
    
    result = []
    queue = Queue()
    queue.enqueue(root)
    
    while not queue.is_empty():
        level_size = queue.size()
        current_level = []
        
        for _ in range(level_size):
            node = queue.dequeue()
            current_level.append(node.val)
            
            if node.left:
                queue.enqueue(node.left)
            if node.right:
                queue.enqueue(node.right)
        
        result.append(current_level)
    
    return result

# Build a sample tree
#       1
#      / \
#     2   3
#    / \
#   4   5
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

print("Level order traversal:")
for level in level_order_traversal(root):
    print(level)
````

## Advanced Variants

### 1. Deque (Double-Ended Queue)

A deque allows insertion and removal from both ends.

````python
from collections import deque

# Deque operations
dq = deque([1, 2, 3])

print(f"Initial deque: {dq}")

# Add to right (like queue enqueue)
dq.append(4)
print(f"After append(4): {dq}")

# Add to left
dq.appendleft(0)
print(f"After appendleft(0): {dq}")

# Remove from right (like stack pop)
right = dq.pop()
print(f"After pop(): {dq}, removed: {right}")

# Remove from left (like queue dequeue)
left = dq.popleft()
print(f"After popleft(): {dq}, removed: {left}")
````

### 2. Priority Queue

Elements are removed based on priority, not just order.

````python
import heapq

class PriorityQueue:
    """Priority queue using heapq"""
    
    def __init__(self):
        self.items = []
    
    def enqueue(self, item, priority):
        """Add item with priority"""
        heapq.heappush(self.items, (priority, item))
    
    def dequeue(self):
        """Remove item with highest priority (lowest number)"""
        if not self.is_empty():
            return heapq.heappop(self.items)[1]
        raise IndexError("dequeue from empty priority queue")
    
    def is_empty(self):
        return len(self.items) == 0
    
    def __str__(self):
        return f"PriorityQueue({self.items})"

# Demo priority queue
pq = PriorityQueue()
pq.enqueue("Low priority task", 3)
pq.enqueue("High priority task", 1)
pq.enqueue("Medium priority task", 2)

print(f"Priority queue: {pq}")
print("Processing tasks by priority:")

while not pq.is_empty():
    task = pq.dequeue()
    print(f"Processing: {task}")
````

### 3. Circular Queue

Efficient use of fixed-size array with wrap-around.

````python
class CircularQueue:
    """Circular queue implementation"""
    
    def __init__(self, capacity):
        self.capacity = capacity
        self.items = [None] * capacity
        self.front = -1
        self.rear = -1
        self.size = 0
    
    def enqueue(self, item):
        """Add item to rear (with wrap-around)"""
        if self.is_full():
            raise OverflowError("Queue is full")
        
        if self.front == -1:  # First element
            self.front = 0
        
        self.rear = (self.rear + 1) % self.capacity
        self.items[self.rear] = item
        self.size += 1
    
    def dequeue(self):
        """Remove item from front (with wrap-around)"""
        if self.is_empty():
            raise IndexError("Queue is empty")
        
        item = self.items[self.front]
        
        if self.front == self.rear:  # Last element
            self.front = -1
            self.rear = -1
        else:
            self.front = (self.front + 1) % self.capacity
        
        self.size -= 1
        return item
    
    def is_empty(self):
        return self.size == 0
    
    def is_full(self):
        return self.size == self.capacity
    
    def __str__(self):
        if self.is_empty():
            return "CircularQueue([])"
        
        result = []
        for i in range(self.size):
            index = (self.front + i) % self.capacity
            result.append(str(self.items[index]))
        
        return f"CircularQueue([{', '.join(result)}])"

# Demo circular queue
cq = CircularQueue(5)
print(f"Initial: {cq}")

for i in range(5):
    cq.enqueue(i)
    print(f"After enqueue({i}): {cq}")

print(f"Is full: {cq.is_full()}")

for i in range(3):
    item = cq.dequeue()
    print(f"After dequeue(): {cq}, removed: {item}")

# Can enqueue more due to wrap-around
for i in range(5, 8):
    cq.enqueue(i)
    print(f"After enqueue({i}): {cq}")
````

## Problem-Solving Patterns

### 1. Monotonic Stack

Stack where elements are always in sorted order. Useful for finding next greater/smaller elements.

````python
def next_greater_element(nums):
    """Find next greater element for each position"""
    result = [-1] * len(nums)
    stack = []  # Store indices
    
    for i in range(len(nums)):
        # While current element is greater than stack top
        while stack and nums[i] > nums[stack[-1]]:
            index = stack.pop()
            result[index] = nums[i]
        
        stack.append(i)
    
    return result

# Test
arr = [2, 1, 2, 4, 3]
print(f"Array: {arr}")
print(f"Next greater: {next_greater_element(arr)}")

# Explanation:
# 2 -> 4
# 1 -> 2
# 2 -> 4
# 4 -> -1 (no greater element)
# 3 -> -1 (no greater element)
````

### 2. Sliding Window Maximum

Use deque to maintain maximum in sliding window efficiently.

````python
def sliding_window_maximum(nums, k):
    """Find maximum in each sliding window of size k"""
    if not nums or k == 0:
        return []
    
    result = []
    window = deque()  # Store indices in decreasing order
    
    for i in range(len(nums)):
        # Remove indices out of window
        while window and window[0] <= i - k:
            window.popleft()
        
        # Remove indices whose values are less than current
        while window and nums[window[-1]] < nums[i]:
            window.pop()
        
        window.append(i)
        
        # Add to result when window is full
        if i >= k - 1:
            result.append(nums[window[0]])
    
    return result

# Test
arr = [1, 3, -1, -3, 5, 3, 6, 7]
k = 3
print(f"Array: {arr}")
print(f"Window size: {k}")
print(f"Sliding window maximum: {sliding_window_maximum(arr, k)}")
# Expected: [3, 3, 5, 5, 6, 7]
````

### 3. Stack for Expression Evaluation

Evaluate mathematical expressions using stacks.

````python
def evaluate_expression(expression):
    """Evaluate mathematical expression with +, -, *, /"""
    def precedence(op):
        if op == '+' or op == '-':
            return 1
        if op == '*' or op == '/':
            return 2
        return 0
    
    def apply_op(a, b, op):
        if op == '+':
            return a + b
        if op == '-':
            return a - b
        if op == '*':
            return a * b
        if op == '/':
            return a // b
    
    values = []  # Stack for numbers
    ops = []     # Stack for operators
    
    i = 0
    while i < len(expression):
        if expression[i] == ' ':
            i += 1
            continue
        
        # If number, extract full number
        if expression[i].isdigit():
            val = 0
            while i < len(expression) and expression[i].isdigit():
                val = val * 10 + int(expression[i])
                i += 1
            values.append(val)
        
        # If opening bracket
        elif expression[i] == '(':
            ops.append(expression[i])
            i += 1
        
        # If closing bracket
        elif expression[i] == ')':
            while ops and ops[-1] != '(':
                val2 = values.pop()
                val1 = values.pop()
                op = ops.pop()
                values.append(apply_op(val1, val2, op))
            
            if ops:
                ops.pop()  # Remove opening bracket
            i += 1
        
        # If operator
        else:
            while ops and precedence(ops[-1]) >= precedence(expression[i]):
                val2 = values.pop()
                val1 = values.pop()
                op = ops.pop()
                values.append(apply_op(val1, val2, op))
            
            ops.append(expression[i])
            i += 1
    
    # Apply remaining operators
    while ops:
        val2 = values.pop()
        val1 = values.pop()
        op = ops.pop()
        values.append(apply_op(val1, val2, op))
    
    return values[-1]

# Test
expressions = [
    "3 + 4 * 2",
    "10 + 2 * 6",
    "100 * 2 + 12",
    "100 * ( 2 + 12 )",
    "100 * ( 2 + 12 ) / 14"
]

for expr in expressions:
    result = evaluate_expression(expr)
    print(f"{expr} = {result}")
````

## Performance Analysis

### Time Complexity Comparison

| Operation | Stack | Queue | Deque | Priority Queue |
|-----------|-------|-------|-------|----------------|
| Insert | O(1) | O(1) | O(1) | O(log n) |
| Delete | O(1) | O(1) | O(1) | O(log n) |
| Peek | O(1) | O(1) | O(1) | O(1) |
| Search | O(n) | O(n) | O(n) | O(n) |

### Space Complexity

- All basic implementations: O(n) where n is number of elements
- Circular queue: O(k) where k is fixed capacity
- Priority queue: O(n) for heap implementation

### When to Use Which

| Use Case | Best Data Structure | Reason |
|----------|-------------------|--------|
| Function calls | Stack | Natural LIFO behavior |
| Task scheduling | Queue | Fair FIFO ordering |
| Both ends access | Deque | Flexible operations |
| Priority-based | Priority Queue | Order by importance |
| Fixed buffer | Circular Queue | Memory efficiency |

## Best Practices

### 1. Choose the Right Implementation

- **Python list for stack**: Simple, but beware of O(n) operations at front
- **collections.deque for queue**: O(1) operations at both ends
- **heapq for priority queue**: Efficient priority operations

### 2. Handle Edge Cases

```python
# Always check before operations
if not stack.is_empty():
    item = stack.pop()

if not queue.is_empty():
    item = queue.dequeue()
```

### 3. Memory Management

- For large datasets, consider memory-efficient implementations
- Use generators when possible for lazy evaluation
- Clear structures when no longer needed

### 4. Common Pitfalls

1. **Using wrong data structure**: Stack for FIFO needs, Queue for LIFO needs
2. **Ignoring overflow/underflow**: Always check bounds
3. **Inefficient operations**: Using list.pop(0) for queue (O(n))
4. **Memory leaks**: Not clearing references in long-running programs
