# Recursion

A function that calls itself to solve smaller subproblems.

---

## Anatomy of Recursion

Every recursive function needs:
1. **Base case** - when to stop
2. **Recursive case** - break problem into smaller subproblems

````python
def factorial(n):
    """Classic example: n! = n * (n-1)!"""
    # Base case
    if n <= 1:
        return 1
    # Recursive case
    return n * factorial(n - 1)

print(f"5! = {factorial(5)}")
print(f"10! = {factorial(10)}")
````

````python
def fibonacci(n):
    """Fibonacci: F(n) = F(n-1) + F(n-2)"""
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(f"Fib(10) = {fibonacci(10)}")
# Note: This is O(2^n) - very slow!
````

## Recursion with Memoization

Cache results to avoid redundant computation.

````python
from functools import lru_cache

@lru_cache(maxsize=None)
def fibonacci_memo(n):
    """Memoized Fibonacci: O(n)"""
    if n <= 1:
        return n
    return fibonacci_memo(n - 1) + fibonacci_memo(n - 2)

print(f"Fib(50) = {fibonacci_memo(50)}")
print(f"Fib(100) = {fibonacci_memo(100)}")
````

## Common Recursive Patterns

### 1. Linear Recursion

````python
def sum_list(arr):
    """Sum elements recursively."""
    if not arr:
        return 0
    return arr[0] + sum_list(arr[1:])

print(f"Sum [1,2,3,4,5] = {sum_list([1,2,3,4,5])}")
````

### 2. Binary Recursion (Divide and Conquer)

````python
def binary_sum(arr, left, right):
    """Sum using divide and conquer."""
    if left > right:
        return 0
    if left == right:
        return arr[left]
    
    mid = (left + right) // 2
    return binary_sum(arr, left, mid) + binary_sum(arr, mid + 1, right)

arr = [1, 2, 3, 4, 5]
print(f"Binary sum = {binary_sum(arr, 0, len(arr) - 1)}")
````

### 3. Backtracking

````python
def permutations(nums):
    """Generate all permutations using backtracking."""
    result = []
    
    def backtrack(path, remaining):
        if not remaining:
            result.append(path[:])
            return
        
        for i in range(len(remaining)):
            path.append(remaining[i])
            backtrack(path, remaining[:i] + remaining[i+1:])
            path.pop()  # Backtrack
    
    backtrack([], nums)
    return result

print(f"Permutations of [1,2,3]:")
for p in permutations([1, 2, 3]):
    print(f"  {p}")
````

````python
def subsets(nums):
    """Generate all subsets."""
    result = []
    
    def backtrack(start, path):
        result.append(path[:])
        
        for i in range(start, len(nums)):
            path.append(nums[i])
            backtrack(i + 1, path)
            path.pop()
    
    backtrack(0, [])
    return result

print(f"Subsets of [1,2,3]: {subsets([1, 2, 3])}")
````

## Reverse Linked List Using Recursion

One of the most elegant applications of recursion is reversing a linked list. This problem demonstrates how recursion can simplify complex pointer manipulation.

### The Problem

Given a singly linked list, reverse it so that the last node becomes the first, and all pointers are reversed.

**Example**:
```
Input:  1 -> 2 -> 3 -> 4 -> 5 -> None
Output: 5 -> 4 -> 3 -> 2 -> 1 -> None
```

````python
# Define the ListNode class
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    
    def __repr__(self):
        """String representation for easy visualization."""
        result = []
        node = self
        while node:
            result.append(str(node.val))
            node = node.next
        return " -> ".join(result) + " -> None"

# Helper function to create a linked list from a list
def create_linked_list(values):
    """Create a linked list from a list of values."""
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for val in values[1:]:
        current.next = ListNode(val)
        current = current.next
    return head

# Test the helper
original = create_linked_list([1, 2, 3, 4, 5])
print(f"Original list: {original}")
````

### The Recursive Approach

**Key Insight**: To reverse a linked list recursively, we:
1. Recursively reverse the rest of the list (everything after the head)
2. Make the second node point back to the head
3. Set the head's next to None (it becomes the tail)
4. Return the new head (which was the last node)

**Base Case**: If the list is empty or has only one node, it's already reversed.

**Recursive Case**: 
- Recursively reverse the rest of the list
- Fix the pointers for the current node

````python
def reverse_linked_list_recursive(head):
    """
    Reverse a linked list using recursion.
    
    Time Complexity: O(n) - visit each node once
    Space Complexity: O(n) - recursion stack depth
    """
    # Base case: empty list or single node
    if not head or not head.next:
        return head
    
    # Recursive case: reverse the rest of the list
    new_head = reverse_linked_list_recursive(head.next)
    
    # Fix the pointers
    # head.next is the second node, make it point back to head
    head.next.next = head
    
    # Current head becomes the tail, so its next should be None
    head.next = None
    
    # Return the new head (which was the last node)
    return new_head

# Test the function
original = create_linked_list([1, 2, 3, 4, 5])
print(f"Original: {original}")

reversed_list = reverse_linked_list_recursive(original)
print(f"Reversed: {reversed_list}")
````

### Step-by-Step Detailed Walkthrough

Let's trace through the recursion for the list `[1, 2, 3, 4, 5]` to understand exactly what happens at each step.

#### Visual Representation of the Process

```
Original List:
1 -> 2 -> 3 -> 4 -> 5 -> None

Call Stack (going down):
reverse(1 -> 2 -> 3 -> 4 -> 5)
  └─> reverse(2 -> 3 -> 4 -> 5)
        └─> reverse(3 -> 4 -> 5)
              └─> reverse(4 -> 5)
                    └─> reverse(5)  ← Base case! Return 5
```

#### Detailed Step-by-Step Execution

**Step 1: Initial Call**
```
reverse(1 -> 2 -> 3 -> 4 -> 5)
  head = 1
  head.next = 2
  Not base case, so recurse...
```

**Step 2: Recurse on rest**
```
reverse(2 -> 3 -> 4 -> 5)
  head = 2
  head.next = 3
  Not base case, so recurse...
```

**Step 3: Continue recursing**
```
reverse(3 -> 4 -> 5)
  head = 3
  head.next = 4
  Not base case, so recurse...
```

**Step 4: Almost at base case**
```
reverse(4 -> 5)
  head = 4
  head.next = 5
  Not base case, so recurse...
```

**Step 5: Base case reached!**
```
reverse(5)
  head = 5
  head.next = None
  BASE CASE! Return 5
  
  Now 5 is the new head of the reversed list
```

**Step 6: Unwinding - Fix pointers for node 4**
```
Back in reverse(4 -> 5):
  new_head = 5 (returned from recursive call)
  head = 4
  head.next = 5
  
  Fix pointers:
  head.next.next = head  →  5.next = 4
  head.next = None       →  4.next = None
  
  List now: 5 -> 4 -> None
  Return new_head (5)
```

**Step 7: Unwinding - Fix pointers for node 3**
```
Back in reverse(3 -> 4 -> 5):
  new_head = 5 (still the head)
  head = 3
  head.next = 4
  
  Fix pointers:
  head.next.next = head  →  4.next = 3
  head.next = None       →  3.next = None
  
  List now: 5 -> 4 -> 3 -> None
  Return new_head (5)
```

**Step 8: Unwinding - Fix pointers for node 2**
```
Back in reverse(2 -> 3 -> 4 -> 5):
  new_head = 5 (still the head)
  head = 2
  head.next = 3
  
  Fix pointers:
  head.next.next = head  →  3.next = 2
  head.next = None       →  2.next = None
  
  List now: 5 -> 4 -> 3 -> 2 -> None
  Return new_head (5)
```

**Step 9: Unwinding - Fix pointers for node 1**
```
Back in reverse(1 -> 2 -> 3 -> 4 -> 5):
  new_head = 5 (still the head)
  head = 1
  head.next = 2
  
  Fix pointers:
  head.next.next = head  →  2.next = 1
  head.next = None       →  1.next = None
  
  Final list: 5 -> 4 -> 3 -> 2 -> 1 -> None
  Return new_head (5)
```

#### Key Observations

1. **Recursion goes all the way to the last node** before doing any work
2. **The last node becomes the new head** and is passed back up through all recursive calls
3. **Each level fixes its own pointers** as the stack unwinds
4. **The pattern `head.next.next = head`** makes the next node point back to current
5. **Setting `head.next = None`** prevents cycles and makes current node the tail

````python
def reverse_linked_list_verbose(head):
    """
    Verbose version showing each step of the recursion.
    Helps visualize what's happening at each recursive call.
    """
    print(f"\n{'='*60}")
    print("Starting recursive reversal with verbose output")
    print(f"{'='*60}")
    
    call_depth = [0]  # Use list to modify in nested function
    
    def reverse_helper(node, depth):
        indent = "  " * depth
        call_depth[0] = max(call_depth[0], depth)
        
        print(f"\n{indent}📞 Call {depth}: reverse({node.val if node else 'None'})")
        
        # Base case
        if not node or not node.next:
            if node:
                print(f"{indent}✅ BASE CASE: Single node {node.val}, returning it as new head")
            else:
                print(f"{indent}✅ BASE CASE: Empty list")
            return node
        
        print(f"{indent}   Current: {node.val} -> {node.next.val}")
        print(f"{indent}   ⬇️  Recursing on rest of list...")
        
        # Recursive call
        new_head = reverse_helper(node.next, depth + 1)
        
        # Fix pointers (unwinding)
        print(f"\n{indent}⬆️  Unwinding from call {depth}")
        print(f"{indent}   Fixing pointers for node {node.val}")
        print(f"{indent}   Before: {node.val}.next = {node.next.val if node.next else 'None'}")
        
        node.next.next = node
        print(f"{indent}   Action: {node.next.val}.next = {node.val}")
        
        node.next = None
        print(f"{indent}   Action: {node.val}.next = None")
        print(f"{indent}   After: {node.val} is now a tail node")
        print(f"{indent}   Returning new_head = {new_head.val}")
        
        return new_head
    
    result = reverse_helper(head, 0)
    
    print(f"\n{'='*60}")
    print(f"Recursion complete! Max depth: {call_depth[0]}")
    print(f"{'='*60}\n")
    
    return result

# Test with verbose output
print("Testing with list [1, 2, 3, 4, 5]:")
original = create_linked_list([1, 2, 3, 4, 5])
reversed_list = reverse_linked_list_verbose(original)
print(f"\nFinal result: {reversed_list}")
````

### Common Mistakes and How to Avoid Them

#### Mistake 1: Forgetting to Set `head.next = None`

**Wrong**:
```python
def reverse_wrong(head):
    if not head or not head.next:
        return head
    new_head = reverse_wrong(head.next)
    head.next.next = head
    # Forgot: head.next = None
    return new_head
```

**Problem**: This creates a cycle! The original head still points to the second node, which now points back to it.

**Result**: `1 <-> 2 <- 3 <- 4 <- 5` (infinite loop between 1 and 2)

**Correct**:
```python
head.next.next = head
head.next = None  # ✅ Must set to None to break the cycle
```

#### Mistake 2: Returning the Wrong Node

**Wrong**:
```python
def reverse_wrong(head):
    if not head or not head.next:
        return head
    new_head = reverse_wrong(head.next)
    head.next.next = head
    head.next = None
    return head  # ❌ Wrong! Returns current node, not new head
```

**Problem**: Returns the original head (which is now the tail), not the new head.

**Correct**:
```python
return new_head  # ✅ Return the new head from recursive call
```

#### Mistake 3: Incorrect Base Case

**Wrong**:
```python
def reverse_wrong(head):
    if not head:  # ❌ Missing check for head.next
        return head
    # ... rest of code
```

**Problem**: When `head.next` is None, we still try to access `head.next.next`, causing an error.

**Correct**:
```python
if not head or not head.next:  # ✅ Check both conditions
    return head
```

### Comparison: Recursive vs Iterative Approach

Both approaches have their merits. Let's compare them:

#### Recursive Approach
```python
def reverse_recursive(head):
    if not head or not head.next:
        return head
    new_head = reverse_recursive(head.next)
    head.next.next = head
    head.next = None
    return new_head
```

**Pros**:
- Clean and elegant code
- Easy to understand the logic
- Natural way to think about the problem

**Cons**:
- O(n) space due to call stack
- Risk of stack overflow for very long lists
- Slower due to function call overhead

#### Iterative Approach
```python
def reverse_iterative(head):
    prev = None
    current = head
    while current:
        next_node = current.next
        current.next = prev
        prev = current
        current = next_node
    return prev
```

**Pros**:
- O(1) space complexity
- No stack overflow risk
- Generally faster

**Cons**:
- More pointer manipulation
- Can be harder to understand initially

| Aspect | Recursive | Iterative |
|--------|-----------|-----------|
| **Code Length** | Shorter | Slightly longer |
| **Readability** | More elegant | More explicit |
| **Time Complexity** | O(n) | O(n) |
| **Space Complexity** | O(n) stack | O(1) |
| **Stack Overflow Risk** | Yes (for long lists) | No |
| **Performance** | Slower (function calls) | Faster |

````python
# Compare both implementations
def reverse_iterative(head):
    """Iterative approach for comparison."""
    prev = None
    current = head
    
    while current:
        next_node = current.next  # Save next
        current.next = prev       # Reverse pointer
        prev = current            # Move prev forward
        current = next_node       # Move current forward
    
    return prev

# Test both approaches
print("Comparing Recursive vs Iterative:\n")

# Test 1
original1 = create_linked_list([1, 2, 3, 4, 5])
original2 = create_linked_list([1, 2, 3, 4, 5])

print("Original list: 1 -> 2 -> 3 -> 4 -> 5")
print(f"Recursive result: {reverse_linked_list_recursive(original1)}")
print(f"Iterative result: {reverse_iterative(original2)}")
print("\nBoth produce the same result! ✅")
````

### Key Takeaways for Recursive Linked List Reversal

| Concept | Key Point |
|---------|-----------|
| **Base Case** | Empty list or single node (already reversed) |
| **Recursive Step** | Reverse the rest, then fix current node's pointers |
| **Critical Line** | `head.next.next = head` makes next node point back |
| **Prevent Cycles** | Must set `head.next = None` after reversing |
| **Return Value** | Always return the new head (from base case) |
| **Space Complexity** | O(n) due to recursion stack |
| **When to Use** | When code elegance matters more than space efficiency |

### The Mental Model

Think of it like a stack of plates:
1. **Going down**: Place each node on a stack (recursion calls)
2. **Hit bottom**: The last node becomes the new head
3. **Coming back up**: Pop each plate and reverse its pointer
4. **Each level**: Fixes its own connection, passes the new head up

The recursion naturally handles the "visit in reverse order" pattern that's needed for reversing a linked list!

## Recursion vs Iteration

| Aspect | Recursion | Iteration |
|--------|-----------|----------|
| Readability | Often cleaner | Can be verbose |
| Space | O(n) stack | O(1) possible |
| Speed | Function call overhead | Generally faster |
| Stack overflow | Risk with deep recursion | No risk |

````python
# Iterative factorial (more efficient)
def factorial_iter(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

print(f"Iterative 10! = {factorial_iter(10)}")
````
